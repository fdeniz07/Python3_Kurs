# Python 3: Operationen für Dictionaries

Dictionaries (Wörterbücher) sind veränderbare Datenstrukturen, die Schlüssel-Wert-Paare (*Key-Value Pairs*) speichern. Da sie auf **Hashing** basieren, haben die Elemente keine feste Reihenfolge. Damit das Hashing bei jedem Zugriff die gleiche Adresse liefert, darf der Schlüssel sich nicht ändern.

---

## 1. Übersicht der wichtigsten Operationen

| Operation | Ergebnis / Beschreibung |
| :--- | :--- |
| `d[k]` | Liefert den Wert zum Schlüssel `k`. Existiert der Schlüssel nicht, gibt es einen **KeyError**. |
| `d[k] = x` | Dem Schlüssel `k` wird der Wert `x` zugeordnet (erstellt den Eintrag neu oder überschreibt ihn). |
| `del d[k]` | Das Item mit dem Schlüssel `k` wird entfernt. Falls `k` nicht existiert, gibt es einen **KeyError**. |
| `d.keys()` | Die Methode liefert ein `dict_keys`-Objekt, das die Menge der Schlüssel von `d` repräsentiert. |
| `d.values()` | Die Methode liefert ein `dict_values`-Objekt, das eine Kollektion der Werte repräsentiert. |

> **Beachte:** Die Methoden `keys()` und `values()` liefern spezielle Objekte, aus denen man leicht eine Menge (Typ `set`) oder Liste (Typ `list`) erzeugen kann.

## 2. Beispiel: Schlüssel und Werte auslesen
Zuerst erstellen wir ein Dictionary `preis` und lassen uns die Schlüssel sowie die Werte separat als Liste ausgeben.

```python
# Dictionary initialisieren
preis = {'Papier': 4.45, 'Radiergummi': 0.45}

# Schlüssel als Liste ausgeben
schluessel_liste = list(preis.keys())
print("Schlüssel:", schluessel_liste)

# Werte als Liste ausgeben
werte_liste = list(preis.values())
print("Werte:", werte_liste)
```

## 3. Ein Dictionary ändern (Element hinzufügen)
Dictionaries sind änderbare Objekte. Man kann Items (Schlüssel-Wert-Paare) im Dictionary *hinzufügen*, *löschen* und *ändern*.

```python
# Neues Element hinzufügen
preis['Laminierfolien'] = 2.50

# Das aktualisierte Dictionary anzeigen
print(preis)
```
