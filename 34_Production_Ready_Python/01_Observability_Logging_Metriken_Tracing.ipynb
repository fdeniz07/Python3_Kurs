## Lernziele

- Tracing-Grundlagen fuer verteilte Systeme einordnen

```python
import logging

logging.basicConfig(level=logging.INFO, format='%(asctime)s %(levelname)s %(message)s')
logger = logging.getLogger('service')
logger.info('service-started')
```

```python
# Einfaches Metrikmodell (Counter)
metrics = {'requests_total': 0, 'errors_total': 0}

def handle_request(ok=True):
    metrics['requests_total'] += 1
    if not ok:
        metrics['errors_total'] += 1

handle_request(ok=True)
handle_request(ok=False)
print(metrics)
```

## Best Practices

- Log-Nachrichten konsistent und maschinenlesbar formatieren
- Metriken nach RED- oder USE-Methode strukturieren
- Correlation IDs fuer Request-Ketten einfuehren

## Uebung

Erweitere das Metrikmodell um `latency_ms` als Liste und berechne Durchschnitt, Minimum und Maximum.

```python
latency_ms = [120, 80, 95, 140]
avg = sum(latency_ms) / len(latency_ms)
print({'avg': avg, 'min': min(latency_ms), 'max': max(latency_ms)})
```
