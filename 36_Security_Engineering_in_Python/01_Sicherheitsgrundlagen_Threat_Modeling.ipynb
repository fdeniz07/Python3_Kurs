## Theorie

Security Engineering bedeutet, Sicherheit von Anfang an als Architekturziel zu behandeln. Threat Modeling hilft, Angriffe vor der Implementierung sichtbar zu machen.

```python
assets = ['user_data', 'auth_tokens', 'payment_records']
threats = {'spoofing': 8, 'tampering': 7, 'dos': 6}
print('Assets:', assets)
print('Top threat:', max(threats, key=threats.get))
```

## Zentrale Fachbegriffe

STRIDE
Attack Surface
Trust Boundary
Likelihood
Impact
Risk Matrix

```python
# Risiko-Matrix (einfaches Beispiel)
items=[{"threat":"spoofing","likelihood":3,"impact":4},{"threat":"tampering","likelihood":2,"impact":5}]
for i in items:
    i["risk"] = i["likelihood"] * i["impact"]
print(sorted(items,key=lambda x:x["risk"],reverse=True))
```
