## Projekt: Digitale Uhr

### Ziel

Aktuelle Uhrzeit anzeigen.

### Verwendete Widgets

- Label
- Button
- Entry
- Text oder Listbox (je nach Projekt)

### Beispiel


```python
import tkinter as tk
import time

root=tk.Tk()

label=tk.Label(root,font=("Arial",20))
label.pack(pady=20)

def aktualisieren():
    label.config(text=time.strftime("%H:%M:%S"))
    root.after(1000,aktualisieren)

aktualisieren()

root.mainloop()
```

### Erweiterung

- Design verbessern
- Fehlerbehandlung ergänzen
- Weitere Funktionen hinzufügen

## Projekt: Würfel

### Ziel

Zufallszahl von 1 bis 6.

### Verwendete Widgets

- Label
- Button
- Entry
- Text oder Listbox (je nach Projekt)

### Beispiel


```python
import tkinter as tk
import random

root=tk.Tk()

ausgabe=tk.Label(root,font=("Arial",20))
ausgabe.pack()

def werfen():
    ausgabe.config(text=random.randint(1,6))

tk.Button(root,text="Würfeln",command=werfen).pack()

root.mainloop()
```

## Projekt: Zahlenraten

### Ziel

Eine Zufallszahl erraten.

### Verwendete Widgets

- Label
- Button
- Entry
- Text oder Listbox (je nach Projekt)

### Beispiel


```python
import tkinter as tk
import random

zahl=random.randint(1,10)

root=tk.Tk()

def pruefen():
    if int(entry.get())==zahl:
        ausgabe.config(text="Richtig!")
    else:
        ausgabe.config(text="Leider falsch.")

entry=tk.Entry(root)
entry.pack()

tk.Button(root,text="Prüfen",command=pruefen).pack()

ausgabe=tk.Label(root)
ausgabe.pack()

root.mainloop()
```
