## Bedingte Anweisungen – Hausaufgaben

### Aufgabe 1: Anzahl der Stellen berechnen

Aufgabenbeschreibung:

Berechnen Sie die Anzahl der Stellen einer zufällig erzeugten Zahl mit maximal 5 Stellen.


### Aufgabe 2: Einkaufs- und Preisberechnung

Aufgabenbeschreibung:

Stellen Sie sich vor, Sie entwickeln ein Programm zur Preisberechnung. Der Preis soll abhängig davon angepasst werden, ob der Benutzer eine Ratenzahlung wünscht und ob eine Kundenkarte vorhanden ist.

Bedingung 1 -> Bei Ratenzahlung erhöht sich der Preis:
- 3 Raten: +3 %
- 6 Raten: +6 %
- 9 Raten: +9 %

Bedingung 2 -> Wenn eine Kundenkarte vorhanden ist, wird ein Rabatt von 10 % gewährt.


### Aufgabe 3: Prüfung der Dreiecksbedingung

Aufgabenbeschreibung:

Lesen Sie die drei Seitenlängen eines Dreiecks ein und prüfen Sie, ob mit diesen Werten (x, y, z) ein Dreieck gebildet werden kann.

Bedingung 1 -> Für die Seite x gilt: |y-z| < x < y+z

Bedingung 2 -> Für die Seite y gilt: |x-z| < y < x+z

Bedingung 3 -> Für die Seite z gilt: |x-y| < z < x+y
