## Schleifen
Schleifen sind Kontrollstrukturen, die einen Codeblock wiederholt ausführen, solange eine bestimmte Bedingung erfüllt ist.

1. While-Schleife
2. For-Schleife

#### While-Schleife

### Wann verwendet man sie?

Die While-Schleife wird verwendet, wenn Wiederholungen von einer Bedingung abhängen oder solange ausgeführt werden sollen, bis eine bestimmte Bedingung nicht mehr erfüllt ist.

### Syntax:

while bedingung:

    anweisungen_im_schleifenblock

### Erklärung:
Solange die Bedingung den Wert True liefert, werden die Anweisungen im Schleifenblock wiederholt ausgeführt.

```python
i=0
print("Die Schleife beginnt...")
while i<10:
    print(i,end=" ")
    i+=1
print("\nDie Schleife wurde beendet...")
```

#### Erklärung des obigen Codes:

In jeder Iteration wird die Variable `i` ausgegeben und anschließend um 1 erhöht. Die Schleife läuft weiter, solange die Bedingung `i < 10` erfüllt ist.

##### Mehrere Bedingungen gleichzeitig prüfen

```python
i=0
j=5
print("Die Schleife beginnt...")
while i<10 and j<10:
    print(i)
    i+=1
    j+=1
print("\nDie Schleife wurde beendet...")
```

#### Erklärung des obigen Codes:

Die Variablen `i` und `j` werden in jeder Iteration erhöht. Die Schleife wird nur ausgeführt, solange sowohl `i < 10` als auch `j < 10` wahr sind.

### ACHTUNG! ES KANN EINE ENDLOSSCHLEIFE ENTSTEHEN

Wenn sich die in der Bedingung geprüfte Variable nie verändert und die Bedingung dadurch immer True bleibt, läuft die Schleife unendlich weiter.

```python
i=0
while i<100:
    print(i)
```

#### Erklärung des obigen Codes:

Da der Wert der Variable `i` innerhalb der Schleife niemals erhöht wird, bleibt die Bedingung `i < 100` immer wahr (True). Dadurch entsteht eine Endlosschleife.
