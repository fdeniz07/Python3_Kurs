## Fallstudie

Bildungsplattform: Kursanlage, Einschreibung, Bezahlung, Benachrichtigung und Zertifikatsablaeufe.

```python
# Einfaches ADR-Datenmodell
adr = {
    'id': 'ADR-001',
    'title': 'Use CQRS for enrollment flow',
    'context': 'High read/write asymmetry',
    'decision': 'Separate command and query handlers',
    'consequences': ['more classes', 'better scaling']
}
print(adr['title'])
```

## Entscheidungsfragen

- Wo liegt in diesem Problem die Komplexitaet?

- Wie werden die Testkosten beeinflusst?

## Best Practices

- Klein starten, Patterns bei Bedarf einfuehren
- Fuer jede grosse Entscheidung ein ADR dokumentieren
- Pattern-Auswahl an messbare Ziele koppeln

## Uebung

Schreibe fuer dein Projekt zwei ADRs: eines mit DDD-Fokus und eines mit EDD-Fokus.

```python
# Loesungsvorlage
def make_adr(adr_id, title, decision):
    return {
        'id': adr_id,
        'title': title,
        'decision': decision
    }

print(make_adr('ADR-002', 'Adopt Strategy for pricing', 'runtime strategy switch'))
```

## Abschluss

## Technischer Tiefgang

Auf fortgeschrittenem Niveau steht nicht nur Funktionalitaet, sondern die technische Nachvollziehbarkeit im Vordergrund.
Begriffe, Risiken, Betriebsaspekte und Qualitaetskriterien werden explizit gemacht, damit Entscheidungen reproduzierbar bleiben.

## Zentrale Fachbegriffe

Design Constraint
Trade-off
Failure Mode
Observability Signal
Quality Gate
Regression Risk

```python
# Zusatzbeispiel: Priorisierung technischer Risiken
risks=[{"name":"regression","score":9},{"name":"operability","score":8},{"name":"complexity","score":7}]
for r in sorted(risks,key=lambda x:x["score"],reverse=True):
    print(r["name"], r["score"])
```

## Haeufige Fehler und Debugging-Checkliste

- Sind Eingaben, Konfiguration und Randbedingungen explizit validiert?
- Sind reproduzierbare Schritte fuer den Fehler dokumentiert?
- Wurden Logs, Metriken und Tests gemeinsam ausgewertet?
- Ist die Korrektur durch einen neuen Test dauerhaft abgesichert?
- Gibt es eine kurze Lessons-Learned-Notiz fuer das Team?
