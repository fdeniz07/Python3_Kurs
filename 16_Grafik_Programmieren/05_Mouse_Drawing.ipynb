# Zeichnen mit der Maus
Dieses Beispiel zeigt, wie du mit Mausereignissen auf dem Canvas zeichnen kannst.

```python
import tkinter as tk

fenster = tk.Tk()
fenster.title('Zeichnen mit der Maus')

canvas = tk.Canvas(fenster, width=600, height=400, bg='white')
canvas.pack()

letzte_position = None

def startzeichnen(event):
    global letzte_position
    letzte_position = (event.x, event.y)

def zeichnen(event):
    global letzte_position
    x, y = event.x, event.y
    if letzte_position is not None:
        x0, y0 = letzte_position
        canvas.create_line(x0, y0, x, y, fill='black', width=2)
    letzte_position = (x, y)

def stopzeichnen(event):
    global letzte_position
    letzte_position = None

canvas.bind('<ButtonPress-1>', startzeichnen)
canvas.bind('<B1-Motion>', zeichnen)
canvas.bind('<ButtonRelease-1>', stopzeichnen)

fenster.mainloop()
```
