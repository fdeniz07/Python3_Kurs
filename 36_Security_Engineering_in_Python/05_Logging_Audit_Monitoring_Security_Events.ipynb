# Kapitel 36.5 - Logging, Audit und Monitoring von Security Events

Sicherheitsvorfaelle lassen sich nur erkennen und analysieren, wenn Ereignisse konsistent protokolliert und ueberwacht werden.

## Lernziele

- Security-Events von normalen Betriebslogs trennen
- Audit-Anforderungen in Log-Strategien integrieren
- Alarmregeln fuer kritische Ereignisse ableiten

```python
events = [
    {'type': 'login_failed', 'user': 'alice'},
    {'type': 'token_revoked', 'user': 'bob'},
    {'type': 'permission_denied', 'user': 'carol'}
]
for e in events:
    print(f"SECURITY_EVENT {e['type']} user={e['user']}")
```
