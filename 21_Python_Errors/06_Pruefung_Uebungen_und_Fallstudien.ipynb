# Beispiel 3 - Fallstudie

Symptom: Batchprozess stoppt bei erster Exception.

Aufgabe:
- robustes Sammelmuster definieren
- Teilergebnisse trotz Fehlern sichern

````python
# Referenzsnippet zu Fallstudie 3
def robust_batch(funktionen):
    ergebnisse = []
    for fn in funktionen:
        try:
            ergebnisse.append({'ok': True, 'wert': fn()})
        except Exception as fehler:
            ergebnisse.append({'ok': False, 'fehler': str(fehler)})
    return ergebnisse

jobs = [lambda: 1/1, lambda: 1/0, lambda: int('3')]
print(robust_batch(jobs))
````
