## Beispiel 1
Neue Datenbank und Tabelle erstellen.

````python
import pymysql

conn = pymysql.connect(
    host='localhost',
    user='root',
    password='passwort'
)
cursor = conn.cursor()

# Datenbank erstellen
cursor.execute("CREATE DATABASE IF NOT EXISTS schule")
cursor.execute("USE schule")

# Tabelle erstellen
cursor.execute('''
    CREATE TABLE IF NOT EXISTS schüler (
        id INT AUTO_INCREMENT PRIMARY KEY,
        name VARCHAR(100) NOT NULL,
        klasse VARCHAR(10)
    )
''')

conn.commit()
conn.close()
````

## Beispiel 2
Mit Kontextmanager arbeiten für sichere Verbindungen.

````python
import pymysql
from contextlib import closing

with closing(pymysql.connect(
    host='localhost',
    user='root',
    password='passwort',
    database='schule'
)) as conn:
    with conn.cursor() as cursor:
        cursor.execute("INSERT INTO schüler (name, klasse) VALUES (%s, %s)", ('Anna', '10A'))
        conn.commit()
        
        cursor.execute("SELECT * FROM schüler")
        print(cursor.fetchall())
````

## Praxisbeispiel
Eine Klasse für Datenbankoperationen.

````python
import pymysql

class MySQLDB:
    def __init__(self, host, user, password, database):
        self.conn = pymysql.connect(
            host=host,
            user=user,
            password=password,
            database=database
        )
        self.cursor = self.conn.cursor()
    
    def execute(self, query, params=None):
        self.cursor.execute(query, params)
        self.conn.commit()
    
    def fetch_all(self, query, params=None):
        self.cursor.execute(query, params)
        return self.cursor.fetchall()
    
    def close(self):
        self.conn.close()

db = MySQLDB('localhost', 'root', 'passwort', 'schule')
db.execute("INSERT INTO schüler (name, klasse) VALUES (%s, %s)", ('Beni', '10B'))
print(db.fetch_all("SELECT * FROM schüler"))
db.close()
````

## Häufige Fehler
- Falsche Verbindungsparameter
- `commit()` vergessen
- Keine Fehlerbehandlung

## Best Practice
Verwende Kontextmanager oder Try-Except für sichere Verbindungen.
