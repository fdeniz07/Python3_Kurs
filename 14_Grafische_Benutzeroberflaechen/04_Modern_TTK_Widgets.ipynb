Vorteile:

```python
import tkinter as tk
from tkinter import ttk

root = tk.Tk()
root.title("Erstes ttk Beispiel")

ttk.Label(root, text="Hallo ttk").pack(pady=10)
ttk.Button(root, text="OK").pack()

root.mainloop()
```

# ttk.Progressbar

## Beschreibung

Fortschrittsanzeige

## Beispiel


```python
import tkinter as tk
from tkinter import ttk

root=tk.Tk()

bar=ttk.Progressbar(root,length=250,mode='determinate')
bar['value']=60
bar.pack()

root.mainloop()
```

# ttk.Notebook

Mit einem Notebook können Registerkarten (Tabs) erstellt werden.


```python
import tkinter as tk
from tkinter import ttk

root=tk.Tk()

tabs=ttk.Notebook(root)

seite1=ttk.Frame(tabs)
seite2=ttk.Frame(tabs)

tabs.add(seite1,text="Start")
tabs.add(seite2,text="Einstellungen")

tabs.pack(expand=True,fill="both")

ttk.Label(seite1,text="Willkommen").pack(pady=20)
ttk.Label(seite2,text="Optionen").pack(pady=20)

root.mainloop()
```

# ttk.Treeview

Treeview wird häufig verwendet, um Tabellen darzustellen.


```python
import tkinter as tk
from tkinter import ttk

root=tk.Tk()

tree=ttk.Treeview(root,columns=("Name","Alter"),show="headings")

tree.heading("Name",text="Name")
tree.heading("Alter",text="Alter")

tree.insert("",tk.END,values=("Ali",25))
tree.insert("",tk.END,values=("Ayşe",30))
tree.insert("",tk.END,values=("Max",41))

tree.pack(expand=True,fill="both")

root.mainloop()
```

# Styles

Mit `ttk.Style()` kann das Aussehen verändert werden.


```python
import tkinter as tk
from tkinter import ttk

root=tk.Tk()

style=ttk.Style()

print(style.theme_names())
print(style.theme_use())

root.mainloop()
```
