## Dateioperationen - Lösungen der Hausaufgaben

#### Dateiinhalt umkehren

**Aufgabenstellung:**
Sie müssen alle Wörter, die Sie aus einer Datei einlesen, in einer zweiten Datei in umgekehrter Reihenfolge speichern.

```python
datei_umkehren = open("umkehren_lesen.txt", "r")
datei_umkehren_ziel = open("umkehren_schreiben.txt", "w")
datei_inhalt = datei_umkehren.read()
print("Original:", datei_inhalt)
datei_inhalt = datei_inhalt[::-1]
print("Umgekehrt:", datei_inhalt)
datei_umkehren_ziel.write(datei_inhalt)
datei_umkehren.close()
datei_umkehren_ziel.close()
```

#### Wortsuche in einer Datei

**Aufgabenstellung:**

Sie müssen überprüfen, ob ein vom Benutzer eingegebenes Wort innerhalb einer bestimmten Datei existiert.

```python
gesuchtes_wort = input("Bitte geben Sie das zu suchende Wort ein: ")
wort_suche_datei = open("wort_suche.txt", "r", encoding="utf-8")
datei_inhalt = wort_suche_datei.read()

# Jedes Wort splitten UND gleichzeitig von Punkt und Komma befreien
wort_liste = [wort.strip(".,!?") for wort in datei_inhalt.split()]

print("Die Datei enthält folgende Wörter:", wort_liste)
  
if gesuchtes_wort in wort_liste:
    print(f"Das Wort '{gesuchtes_wort}' wurde in der Datei gefunden. Das gesuchte Wort ist das {wort_liste.index(gesuchtes_wort) + 1}. Wort in der Datei.")
else:
    print(f"Das Wort '{gesuchtes_wort}' existiert nicht in der Datei.")
  
wort_suche_datei.close()
```

#### Abfrage von in einer Datei gespeicherten Filmdaten

**Aufgabenstellung:**

Überprüfen Sie, ob ein vom Benutzer eingegebener Film in der Datei vorhanden ist. Falls der Film in der Datei gefunden wird, geben Sie den Namen des Films, das Veröffentlichungsdatum und die Bewertung in der Konsole aus. Andernfalls geben Sie die Meldung "Der gesuchte Film ist nicht in der Liste vorhanden!" aus.

```python
gesuchter_film = input("Bitte geben Sie den Namen des gesuchten Films ein: ")
film_datei = open("filme.txt", "r")
film_daten = film_datei.readlines()

print(film_daten)

gefunden = False
for zeile in film_daten:
    zeile = zeile.strip()
    if zeile:
        film_name, film_datum, bewertung = zeile.split(",")
        if film_name.lower() == gesuchter_film.lower():
            print(f"Filmname: {film_name}\nVeröffentlichungsdatum: {film_datum}\nFilmbewertung: {bewertung}")
            gefunden = True
            break

if not gefunden:
    print("Der gesuchte Film ist nicht in der Liste vorhanden!")
film_datei.close()
```
