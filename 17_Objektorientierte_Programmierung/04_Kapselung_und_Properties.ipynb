# Kapitel 17.4 — Kapselung (Encapsulation) und Properties

## Lernziele
- Den Begriff Kapselung verstehen.
- Private Attribute schützen.
- Properties einsetzen.

## Theorie
Kapselung bedeutet, dass interne Daten vor ungewolltem Zugriff geschützt werden.
In Python erreichen wir das vor allem über `_` und `__` sowie Properties.

```python
class Bankkonto:
    def __init__(self, inhaber, saldo):
        self.inhaber = inhaber
        self.__saldo = saldo

    @property
    def saldo(self):
        return self.__saldo

    @saldo.setter
    def saldo(self, wert):
        if wert >= 0:
            self.__saldo = wert
        else:
            raise ValueError("Saldo darf nicht negativ sein")

konto = Bankkonto("Nora", 250)
print(konto.saldo)
konto.saldo = 300
print(konto.saldo)
```

## Beispiel 1
`__saldo` kann nicht direkt von außen verändert werden.

```python
class Person:
    def __init__(self, alter):
        self.__alter = alter

    @property
    def alter(self):
        return self.__alter

    @alter.setter
    def alter(self, wert):
        if wert >= 0:
            self.__alter = wert

p = Person(20)
print(p.alter)
```

## Praxisbeispiel
Eine Klasse `Temperatur` mit Prüfung auf sinnvolle Werte.

```python
class Temperatur:
    def __init__(self, grad):
        self.grad = grad

    @property
    def grad(self):
        return self.__grad

    @grad.setter
    def grad(self, wert):
        if -100 <= wert <= 100:
            self.__grad = wert
        else:
            raise ValueError("Ungültiger Temperaturwert")

t = Temperatur(25)
print(t.grad)
```

## Häufige Fehler
- Zu viel direktes Ändern von Attributen
- Properties ohne Setter verwenden
- Private Attribute als öffentlich behandeln

## Best Practice
Verwende Properties, wenn ein Attribut geprüft oder umgewandelt werden soll.

## Übung
Erstelle eine Klasse `Auto`, die nur Fahrzeugwerte zwischen 1900 und 2100 zulässt.

```python
class Auto:
    def __init__(self, baujahr):
        self.baujahr = baujahr

    @property
    def baujahr(self):
        return self.__baujahr

    @baujahr.setter
    def baujahr(self, wert):
        if 1900 <= wert <= 2100:
            self.__baujahr = wert
        else:
            raise ValueError("Ungültiger Baujahrwert")

a = Auto(2024)
print(a.baujahr)
```

## Zusammenfassung
Kapselung schützt Daten und macht Programme robuster.
