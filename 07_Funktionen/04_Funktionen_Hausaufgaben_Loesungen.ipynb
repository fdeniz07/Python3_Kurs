#### Funktion zur Berechnung der Anzahl von Vokalen
Funktionsbeschreibung:

Die Funktion nimmt eine Zeichenkette (String) von außen entgegen und gibt die Anzahl der darin enthaltenen Vokale aus.

vokale = ['a', 'e', 'i', 'u', 'o']

```python
def berechneVokale(text):
    vokale = ['a', 'e', 'i', 'u', 'o']
    vokalAnzahlen = [0 for i in range(5)]
    for index in range(len(vokale)):
        if vokale[index] in text:
            anzahl = text.count(vokale[index])
            vokalAnzahlen[index] = anzahl
    for index in range(len(vokale)):
        print("Der Buchstabe {} kommt {} Mal vor.".format(vokale[index], vokalAnzahlen[index]))
```

```python
berechneVokale("fatihdeniz")
```

#### Funktion zur Berechnung der durchschnittlichen Zeichenanzahl pro Wort
Funktionsbeschreibung:

Die Funktion nimmt einen Satz von außen entgegen und gibt die durchschnittliche Anzahl von Zeichen pro Wort zurück (als `return`).

Die durchschnittliche Zeichenanzahl kann wie folgt berechnet werden: Gesamtzahl der Zeichen / Gesamtzahl der Wörter

```python
def berechneDurchschnittlicheWortlaenge(text):
    gesamtWoerter = 0
    gesamtZeichen = 0
    for ch in text:
        if ch == " ":
            gesamtWoerter += 1
        gesamtZeichen += 1
    gesamtWoerter += 1
    print("Gesamte Zeichenanzahl: {} und gesamte Wortanzahl: {}".format(gesamtZeichen, gesamtWoerter))
    return gesamtZeichen // gesamtWoerter
```

```python
print(berechneDurchschnittlicheWortlaenge("Fatih Deniz lernt gerade Python."))
```

#### Funktion zur Berechnung der Anzahl eindeutiger Zeichen

Funktionsbeschreibung:

Die Funktion nimmt eine Zeichenkette (String) von außen entgegen und gibt die Anzahl der eindeutigen (einzigartigen) Zeichen zurück (als `return`).

Eindeutiges Zeichen: Jedes Zeichen wird nur genau einmal gezählt, mehrfach vorkommende Zeichen/Buchstaben werden kein zweites Mal berücksichtigt.

```python
def berechneEindeutigeZeichen(text):
    zeichenListe = []
    for ch in text:
        if ch not in zeichenListe:
            zeichenListe.append(ch)
    print(zeichenListe)
    return len(zeichenListe)
        
```

```python
print(berechneEindeutigeZeichen("fatihdeniz"))
```

#### Funktion zur Bestimmung der N-t größten und N-t kleinsten Zahl in einer Liste

Funktionsbeschreibung:
Die Funktion nimmt eine Liste und eine Zahl N von außen entgegen. Sie gibt jeweils die N-t größte und N-t kleinste Zahl dieser Liste zurück (als `return`).

Beispiel: Wenn die Liste = [1, 54, 32, 65, 23, 93] und die Zahl N = 2 ist,

sollten die Ergebnisse 65 (die zweitgrößte) und 23 (die zweitkleinste) zurückgegeben werden.


```python
def findeGruszteNUndKleinsteN(liste, N):
    liste.sort()
    print(liste)
    return liste[-N], liste[N-1]
```

```python
grossesN, kleinesN = findeGruszteNUndKleinsteN([1, 54, 32, 65, 23, 93], 3)
print("{} und {}".format(grossesN, kleinesN))
```

#### Rekursive Funktion zur Berechnung von ggT und kgV (OBEB-OKEK)

Funktionsbeschreibung:

Die Funktion nimmt zwei Zahlen von außen entgegen und berechnet rekursiv die Werte für den größten gemeinsamen Teiler (ggT) und das kleinste gemeinsame Vielfache (kgV). Sie können sich hierfür den Euklidischen Algorithmus (GCD) ansehen.

```python
def recursiveGGT(zahl1, zahl2):
    if zahl2 == 0:
        return zahl1
    return recursiveGGT(zahl2, zahl1 % zahl2)

def recursiveKGV(zahl1, zahl2):
    return zahl1 * zahl2 // recursiveGGT(zahl1, zahl2)
```

```python
print("Die ggT- und kgV-Ergebnisse für die Zahlen {} und {} lauten jeweils: {} und {}.".format(4, 7, recursiveGGT(4, 7), recursiveKGV(4, 7)))
```

##### Erklärung des obigen Codes:

`recursiveGGT(3, 5)`

Funktionsaufruf -> `return recursiveGGT(5, 3)`

Funktionsaufruf -> `return recursiveGGT(3, 2)`

Funktionsaufruf -> `return recursiveGGT(2, 1)`

Funktionsaufruf -> `return recursiveGGT(1, 0)` -> als Ergebnis wird `1` zurückgegeben.

`recursiveKGV(3, 5)`

Funktionsaufruf: `return 3 * 5 // recursiveGGT(3, 5)` -> `15 // 1 = 15` wird als Ergebnis zurückgegeben.
