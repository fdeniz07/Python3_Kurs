## Was sind Operatoren?

##### Arithmetische Operatoren: Addition, Subtraktion, Multiplikation, Ganzzahldivision, Gleitkommadivision, Modulo, Potenzierung
##### Vergleichsoperatoren: ==, !=, >, <, >=, <=
##### Logische Operatoren: AND, OR, NOT
##### Bitweise Operatoren: &, |, ^, ~, <<, >>
##### Extra: in, not in, is, is not

### Wie werden bitweise Operationen durchgeführt?
Jede Zahl hat eine Entsprechung auf der 0-1 Ebene (Binärebene).

64 -> 1000000 
64 = 2^6*1 + 2^5*0 + 2^4*0 + 2^3*0 + 2^2*0 + 2^1*0 + 2^0*0


35 -> 100011
35 = 2^5*1 + 2^1*1 + 2^0*1

### Bitweise Operatoren

##### & (AND) Operator
Ergibt True(1), wenn beide Werte True(1) sind, andernfalls False(0)

```python
a = 4 # 100 -> 2^2+0+0
b = 2 # 010 -> 2^1+0+0
# and 000 -> 0+0+0
print(a & b)
```

##### | (OR) Operator
Ergibt True(1), wenn mindestens einer der beiden Werte True(1) ist, andernfalls False(0)

```python
a = 4 # 100 -> 2^2+0+0
b = 2 # 010 -> 2^1+0+0
# or # 110 -> 2^2+2^1+0
print(a | b)
```

##### ^ (XOR) Operator
Ergibt False(0), wenn beide Werte gleich sind, andernfalls True(1)

```python
a = 4  # 100 -> 2^2+0+0
b = 5  # 101 -> 2^2+2^0
# xor # 001 -> 2^0
print(a ^ b)
```

##### ~ Operator
Funktioniert nach dem gleichen Prinzip wie NOT; ergibt False(0), wenn True(1) eingegeben wird, und True(1), wenn False(0) eingegeben wird

```python
a = 4  # 100 -> 2^2+0+0
b = ~a # 011 -> 2^0+2^1+0
print(b) # Der Grund für das negative Ergebnis liegt am Vorzeichen-Bit am Ende
```

##### << Operator
Wird für die bitweise Linksverschiebung verwendet; eine Verschiebung um ein Bit nach links bedeutet, dass die Zahl mit 2 multipliziert wird

```python
a = 4   # 100  ->  # 1000000
b = a << 4
print(b)
```

##### >> Operator
Wird für die bitweise Rechtsverschiebung verwendet; eine Verschiebung um ein Bit nach rechts bedeutet, dass die Zahl durch 2 geteilt wird

```python
a = 4 # 100    -> # 1
b = a >> 2
print(b)
```
