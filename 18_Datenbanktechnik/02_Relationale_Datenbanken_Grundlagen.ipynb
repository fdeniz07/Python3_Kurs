# Kapitel 18.2 — Relationale Datenbanken Grundlagen

## Lernziele
- Tabellen, Spalten und Zeilen verstehen.
- Primär- und Fremdschlüssel erklären.
- Beziehungen zwischen Tabellen aufbauen.

## Theorie

Eine relationale Datenbank besteht aus:
- **Tabellen**: Sammlungen von Daten mit gleichem Format
- **Spalten (Attribute)**: Die Eigenschaften der Daten
- **Zeilen (Datensätze)**: Einzelne Einträge von Daten
- **Schlüssel**: Eindeutige Identifikatoren für Zeilen

## Erklärung

**Primärschlüssel**: Eine Spalte oder Kombination von Spalten, die jeden Datensatz eindeutig identifiziert.

**Fremdschlüssel**: Eine Spalte, die auf den Primärschlüssel einer anderen Tabelle verweist und Beziehungen zwischen Tabellen herstellt.

## Beispiel 1
Eine Tabelle mit Primärschlüssel.

```
Tabelle: Kurse
| kurs_id (PK) | kursname      | lehrer   |
|--------------|---------------|----------|
| 1            | Mathematik    | Herr R   |
| 2            | Deutsch       | Frau M   |
| 3            | Englisch      | Herr B   |
```

## Beispiel 2
Beziehung zwischen zwei Tabellen über Fremdschlüssel.

```
Tabelle: Schüler
| schüler_id (PK) | name  | kurs_id (FK) |
|-----------------|-------|──────────────|
| 1               | Anna  | 1            |
| 2               | Beni  | 2            |
| 3               | Carla | 1            |
```

## Beziehungstypen

- **1:1 (Eins zu Eins)**: Ein Datensatz in Tabelle A bezieht sich auf genau einen Datensatz in Tabelle B
- **1:N (Eins zu Viele)**: Ein Datensatz in Tabelle A kann mehrere Datensätze in Tabelle B haben
- **M:N (Viele zu Viele)**: Viele Datensätze in Tabelle A können sich auf viele Datensätze in Tabelle B beziehen

## Praxisbeispiel
Ein Beispiel für eine 1:N-Beziehung: Ein Lehrer unterrichtet viele Schüler.

```
Tabelle: Lehrer
| lehrer_id (PK) | name      |
|-----------------|----------|
| 1               | Herr R   |
| 2               | Frau M   |

Tabelle: Schüler
| schüler_id (PK) | name  | lehrer_id (FK) |
|-----------------|-------|────────────────|
| 1               | Anna  | 1              |
| 2               | Beni  | 1              |
| 3               | Carla | 2              |
```

## Häufige Fehler
- Keinen Primärschlüssel definieren
- Fremdschlüssel-Beziehungen ignorieren
- Falsche Beziehungstypen verwenden

## Best Practice
Plane deine Tabellen und ihre Beziehungen sorgfältig, bevor du die Datenbank erstellst.

## Zusammenfassung
Schlüssel und Beziehungen sind das Fundament relationaler Datenbanken.
