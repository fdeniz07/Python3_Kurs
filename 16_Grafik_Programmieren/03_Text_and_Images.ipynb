# Text, Schriftarten und Bilder
Beispiele für Textplatzierung auf dem Canvas und einfache Bildanzeige (PIL optional).

## Tkinter: Text und Bilder
Mit Tkinter können wir einfache Fenster, Text und Bilder anzeigen. In diesem Abschnitt sehen wir zuerst, wie man Text auf einem Canvas darstellt.

## Beispiel 1: Ein einfaches Bild erzeugen
In diesem Beispiel erstellen wir ein neues Bild mit einem hellblauen Hintergrund, einem Rahmen und einem Text. So lernst du, wie du mit Pillow ein Bild von Grund auf aufbaust.

````python
from PIL import Image, ImageDraw

bild = Image.new('RGB', (300, 200), color='lightblue')
draw = ImageDraw.Draw(bild)
draw.rectangle((40, 40, 260, 160), outline='darkblue', width=4)
draw.text((70, 80), 'Hallo aus Pillow', fill='black')
bild.save('pillow_beispiel.png')
bild.show()
````

## Beispiel 2: Formen und Linien zeichnen
Hier zeichnen wir einen Kreis, eine Linie und einen Text. Das zeigt, wie du mit Pillow einfache geometrische Formen gestalten kannst.

````python
from PIL import Image, ImageDraw

bild_2 = Image.new('RGB', (300, 200), color='white')
draw_2 = ImageDraw.Draw(bild_2)
draw_2.ellipse((50, 40, 250, 160), fill='orange', outline='red', width=3)
draw_2.line((50, 160, 250, 40), fill='black', width=3)
draw_2.text((80, 170), 'Experiment mit Formen', fill='darkgreen')
bild_2.save('pillow_formen.png')
bild_2.show()
````

## Beispiel 3: Ein kleines Experiment mit Gittern und Formen
Dieses Experiment kombiniert Linien, Rechtecke und Kreise. Es hilft dir, die Grundbausteine von Pillow besser zu verstehen.

````python
from PIL import Image, ImageDraw, ImageFont

bild_3 = Image.new('RGB', (400, 250), color='white')
draw_3 = ImageDraw.Draw(bild_3)
for x in range(0, 400, 40):
    draw_3.line((x, 0, x, 250), fill='lightgray', width=1)

for y in range(0, 250, 25):
    draw_3.line((0, y, 400, y), fill='lightgray', width=1)

draw_3.rectangle((80, 60, 320, 190), outline='purple', width=4)
draw_3.ellipse((120, 90, 280, 160), fill='yellow', outline='orange', width=3)
try:
    font = ImageFont.truetype('DejaVuSans.ttf', 24)
except OSError:
    font = ImageFont.load_default()

draw_3.text((95, 200), 'Pillow funktioniert!', font=font, fill='black')
bild_3.save('pillow_experiment.png')
bild_3.show()
````

## Mehr zu Pillow
Bevor wir mit Pillow arbeiten, müssen wir die richtige Bibliothek installieren. Pillow ist die moderne Bibliothek für Bildverarbeitung in Python. Sie wird oft mit dem Namen `PIL` verwendet, weil sie aus der älteren Bibliothek „Python Imaging Library“ entstanden ist.

### 1. Welche Bibliothek wird benötigt?
Für die meisten Beispiele verwenden wir:

```python
from PIL import Image, ImageDraw, ImageFont
```

Hier bedeutet:
- `Image`: für das Erstellen, Öffnen und Speichern von Bildern
- `ImageDraw`: für das Zeichnen von Formen und Text
- `ImageFont`: für Schriftarten

### 2. Installation
Falls Pillow noch nicht installiert ist, kannst du es so installieren:

```bash
pip install pillow
```

### 3. Versionsprüfung
Es ist hilfreich zu prüfen, ob Pillow korrekt installiert wurde. Dazu kannst du die Version anzeigen:

```python
import PIL
print(PIL.__version__)
```

Damit prüfst du, dass die Bibliothek verfügbar ist und welche Version gerade genutzt wird.

### 4. Wichtige Begriffe
- `Bild`: ein Objekt mit Breite, Höhe und Farbmodus
- `Modus`: beschreibt die Art der Farben, zum Beispiel `RGB` für Farben oder `L` für Graustufen
- `Pixel`: die kleinen Punkte, aus denen ein Bild besteht
- `Canvas`: die Zeichenfläche, auf der wir zeichnen

### 5. Typische Arbeitsweise mit Pillow
Ein typischer Ablauf sieht so aus:

1. Ein neues Bild erzeugen
2. Mit `ImageDraw` zeichnen
3. Das Bild speichern
4. Das Bild anzeigen

### 6. Beispiel für die Versionsprüfung
```python
from PIL import Image
import PIL

print("Pillow-Version:", PIL.__version__)
bild = Image.new("RGB", (100, 60), color="lightgreen")
print("Bildgröße:", bild.size)
print("Farbmodus:", bild.mode)
```

Diese Zeilen zeigen dir, dass Pillow funktioniert und dass du wichtige Eigenschaften eines Bildes direkt abfragen kannst.

### 7. Wichtige Methoden im Überblick
- `Image.new(...)`: erzeugt ein neues Bild
- `bild.size`: gibt Breite und Höhe zurück
- `bild.mode`: zeigt den Farbmodus an
- `ImageDraw.Draw(bild)`: erstellt ein Zeichenwerkzeug
- `draw.rectangle(...)`, `draw.ellipse(...)`, `draw.line(...)`, `draw.text(...)`: zeichnen Formen und Text
- `bild.save(...)`: speichert das Bild
- `bild.show()`: öffnet das Bild im Standardbildbetrachter

Diese Methoden helfen dir, Bilder gezielt zu erzeugen, zu bearbeiten und zu speichern.

## Übung
Baue ein eigenes Bild mit einem Hintergrund, einem Rechteck und einem Text. Ändere die Farben selbst und speichere das Bild unter einem anderen Namen.

````python
import tkinter as tk

fenster = tk.Tk()
fenster.title('Text und Bilder')

canvas = tk.Canvas(fenster, width=500, height=300, bg='white')
canvas.pack()

# Text
canvas.create_text(250, 40, text='Grafik mit Text', font=('Arial', 20, 'bold'), fill='darkblue')
canvas.create_text(250, 80, text='Hier lernen wir Schriftarten und Farben', font=('Arial', 12), fill='black')

# Optional: Bild anzeigen (PIL benötigt)
# from PIL import Image, ImageTk
# img = Image.open('path/to/image.png')
# photo = ImageTk.PhotoImage(img)
# canvas.create_image(250, 180, image=photo)

fenster.mainloop()
````
