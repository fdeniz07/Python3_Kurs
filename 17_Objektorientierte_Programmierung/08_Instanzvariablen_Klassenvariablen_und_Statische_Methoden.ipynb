## Theorie
Eine Instanzvariable gehört zu einem einzelnen Objekt. Eine Klassenvariable ist für alle Instanzen einer Klasse gleich.
Statische Methoden hängen an der Klasse, aber nicht an einem konkreten Objekt. Klassenmethoden erhalten die Klasse selbst als ersten Parameter.

```python
class Hund:
Canis familiaris"  # Klassenvariable

    def __init__(self, name):
        self.name = name  # Instanzvariable

h1 = Hund("Bello")
h2 = Hund("Milo")
print(h1.name)
print(h2.name)
print(h1.art)
print(h2.art)
```
