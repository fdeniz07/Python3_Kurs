# Theorie

NumPy ist die Basis numerischer Arbeit in Python.
Arrays sind kompakter und schneller als reine Python-Listen bei vielen Operationen.

# Erklaerung

Wichtige Kernideen:
- `ndarray` mit Form (`shape`) und Datentyp (`dtype`)
- Vektorisierung statt expliziter Schleifen
- Broadcasting fuer Operationen unterschiedlicher Formen

# Syntax

```python
import numpy as np
a = np.array([1,2,3])
b = a * 2
```

# Merke

NumPy-Performance kommt durch Array-Operationen, nicht durch Python-Schleifen.

# Parameter

- `dtype` bei `np.array`
- `axis` bei Aggregationen (`sum`, `mean`)

# Rueckgabewert

NumPy-Funktionen liefern meist Arrays oder skalare Werte, abhaengig von der Operation.

````python
import numpy as np

# Beispiel 1
a = np.array([1, 2, 3, 4], dtype=np.float64)
print(a, a.dtype, a.shape)
````

````python
# Beispiel 2
matrix = np.array([[1, 2], [3, 4], [5, 6]])
print('sum axis=0:', matrix.sum(axis=0))
print('sum axis=1:', matrix.sum(axis=1))
````

````python
# Beispiel 3: Broadcasting
x = np.array([1, 2, 3])
y = np.array([[10], [20], [30]])
print(x + y)
````

# Haeufige Fehler

1. Falsche Form (`shape`) bei Operationen.
2. Python-Listen mit NumPy-Arrays verwechseln.
3. Datentyp-Konvertierungen ignorieren.

# Best Practice

- Shapes frueh ausgeben und pruefen.
- Rechenpipelines in kleine Schritte teilen.
