# Theorie

Seaborn baut auf Matplotlib auf und liefert ausdrucksstarke Standardstile fuer statistische Grafiken.

# Erklaerung

Wichtige Plottypen:
- `histplot`, `kdeplot`
- `boxplot`, `violinplot`
- `scatterplot`, `lineplot`, `pairplot`

# Syntax

```python
import seaborn as sns
sns.boxplot(data=df, x='gruppe', y='wert')
```

# Merke

Weniger Plottypen, dafuer sauber interpretiert, ist didaktisch staerker.

# Parameter

- `hue` fuer Untergruppen
- `palette` fuer Farben
- `kind` bei facettenbasierten Plots

# Rueckgabewert

Seaborn gibt Matplotlib-Axes-Objekte zurueck und bleibt so fein anpassbar.

````python
import seaborn as sns
import pandas as pd
import matplotlib.pyplot as plt

df = pd.DataFrame({
    'gruppe': ['A','A','B','B','C','C'],
    'wert': [10,12,9,15,14,13],
    'kategorie': ['x','y','x','y','x','y']
})

# Beispiel 1
sns.barplot(data=df, x='gruppe', y='wert')
plt.title('Seaborn Barplot')
plt.show()
````

````python
# Beispiel 2
sns.boxplot(data=df, x='gruppe', y='wert')
plt.title('Boxplot je Gruppe')
plt.show()
````

````python
# Beispiel 3
sns.scatterplot(data=df, x='wert', y='gruppe', hue='kategorie', s=100)
plt.title('Scatter mit hue')
plt.show()
````
