````python
# Beispiel 1: Simulierte Messdaten
import numpy as np
import pandas as pd

np.random.seed(42)
df = pd.DataFrame({
    'gruppe': np.random.choice(['kontrolle', 'test'], size=100),
    'wert': np.random.normal(loc=50, scale=10, size=100)
})
print(df.head())
````

````python
# Beispiel 2: Gruppenvergleich
print(df.groupby('gruppe')['wert'].agg(['mean', 'std', 'count']))
````

````python
# Beispiel 3: einfache Visualisierung
import matplotlib.pyplot as plt

df.boxplot(column='wert', by='gruppe', figsize=(6,4))
plt.title('Gruppenvergleich')
plt.suptitle('')
plt.xlabel('Gruppe')
plt.ylabel('Wert')
plt.show()
````
