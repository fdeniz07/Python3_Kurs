## Modulkonzept


### Was ist ein Modul? Wozu dient es?

Jede separate Datei, die wir in Python schreiben, wird als ein Modul betrachtet. Wir können den Code, die Funktionen und die Klassen, die wir innerhalb von Modulen schreiben, später in anderen Programmen oder an verschiedenen Stellen desselben Programms aufrufen und verwenden.

### Ein Modul hinzufügen

Um ein zuvor geschriebenes Modul in einer beliebigen Codedatei wiederzuverwenden, müssen wir das Modul mit dem Schlüsselwort `import` in die Codedatei einbinden.

````python
import math # Ich habe das Math-Modul in diese Python-Datei eingebunden
import time # Ich habe das Time-Modul in diese Python-Datei eingebunden
import os   # Ich habe das OS-Modul in diese Python-Datei eingebunden
import sys  # Ich habe das Sys-Modul in diese Python-Datei eingebunden
````

### Einen bestimmten Teil eines Moduls hinzufügen

Syntax-Struktur:

```python
from modul_name import funktion_name
```

````python
from os import name
````

````python
name
````

### Alle Funktionen innerhalb eines Moduls hinzufügen

Syntax-Struktur:

```python
from modul_name import *
```

````python
from os import *
````

### Module beim Hinzufügen umbenennen

Beim Einbinden eines Moduls in den Codeblock können wir ihm mit dem Schlüsselwort `as` einen selbst definierten Namen geben.

````python
import math as mathematik
import os as betriebssystem
import sys as system
````

### Funktionen innerhalb eines Moduls anzeigen

Mit der Funktion `dir()` können wir die Funktionen innerhalb eines Moduls auflisten.

````python
dir(mathematik)
````

````python
dir(betriebssystem)
````

### Die Funktionalität von Funktionen innerhalb eines Moduls einsehen

Mit der Funktion `help()` können wir die Funktionalitäten der Funktionen innerhalb eines Moduls auflisten.

````python
help(mathematik)
````
