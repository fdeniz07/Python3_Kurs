Thema: `random`-Modul, eigene Funktionen, if-Kontrollstruktur und for-Schleife

## a) Modul `random` importieren und Zufallszahl generieren

**Aufgabe:** Importiere das Modul `random` und verwende eine Funktion aus diesem Modul, um eine Zufallszahl zwischen 1 und 10 zu generieren. Speichere diese Zahl in einer Variablen.

```python
import random

zufallszahl = random.randint(1, 10)
print("Generierte Zufallszahl:", zufallszahl)
```

## b) Funktion `berechne_quadrat` definieren

**Aufgabe:** Definiere eine Funktion namens `berechne_quadrat`, die einen Parameter nimmt, dessen Quadrat berechnet und das Ergebnis zurückgibt.

```python
def berechne_quadrat(zahl):
    return zahl ** 2
```

## c) if-Kontrollstruktur

**Aufgabe:** Überprüfe, ob die generierte Zufallszahl größer als 5 ist. Falls ja, rufe `berechne_quadrat` mit der Zufallszahl auf und gib das Ergebnis aus. Falls nein, gib eine Nachricht aus, dass die Zahl kleiner oder gleich 5 ist.

```python
if zufallszahl > 5:
    ergebnis = berechne_quadrat(zufallszahl)
    print("Das Quadrat von", zufallszahl, "ist", ergebnis)
else:
    print("Die Zahl", zufallszahl, "ist kleiner oder gleich 5.")
```

## d) for-Schleife bis zur Zufallszahl

**Aufgabe:** Implementiere eine Schleife, die von 1 bis zur generierten Zufallszahl läuft und dabei jedes Mal die aktuelle Zahl ausgibt.

```python
for i in range(1, zufallszahl + 1):
    print(i)
```

## Komplettes Programm (alle Teile zusammen)

```python
import random

# a) Zufallszahl zwischen 1 und 10 generieren
zufallszahl = random.randint(1, 10)
print("Generierte Zufallszahl:", zufallszahl)

# b) Funktion zur Berechnung des Quadrats
def berechne_quadrat(zahl):
    return zahl ** 2

# c) if-Kontrollstruktur: Überprüfung ob zufallszahl > 5
if zufallszahl > 5:
    ergebnis = berechne_quadrat(zufallszahl)
    print("Das Quadrat von", zufallszahl, "ist", ergebnis)  # print(f"Das Quadrat von {zufallszahl} ist {ergebnis}.")
else:
    print("Die Zahl", zufallszahl, "ist kleiner oder gleich 5.") # print(f"Die Zahl {zufallszahl} ist kleiner oder gleich 5.")

# d) for-Schleife von 1 bis zur Zufallszahl
for i in range(1, zufallszahl + 1):
    print(i)
```
