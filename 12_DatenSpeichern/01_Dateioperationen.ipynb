## Dateioperationen (Datei-Handling)

#### Warum brauchen wir das?

Während des Programmablaufs speichern wir Daten, Informationen und Werte, die wir vom Benutzer erhalten oder berechnen, im Arbeitsspeicher des Computers. Wenn das Programm beendet wird, werden diese Informationen gelöscht und es ist nicht mehr möglich, darauf zuzugreifen. Um später darauf zuzugreifen zu können, müssen wir diese Daten dauerhaft (persistent) speichern. Dafür gibt es viele Methoden; die grundlegendste besteht darin, sie in eine Datei auf einem Datenträger zu schreiben, sodass wir später wieder darauf zugreifen können.

### Datei erstellen, öffnen und schließen
1. `open()`-Funktion: Wird zum Erstellen oder Öffnen einer Datei verwendet.
2. `close()`-Funktion: Schließt die Datei, um Ressourcen freizugeben.

`open("vollständiger_dateipfad", datei_öffnungsmodus)`

#### Dateimodus-Erklärungen (File Modes)

Beim Öffnen von Dateien mit der Funktion `open()` müssen wir angeben, zu welchem Zweck die Datei geöffnet wird:

| Modus | Beschreibung | Wenn Datei nicht existiert? | Wenn Datei bereits existiert? |
| :--- | :--- | :--- | :--- |
| **`r`** | Nur Lesen (Read). Der Cursor ist am Anfang. | Fehler (`FileNotFoundError`) | Datei wird geöffnet, Inhalt bleibt unverändert. |
| **`w`** | Nur Schreiben (Write). | Erstellt eine neue leere Datei. | Löscht den gesamten Inhalt (Überschreibt komplett). |
| **`a`** | Anhängen (Append). Der Cursor ist am Ende. | Erstellt eine neue leere Datei. | Bestehender Inhalt bleibt erhalten, neuer Text wird angehängt. |
| **`r+`** | Lesen und Schreiben. Der Cursor ist am Anfang. | Fehler (`FileNotFoundError`) | Neuer Text überschreibt alten Text Zeichen für Zeichen ab dem Anfang. |
| **`w+`** | Schreiben und Lesen. | Erstellt eine neue leere Datei. | Löscht den gesamten Inhalt (Überschreibt komplett). |
| **`a+`** | Anhängen und Lesen. | Erstellt eine neue leere Datei. | Bestehender Inhalt bleibt erhalten, neuer Text wird am Ende angehängt. |
| **`rb` / `wb`** | Binärmodus für Lesen / Schreiben. | Verhält sich wie oben, aber für Binärdaten (Bilder, Audio). | - |



#### Datei-Öffnungsmodi

1. `"w"`: Steht für "write" (Schreiben) und ermöglicht es uns, die Datei im Schreibmodus zu öffnen. Wenn der angegebene Dateiname nicht existiert, wird eine neue Datei erstellt. Wenn die Datei bereits existiert, wird ihr Inhalt gelöscht und neue Daten werden darüber geschrieben.

2. `"a"`: Steht für "append" (Anhängen). Wenn die Datei nicht existiert, wird sie neu erstellt. Wenn sie bereits existiert, werden die neuen Daten am Ende der Datei angehängt, ohne den bestehenden Inhalt zu löschen.

3. `"r"`: Steht für "read" (Lesen). Wenn die zu öffnende Datei nicht existiert, tritt ein Fehler auf.

```python
try:
    datei1 = open("beispiel.txt", "w")
    datei2 = open("beispiel2.txt", "a")
    datei3 = open("beispiel3.txt", "r")
    datei3.close()
except FileNotFoundError:
    print("Datei wurde nicht gefunden!")
```

```python
# Beispiel: Verwendung von seek und tell
with open("test.txt", "w+", encoding="utf-8") as f:
    f.write("Hallo Welt")
    f.seek(0)
    print("Position nach seek(0):", f.tell())
    print(f.read(5))
```

```python
# Beispiel für den 'w' und 'r' Modus mit UTF-8 Kodierung
try:
    with open("beispiel_datei.txt", "w", encoding="utf-8") as datei:
        datei.write("Erste Zeile in Python\nZweite Zeile\nDritte Zeile")
    print("Datei erfolgreich im 'w'-Modus erstellt.")
except Exception as e:
    print("Ein Fehler ist aufgetreten:", e)
```

#### Daten in eine Datei schreiben

1. `write()`-Funktion: Nach dem Öffnen der Datei können wir mit der Funktion `write()` Daten in die Datei schreiben.

```python
datei = open("beispiel.txt", mode="w")
datei.write("Max Mustermann\n")
datei.write("Python-Programmierung\n")
datei.write("Programmieren macht Spaß")
datei.close()
```

```python
try:
    datei2 = open("beispiel2.txt", "r")
    datei2.write("MaxMustermann")
except:
    print("Fehler beim Schreiben im Lesemodus!")
```

#### Daten aus einer Datei lesen
1. `read()`-Funktion: Nach dem Öffnen der Datei im Lesemodus (`"r"`) können wir den gesamten Inhalt der Datei mit der Funktion `read()` auslesen.

```python
try:
    datei = open("beispiel.txt", mode="r")
    print(datei.read())
    datei.close()
except FileNotFoundError:
    print("Datei nicht gefunden!")
```

```python
try:
    datei = open("beispiel.txt", mode="r")
    print(datei.readline())
    print(datei.readline())
    print(datei.readline())
    datei.close()
except FileNotFoundError:
    print("Datei nicht gefunden!")
```

```python
try:
    datei = open("beispiel.txt", mode="r")
    alle_zeilen = datei.readlines()
    for zeile in alle_zeilen:
        print(zeile, end="")
    datei.close()
except FileNotFoundError:
    print("Datei nicht gefunden!")
```

#### Daten schreiben, ohne die Datei zu löschen (Anhängen)

1. `"a"` (Append)-Modus: Ermöglicht das Hinzufügen von neuen Zeilen oder Daten am Ende einer bestehenden Datei, ohne den bisherigen Inhalt zu überschreiben.

```python
datei = open("beispiel.txt", mode="a")
datei.write("\nPython-Programmierkurs")
datei.close()
```

#### Stream-Methoden und Cursor-Verwaltung

Um sich innerhalb einer Datei zu bewegen und die genaue Position der Daten zu steuern, nutzen wir Stream-Methoden:

* **`tell()`**: Gibt die aktuelle Position des Dateicursors (in Bytes) zurück.
* **`seek(offset, whence)`**: Bewegt den Cursor an eine bestimmte Byte-Position.
    * `seek(0)`: Bringt den Cursor zurück an den absoluten Anfang der Datei.
    * `seek(0, 2)`: Bringt den Cursor an das absolute Ende der Datei.
* **`flush()`**: Schreibt alle Daten, die sich noch im Zwischenspeicher (Buffer) befinden, sofort auf die Festplatte.
* **`truncate(size)`**: Schneidet die Datei auf die angegebene Größe (in Bytes) ab.

```python
# Demonstration von seek() und tell()
with open("cursor_test.txt", "w+", encoding="utf-8") as f:
    print("Anfangsposition des Cursors:", f.tell()) # Gibt 0 aus
    f.write("Hallo Welt")
    print("Position nach dem Schreiben:", f.tell()) # Gibt die Anzahl der Bytes aus
    
    # Wenn wir jetzt lesen, ist es leer, weil der Cursor am Ende steht!
    # Daher bringen wir den Cursor mit seek(0) zurück an den Anfang:
    f.seek(0)
    print("Position nach seek(0):", f.tell()) # Gibt wieder 0 aus
    print("Inhalt der Datei:", f.read())
```

```python
# Demonstration von truncate() zum Kürzen von Dateien
with open("kuerzen_test.txt", "w+", encoding="utf-8") as f:
    f.write("ABCDEFGHIJ") # 10 Zeichen/Bytes geschrieben
    f.flush() # Daten sofort auf die Festplatte schreiben
    
    f.truncate(5) # Die Datei nach 5 Bytes abschneiden (nur 'ABCDE' bleibt)
    f.seek(0)
    print("Gekürzter Inhalt:", f.read()) # Ausgabe: ABCDE
```

```python
# Erweitertes Beispiel: Ändern von Daten mitten in der Datei mit 'r+'
with open("zahlen.txt", "w+", encoding="utf-8") as f:
    f.write("1234567890")

# Datei im Lese- und Schreibmodus öffnen
with open("zahlen.txt", "r+", encoding="utf-8") as f:
    f.seek(5) # Springe zum Index 5 (6. Zeichen)
    f.write("X") # Überschreibt die '6' mit einem 'X'
    f.seek(0)
    print("Aktualisierte Datei:", f.read()) # Ausgabe: 12345X7890
```
