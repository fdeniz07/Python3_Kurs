## Schleifen

#### Sprunganweisungen

###### break:
Ein Schlüsselwort, das verwendet wird, um eine Schleife unter einer bestimmten Bedingung sofort zu beenden.
###### continue:
Ein Schlüsselwort, das die aktuelle Iteration beendet und direkt zur nächsten Iteration übergeht.
###### pass:
Ein Schlüsselwort, das als Platzhalter dient und keine Aktion ausführt.

```python
i=0
while True:
    print(i,end=" ")
    if i>=50:
        print("Die Schleife wird beendet")
        break
    i+=2
```

```python
for i in range(100):
    if not i%5==0:
        continue
    print(i,end=" ")
```

```python
for i in range(100):
    if not i%5==0:
        pass
    print(i,end=" ")
```
