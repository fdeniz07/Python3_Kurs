### Aufgabe 2: Passwortüberprüfung

Aufgabenbeschreibung: Überprüfen Sie, ob ein vom Benutzer eingegebenes Passwort die Anforderungen erfüllt. Wenn alle Bedingungen erfüllt sind, soll "Passwort ist gültig" ausgegeben werden, andernfalls eine passende Fehlermeldung.

Bedingung 1 -> Das Passwort muss Groß- und Kleinbuchstaben enthalten.

Bedingung 2 -> Das Passwort muss mindestens eine Zahl enthalten.

Bedingung 3 -> Das Passwort muss länger als 8 Zeichen sein.

Bedingung 4 -> Das Passwort muss kürzer als 16 Zeichen sein.

Beispiele:

passwort="maxMst123" → "Passwort ist gültig"

passwort="maxmst123" → "Es muss mindestens ein Großbuchstabe verwendet werden"

passwort="mA12" → "Das Passwort muss mindestens 8 Zeichen enthalten"

passwort="MaxMustermann123456" → "Das Passwort darf höchstens 16 Zeichen enthalten"

passwort="maxmustermann" → "Das Passwort muss mindestens eine Zahl enthalten" 

### Aufgabe 3: Prüfung auf perfekte Zahlen

Aufgabenbeschreibung: Lesen Sie eine Zahl ein und prüfen Sie, ob es sich um eine perfekte Zahl handelt.

Perfekte Zahl: Eine Zahl heißt perfekt, wenn die Summe ihrer positiven Teiler (ohne die Zahl selbst) wieder die Zahl ergibt.

Beispiel:
Die Zahl 6 ist eine perfekte Zahl, da 1 + 2 + 3 = 6.

##### Zusatzaufgabe:
Geben Sie alle perfekten Zahlen zwischen 1 und 1000 aus.

### Aufgabe 5: Armstrong-Zahl prüfen

Aufgabenbeschreibung: Prüfen Sie, ob eine vom Benutzer eingegebene Zahl eine Armstrong-Zahl ist.

Armstrong-Zahl: Eine Zahl heißt Armstrong-Zahl, wenn die Summe der dritten Potenzen ihrer Ziffern wieder die Zahl selbst ergibt.

Beispiel:
371 = 3³ + 7³ + 1³

##### Zusatzaufgabe:
Geben Sie alle Armstrong-Zahlen zwischen 1 und 1000 aus.
