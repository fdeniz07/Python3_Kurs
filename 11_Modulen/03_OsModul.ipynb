## OS-Modul

Ermöglicht es Ihnen, Informationen über das von Ihnen verwendete Betriebssystem zu erhalten und verschiedene Operationen mit dem Betriebssystem durchzuführen.

Das Modul sys ermöglicht es dir, die Arbeitsweise des Laufzeitsystems zu beobachten und zu beeinflussen.

```python
import os
```

Hier ist eine Übersicht der wichtigsten Funktionen des `os`-Moduls.
   
| Funktion/Attribut | Beschreibung | Verwendung |
| :--- | :--- | :--- |
| `os.name` | Name des Betriebssystems ('nt', 'posix') | `os.name` |
| `os.curdir` | Aktuelles Verzeichnis (Zeichenfolge, meist '.') | `os.curdir` |
| `os.pardir` | Übergeordnetes Verzeichnis (Zeichenfolge, meist '..') | `os.pardir` |
| `os.getcwd()` | Aktuelles Verzeichnis abrufen | `os.getcwd()` |
| `os.chdir(path)` | Verzeichnis wechseln | `os.chdir('Pfad')` |
| `os.listdir()` | Inhalt eines Ordners auflisten | `os.listdir()` |
| `os.startfile(f)` | Datei/Ordner wie mit Doppelklick öffnen | `os.startfile('test.txt')` |
| `os.mkdir(path)` | Neuen Ordner erstellen | `os.mkdir('Name')` |
| `os.makedirs(p)` | Geschachtelte Ordner erstellen | `os.makedirs('a/b/c')` |
| `os.rename(s, d)` | Datei/Ordner umbenennen | `os.rename('alt', 'neu')` |
| `os.remove(path)` | Datei löschen | `os.remove('datei.txt')` |
| `os.rmdir(path)` | Leeren Ordner löschen | `os.rmdir('Ordner')` |
| `os.removedirs(p)` | Leere Unterordner rekursiv löschen | `os.removedirs('a/b/c')` |

| `os.path.dirname(p)` | Verzeichnisanteil eines Pfades | `os.path.dirname(pfad)` |
| `os.path.exists(p)` | Prüfen, ob Pfad existiert | `os.path.exists('Pfad')` |
| `os.path.isdir(p)` | Prüfen, ob Pfad ein Ordner ist | `os.path.isdir('Pfad')` |
| `os.path.isfile(p)` | Prüfen, ob Pfad eine Datei ist | `os.path.isfile('Pfad')` |
| `os.path.join(a, b)` | Pfad plattformunabhängig zusammenfügen | `os.path.join('a', 'b')` |
| `os.path.split(p)` | Pfad in (Verzeichnis, Dateiname) splitten | `os.path.split('pfad')` |
| `os.path.splitext(f)` | Pfad in (Name, Endung) splitten | `os.path.splitext('file.txt')` |
| `os.walk(top)` | Verzeichnisbaum rekursiv durchlaufen | `os.walk('.')` |

### Attribut `name`

Gibt den Namen des Betriebssystems zurück;
Für Windows wird 'nt' ausgegeben.
Für macOS und GNU/Linux wird 'posix' ausgegeben.

```python
os.name
```

### Attribut `sep`

Gibt das Trennzeichen für Dateipfade an, das vom jeweiligen Betriebssystem verwendet wird;

Für macOS und GNU/Linux '/'

### Funktion `path.walk`

Es wird der gesamte Verzeihnisbaum mit Wurzel `top` durchlaufen und eine Folge und Tupeln zurückgegeben. Jedes Tupel dokumentiert ein Verzeichnis und hat den Aufbau `(Pfad, Unterverzeichnisse, Dateien)`.

```python
os.chdir(r'c:\\PROJEKTE_LOKAL\\BILDUNGS\\VelpTEC\\Python3_Kurs\\11_Modulen')
os.getcwd()

for wurzel, ordner, dateien in os.walk('.'): # Iteriere über alle Verzeichnisse und Unterverzeichnisse
    print(f"Aktuelles Verzeichnis: {wurzel}")
    for datei in dateien: # Iteriere über die Dateien im aktuellen Verzeichnis
        pfad = os.path.join(wurzel, datei) # Erstelle den vollständigen Pfad zur Datei
        print(f" - Datei gefunden: {datei} (Pfad: {pfad})")
```
