# Kapitel 18.9 — NoSQL Datenbanken mit MongoDB

## Lernziele
- MongoDB verstehen.
- PyMongo verwenden.
- Dokumente einfügen, abrufen und ändern.

## Theorie

MongoDB ist eine NoSQL-Datenbank, die Daten als JSON-ähnliche Dokumente speichert. Sie ist flexibler als relationale Datenbanken und besser für unstrukturierte Daten.

## Installation

````python
from pymongo import MongoClient

# Verbindung
client = MongoClient('mongodb://localhost:27017/')
db = client['schule']
collection = db['schüler']

print("Verbindung erfolgreich")
````

## Beispiel 1
Dokumente einfügen.

````python
# Einzelnes Dokument
schueler1 = {
    'name': 'Anna',
    'klasse': '10A',
    'noten': [5, 4, 5]
}
result = collection.insert_one(schueler1)
print(f"Eingefügt mit ID: {result.inserted_id}")

# Mehrere Dokumente
schueler_liste = [
    {'name': 'Beni', 'klasse': '10B', 'noten': [4, 4, 3]},
    {'name': 'Carla', 'klasse': '10A', 'noten': [5, 5, 4]},
    {'name': 'David', 'klasse': '11A', 'noten': [3, 3, 4]}
]
result = collection.insert_many(schueler_liste)
print(f"Eingefügt {len(result.inserted_ids)} Dokumente")
````

## Beispiel 2
Dokumente abrufen.

````python
# Alle Dokumente
alle = collection.find()
for doc in alle:
    print(doc)

# Mit Filter
schüler_10a = collection.find({'klasse': '10A'})
for doc in schüler_10a:
    print(doc)

# Einzelnes Dokument
anna = collection.find_one({'name': 'Anna'})
print(anna)
````

## Beispiel 3
Dokumente aktualisieren und löschen.

````python
# Aktualisieren
collection.update_one(
    {'name': 'Anna'},
    {'$set': {'klasse': '10B'}}
)

# Mehrere aktualisieren
collection.update_many(
    {'klasse': '10A'},
    {'$inc': {'durchschnitt_erhöht': 0.1}}
)

# Löschen
collection.delete_one({'name': 'David'})
print("Gelöscht")
````

## Praxisbeispiel
Eine Klasse für MongoDB-Operationen.

````python
from pymongo import MongoClient

class MongoDBWrapper:
    def __init__(self, uri, db_name, collection_name):
        self.client = MongoClient(uri)
        self.db = self.client[db_name]
        self.collection = self.db[collection_name]
    
    def insert_one(self, doc):
        return self.collection.insert_one(doc).inserted_id
    
    def find_all(self):
        return list(self.collection.find())
    
    def find_by_filter(self, filter_dict):
        return list(self.collection.find(filter_dict))
    
    def close(self):
        self.client.close()

db = MongoDBWrapper('mongodb://localhost:27017/', 'schule', 'schüler')
print(db.find_all())
db.close()
````
