## Theorie

Eine Entity wird ueber Identitaet definiert, ein Value Object ueber seinen Wert.
Ein Aggregate bildet eine Konsistenzgrenze.

```python
from dataclasses import dataclass

@dataclass(frozen=True)
class EmailAddress:
    value: str

    def __post_init__(self):
        if '@' not in self.value:
            raise ValueError('invalid email')
```

```python
class Student:
    def __init__(self, student_id: int, email: EmailAddress):
        self.student_id = student_id
        self.email = email

    def __repr__(self):
        return f'Student(id={self.student_id}, email={self.email.value})'

s = Student(1, EmailAddress('test@example.com'))
print(s)
```

## Aggregate Beispiel

Ein Course-Aggregate kann eine Invariante enthalten, die eine Kapazitaetsueberschreitung verhindert.

```python
class CourseAggregate:
    def __init__(self, max_seats: int):
        self.max_seats = max_seats
        self.students = []

    def enroll(self, student_id: int):
        if len(self.students) >= self.max_seats:
            raise ValueError('capacity exceeded')
        self.students.append(student_id)

c = CourseAggregate(2)
c.enroll(1)
c.enroll(2)
print(c.students)
```

## Uebungen

- Implementiere eine Money-Klasse als Value Object
- Verhindere doppelte Einschreibungen innerhalb des Aggregates

```python
# Loesungsansatz
class Money:
    def __init__(self, amount: float, currency: str):
        if amount < 0:
            raise ValueError('negative amount')
        self.amount = amount
        self.currency = currency

print(Money(10, 'EUR').currency)
```

## Zusammenfassung

Wenn Entity, Value Object und Aggregate korrekt eingesetzt werden, wird das Domainmodell robuster und verstaendlicher.
