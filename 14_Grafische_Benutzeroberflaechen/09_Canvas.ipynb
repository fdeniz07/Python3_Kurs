# Kapitel 9 – Das Canvas-Widget

Das `Canvas`-Widget wird verwendet, um Grafiken, Formen, Texte und einfache Animationen zu erstellen.


# Lernziele

Nach diesem Kapitel kannst du:

- Ein Canvas erstellen
- Linien und Formen zeichnen
- Texte ausgeben
- Objekte verschieben
- Objekte löschen
- Einfache Animationen programmieren


```python
import tkinter as tk

root = tk.Tk()

canvas = tk.Canvas(root, width=400, height=250, bg="white")
canvas.pack()

root.mainloop()
```

## create_line()

**Beschreibung:** Eine Linie zeichnen

```python
import tkinter as tk

root=tk.Tk()

canvas=tk.Canvas(root,width=350,height=220,bg="white")
canvas.pack()

canvas.create_line(20,20,250,100,width=3,fill="blue")

root.mainloop()
```

## create_rectangle()

**Beschreibung:** Ein Rechteck zeichnen

```python
import tkinter as tk

root=tk.Tk()

canvas=tk.Canvas(root,width=350,height=220,bg="white")
canvas.pack()

canvas.create_rectangle(50,40,180,120,fill="lightblue",outline="black")

root.mainloop()
```

## create_oval()

**Beschreibung:** Einen Kreis oder eine Ellipse zeichnen

```python
import tkinter as tk

root=tk.Tk()

canvas=tk.Canvas(root,width=350,height=220,bg="white")
canvas.pack()

canvas.create_oval(60,50,180,170,fill="yellow")

root.mainloop()
```

## create_arc()

**Beschreibung:** Einen Kreisbogen zeichnen

```python
import tkinter as tk

root=tk.Tk()

canvas=tk.Canvas(root,width=350,height=220,bg="white")
canvas.pack()

canvas.create_arc(50,50,180,180,start=0,extent=120,fill="orange")

root.mainloop()
```

## create_polygon()

**Beschreibung:** Ein Polygon zeichnen

```python
import tkinter as tk

root=tk.Tk()

canvas=tk.Canvas(root,width=350,height=220,bg="white")
canvas.pack()

canvas.create_polygon(60,150,120,50,180,150,fill="green")

root.mainloop()
```

## create_text()

**Beschreibung:** Text auf dem Canvas ausgeben

```python
import tkinter as tk

root=tk.Tk()

canvas=tk.Canvas(root,width=350,height=220,bg="white")
canvas.pack()

canvas.create_text(180,40,text="Hallo Canvas!",font=("Arial",16))

root.mainloop()
```

## Objekte verschieben (`move()`)

```python
import tkinter as tk

root=tk.Tk()

canvas=tk.Canvas(root,width=300,height=200,bg="white")
canvas.pack()

ball=canvas.create_oval(20,80,60,120,fill="red")

def bewegen():
    canvas.move(ball,10,0)

tk.Button(root,text="Bewegen",command=bewegen).pack()

root.mainloop()
```

## Koordinaten ändern (`coords()`)

```python
import tkinter as tk

root=tk.Tk()

canvas=tk.Canvas(root,width=300,height=200)
canvas.pack()

kreis=canvas.create_oval(20,20,80,80,fill="blue")

def verschieben():
    canvas.coords(kreis,120,60,180,120)

tk.Button(root,text="Neue Position",command=verschieben).pack()

root.mainloop()
```

## Objekt löschen (`delete()`)

```python
import tkinter as tk

root=tk.Tk()

canvas=tk.Canvas(root,width=300,height=200)
canvas.pack()

objekt=canvas.create_rectangle(40,40,160,120,fill="green")

tk.Button(root,text="Löschen",command=lambda: canvas.delete(objekt)).pack()

root.mainloop()
```

## Einfache Animation mit `after()`

```python
import tkinter as tk

root=tk.Tk()

canvas=tk.Canvas(root,width=400,height=120,bg="white")
canvas.pack()

ball=canvas.create_oval(10,40,40,70,fill="red")

def animation():
    canvas.move(ball,5,0)
    if canvas.coords(ball)[2] < 400:
        root.after(40,animation)

animation()

root.mainloop()
```
