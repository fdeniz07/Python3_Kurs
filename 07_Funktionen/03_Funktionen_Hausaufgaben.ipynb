## Funktionen - Hausaufgaben (Homework)

#### Funktion zur Berechnung der Anzahl von Vokalen
Funktionsbeschreibung:

Die Funktion nimmt eine Zeichenkette (String) von außen entgegen und gibt die Anzahl der darin enthaltenen Vokale aus.

vokale = ['a', 'e', 'i', 'u', 'o']

#### Funktion zur Berechnung der durchschnittlichen Zeichenanzahl pro Wort
Funktionsbeschreibung:

Die Funktion nimmt einen Satz von außen entgegen und gibt die durchschnittliche Anzahl von Zeichen pro Wort zurück (als `return`).

Die durchschnittliche Zeichenanzahl kann wie folgt berechnet werden: Gesamtzahl der Zeichen / Gesamtzahl der Wörter

#### Funktion zur Berechnung der Anzahl eindeutiger Zeichen

Funktionsbeschreibung:

Die Funktion nimmt eine Zeichenkette (String) von außen entgegen und gibt die Anzahl der eindeutigen (einzigartigen) Zeichen zurück (als `return`).

Eindeutiges Zeichen: Jedes Zeichen wird nur genau einmal gezählt, mehrfach vorkommende Zeichen/Buchstaben werden kein zweites Mal berücksichtigt.

#### Funktion zur Bestimmung der N-t größten und N-t kleinsten Zahl in einer Liste

Funktionsbeschreibung:
Die Funktion nimmt eine Liste und eine Zahl N von außen entgegen. Sie gibt jeweils die N-t größte und N-t kleinste Zahl dieser Liste zurück (als `return`).

Beispiel: Wenn die Liste = [1, 54, 32, 65, 23, 93] und die Zahl N = 2 ist,

sollten die Ergebnisse 65 (die zweitgrößte) und 23 (die zweitkleinste) zurückgegeben werden.


#### Rekursive Funktion zur Berechnung von ggT und kgV (OBEB-OKEK)

Funktionsbeschreibung:

Die Funktion nimmt zwei Zahlen von außen entgegen und berechnet rekursiv die Werte für den größten gemeinsamen Teiler (ggT) und das kleinste gemeinsame Vielfache (kgV). Sie können sich hierfür den Euklidischen Algorithmus (GCD) ansehen.
