## Dateioperationen - Hausaufgaben

#### Dateiinhalt umkehren

**Aufgabenstellung:**
Sie müssen alle Wörter, die Sie aus einer Datei einlesen, in einer zweiten Datei in umgekehrter Reihenfolge speichern.

#### Wortsuche in einer Datei

**Aufgabenstellung:**

Sie müssen überprüfen, ob ein vom Benutzer eingegebenes Wort innerhalb einer bestimmten Datei existiert.

#### Abfrage von in einer Datei gespeicherten Filmdaten

**Aufgabenstellung:**

Überprüfen Sie, ob ein vom Benutzer eingegebener Film in der Datei vorhanden ist. Falls der Film in der Datei gefunden wird, geben Sie den Namen des Films, das Veröffentlichungsdatum und die Bewertung in der Konsole aus. Andernfalls geben Sie die Meldung "Der gesuchte Film ist nicht in der Liste vorhanden!" aus.

**Dateiinhalt-Struktur:** `filmName,filmDatum,bewertung`
