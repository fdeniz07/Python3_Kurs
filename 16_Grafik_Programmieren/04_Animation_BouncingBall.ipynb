# Einfache Animation: Springender Ball
Dieses Beispiel zeigt eine kleine Animation mit `after()` und Bewegung auf dem Canvas.

```python
import tkinter as tk

fenster = tk.Tk()
fenster.title('Springender Ball')

W = 500
H = 350
canvas = tk.Canvas(fenster, width=W, height=H, bg='white')
canvas.pack()

ball = canvas.create_oval(10, 10, 50, 50, fill='red')

vx = 4
vy = 3

def bewegen():
    global vx, vy
    canvas.move(ball, vx, vy)
    x1, y1, x2, y2 = canvas.coords(ball)
    if x1 <= 0 or x2 >= W:
        vx = -vx
    if y1 <= 0 or y2 >= H:
        vy = -vy
    fenster.after(20, bewegen)

bewegen()
fenster.mainloop()
```
