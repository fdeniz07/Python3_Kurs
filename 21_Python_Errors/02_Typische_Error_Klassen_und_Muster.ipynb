# Theorie

Wichtige Klassen:
- `TypeError`
- `ValueError`
- `KeyError`
- `IndexError`
- `AttributeError`
- `FileNotFoundError`

# Erklaerung

Fehlerklassen spiegeln meist eine verletzte Annahme wider:
- falscher Typ
- ungueltiger Wert
- fehlender Schluessel
- ungueltiger Index

# Syntax

```python
try:
    ...
except (TypeError, ValueError) as fehler:
    ...
```

# Merke

Jede Exception-Klasse sollte gezielt behandelt werden, nicht pauschal.

# Parameter

Oft sinnvoll:
- problematischer Eingabewert
- erlaubter Wertebereich

# Rueckgabewert

Definiere klar, ob bei Fehlern Exceptions geworfen oder Ersatzwerte geliefert werden.

````python
# Beispiel 1: TypeError
try:
    print('Alter: ' + 30)
except TypeError as fehler:
    print('TypeError:', fehler)
````

````python
# Beispiel 2: KeyError
person = {'name': 'Ada'}
try:
    print(person['alter'])
except KeyError as fehler:
    print('KeyError:', fehler)
````

````python
# Beispiel 3: FileNotFoundError
try:
    with open('nicht_vorhanden.txt', 'r', encoding='utf-8') as datei:
        print(datei.read())
except FileNotFoundError as fehler:
    print('Dateifehler:', fehler)
````

````python
# Fehlerbehandlung mit Kontext
def parse_int(value):
    try:
        return int(value)
    except ValueError as exc:
        raise ValueError(f"ungueltige Zahl: {value}") from exc
print(parse_int("42"))
````
