# Kapitel 18.8 — ORM mit SQLAlchemy

## Lernziele
- Object-Relational Mapping verstehen.
- SQLAlchemy verwenden.
- Modelle definieren und verwenden.

## Theorie

Ein ORM (Object-Relational Mapping) mappt Datenbank-Tabellen auf Python-Klassen. SQLAlchemy ist das populärste ORM in Python und abstrahiert die Datenbank-Logik.

## Installation

```bash
pip install sqlalchemy
```

````python
from sqlalchemy import create_engine, Column, Integer, String
from sqlalchemy.ext.declarative import declarative_base
from sqlalchemy.orm import sessionmaker

# Engine erstellen
engine = create_engine('sqlite:///schule.db')
Base = declarative_base()

# Modell definieren
class Schueler(Base):
    __tablename__ = 'schüler'
    id = Column(Integer, primary_key=True)
    name = Column(String(100), nullable=False)
    klasse = Column(String(10))

# Tabelle erstellen
Base.metadata.create_all(engine)

# Session erstellen
Session = sessionmaker(bind=engine)
session = Session()
````

## Beispiel 1
Daten einfügen mit ORM.

````python
# Neue Schüler erstellen
s1 = Schueler(name='Anna', klasse='10A')
s2 = Schueler(name='Beni', klasse='10B')
s3 = Schueler(name='Carla', klasse='10A')

# Zur Session hinzufügen
session.add(s1)
session.add(s2)
session.add(s3)

# Änderungen speichern
session.commit()
````

## Beispiel 2
Daten abfragen.

````python
# Alle Schüler
alle_schüler = session.query(Schueler).all()
for s in alle_schüler:
    print(f"{s.id}: {s.name} ({s.klasse})")

# Mit Filter
schüler_10a = session.query(Schueler).filter(Schueler.klasse == '10A').all()
print(schüler_10a)

# Einzelnen Datensatz
anna = session.query(Schueler).filter(Schueler.name == 'Anna').first()
print(anna.name)
````

## Beispiel 3
Daten aktualisieren und löschen.

````python
# Aktualisieren
anna = session.query(Schueler).filter(Schueler.name == 'Anna').first()
anna.klasse = '10B'
session.commit()

# Löschen
s_löschen = session.query(Schueler).filter(Schueler.id == 3).first()
if s_löschen:
    session.delete(s_löschen)
    session.commit()
````

## Praxisbeispiel
Beziehungen zwischen Modellen.

````python
from sqlalchemy import ForeignKey
from sqlalchemy.orm import relationship

class Kurs(Base):
    __tablename__ = 'kurse'
    id = Column(Integer, primary_key=True)
    kursname = Column(String(100), nullable=False)
    schüler = relationship('Schueler', back_populates='kurs')

class Schueler(Base):
    __tablename__ = 'schüler'
    id = Column(Integer, primary_key=True)
    name = Column(String(100), nullable=False)
    kurs_id = Column(Integer, ForeignKey('kurse.id'))
    kurs = relationship('Kurs', back_populates='schüler')

# Daten
kurs = Kurs(kursname='Mathematik')
s = Schueler(name='Anna', kurs=kurs)

session.add(kurs)
session.add(s)
session.commit()
````

## Häufige Fehler
- Modelle nicht correct definieren
- `session.commit()` vergessen
- Foreign Keys nicht richtig setzen

## Best Practice
Nutze ORM für abstrakteren und wartbareren Code.
