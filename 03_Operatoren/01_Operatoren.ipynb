## Was sind Operatoren?

##### Arithmetische Operatoren: Addition, Subtraktion, Multiplikation, Ganzzahldivision, Fließkommadivision, Modulo, Potenzierung
##### Vergleichsoperatoren: ==, !=, >, <, >=, <=
##### Logische Operatoren: AND, OR, NOT
##### Bitweise Operatoren: &, |, ^, ~, <<, >>
##### Zusätzlich: in, not in, is, is not

### Arithmetische Operatoren

```python
a,b=12,5

summe=a+b
subtraktion=a-b
multiplikation=a*b
ganzzahlDivision=a//b
kommaDivision=a/b
modulo=a%b
potenz=a**b

print(summe)
print(subtraktion)
print(multiplikation)
print(ganzzahlDivision)
print(kommaDivision)
print(modulo)
print(potenz)
```

#### Zuweisung und arithmetische Operation gleichzeitig durchführen

```python
c,d=5,2
print(c)
print(d)
c=c+10
c+=10
d*=2
d=d*2
print(c)
print(d)
```

#### Operatorpriorität

Sie entspricht der mathematischen Operatorpriorität. Mit Klammern kann die Auswertungsreihenfolge gezielt festgelegt werden.


```python
a=10/5*3
b=10/(5*3)
print(a)
print(b)
```
