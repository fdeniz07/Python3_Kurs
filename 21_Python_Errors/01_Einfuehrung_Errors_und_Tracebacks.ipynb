# Theorie

Ein Fehler hat meist drei Ebenen:
1. Fehlertyp
2. Ort (Datei, Zeile)
3. Nachricht

Tracebacks zeigen die Aufrufkette bis zur fehlerhaften Stelle.

# Erklaerung

Viele Lernende lesen nur die letzte Zeile. Besser ist:
- zuerst Fehlertyp
- dann letzte relevante eigene Zeile
- danach Ursache pruefen

# Syntax

```python
try:
    ...
except ValueError as fehler:
    print(fehler)
```

# Merke

Fehler sind Lernsignal, kein Scheitern. Gute Entwickler nutzen sie als Diagnosewerkzeug.

# Parameter

Bei Exceptions sind oft relevant:
- Eingabewert
- erwarteter Datentyp
- Kontextinformationen

# Rueckgabewert

Fehlerbehandlung kann Rueckgabewerte ersetzen (z. B. `None`) oder Exceptions weitergeben.

````python
# Beispiel 1: SyntaxError (nur als Textbeispiel)
beispiel = "print('Hallo'"
print('Code mit fehlender Klammer:', beispiel)
````

````python
# Beispiel 2: ValueError
wert = 'abc'
try:
    zahl = int(wert)
except ValueError as fehler:
    print('ValueError erkannt:', fehler)
````

````python
# Beispiel 3: Traceback-Denke
def ebene_c():
    return 10 / 0

def ebene_b():
    return ebene_c()

def ebene_a():
    return ebene_b()

try:
    ebene_a()
except ZeroDivisionError as fehler:
    print('Gefangen:', fehler)
````

````python
# Fehlerbehandlung mit Kontext
def parse_int(value):
    try:
        return int(value)
    except ValueError as exc:
        raise ValueError(f"ungueltige Zahl: {value}") from exc
print(parse_int("42"))
````
