## OS-Modul

Ermöglicht es Ihnen, Informationen über das von Ihnen verwendete Betriebssystem zu erhalten und verschiedene Operationen mit dem Betriebssystem durchzuführen.

```python
import os
```

### Attribut `name`

Gibt den Namen des Betriebssystems zurück;
Für Windows wird 'nt' ausgegeben.
Für macOS und GNU/Linux wird 'posix' ausgegeben.

```python
os.name
```

### Attribut `sep`

Gibt das Trennzeichen für Dateipfade an, das vom jeweiligen Betriebssystem verwendet wird;
Für Windows '\\'
Für macOS und GNU/Linux '/'

```python
os.sep
```

### Attribut `curdir`

Gibt die Zeichenkette zurück, die das aktuelle Verzeichnis darstellt. In den meisten Betriebssystemen wird dies durch '.' ausgedrückt.

```python
os.curdir
```

### Attribut `pardir`

Gibt die Zeichenkette zurück, die das übergeordnete Verzeichnis des aktuellen Verzeichnisses darstellt. In den meisten Betriebssystemen wird dies durch '..' ausgedrückt.

```python
os.pardir
```

### Funktion `getcwd`

Gibt das aktuelle Dateiverzeichnis zurück, in dem sich das Modul befindet.

```python
os.getcwd()
```

### Funktion `chdir`

Ermöglicht den Wechsel von einem Verzeichnis in ein anderes.

```python
os.chdir(r'c:\\PROJEKTE_LOKAL\\BILDUNGS\\VelpTEC\\Python3_Kurs')
```

```python
os.getcwd()
```

### Funktion `listdir`

Listet die Dateien und Ordner innerhalb eines Verzeichnisses in Form einer Liste auf.

```python
os.listdir()
```

```python
for verzeichnis in os.listdir():
    print(verzeichnis)
```

### Funktion `startfile`

Ermöglicht das Öffnen einer Datei oder eines Ordners basierend auf dem angegebenen Dateipfad und Dateinamen.

```python
os.startfile('test.txt')
```

```python
os.startfile('.')
```

### Funktion `mkdir`

Ermöglicht das Erstellen eines neuen Verzeichnisses basierend auf dem angegebenen Dateipfad und Ordnernamen.

```python
os.mkdir(r'c:\\PROJEKTE_LOKAL\\BILDUNGS\\VelpTEC\\Python3_Kurs\\MitOsModulOeffnen')
```

```python
for verzeichnis in os.listdir():
    print(verzeichnis)
```

### Funktion `makedirs`

Ermöglicht das Erstellen von geschachtelten (untergeordneten) neuen Verzeichnissen basierend auf dem angegebenen Dateipfad.

```python
os.makedirs(r'c:\\PROJEKTE_LOKAL\\BILDUNGS\\VelpTEC\\Python3_Kurs\\MitOsModulOeffnen\\MitOsModulOeffnen2')
```

### Funktion `rename`

Ermöglicht es uns, die Namen von Verzeichnissen oder Dateien zu ändern.

```python
os.getcwd()
```

```python
os.rename("test.txt", "neu.txt")
```

### Funktion `remove`

Ermöglicht es uns, Dateien zu löschen.

```python
os.remove("neu.txt")
```

### Funktion `removedirs`

Wird verwendet, um leere Verzeichnisbäume (geschachtelte leere Ordner) zu löschen.

```python
os.removedirs(r'c:\\PROJEKTE_LOKAL\\BILDUNGS\\VelpTEC\\Python3_Kurs\\MitOsModulOeffnen\\MitOsModulOeffnen2\\MitOsModulOeffnen3')
```

### Funktion `path.abspath`

Gibt den absoluten (vollständigen) Pfad der angegebenen Datei zurück.

```python
os.path.abspath("test.txt")
```

### Funktion `path.dirname`

Gibt den Verzeichnisteil (Ordnerpfad) eines angegebenen Dateipfads zurück.

```python
os.path.dirname(os.path.abspath("test.txt"))
```

### Funktion `path.exists`

Überprüft, ob die angegebene Datei oder das Verzeichnis existiert.

```python
os.path.exists(os.path.abspath("test.txt"))
```

```python
os.path.exists(os.path.abspath("test2.txt"))
```

### Funktion `path.isdir`

Überprüft, ob der angegebene Parameter ein Verzeichnis (Ordner) ist.

```python
os.path.isdir("c:\\PROJEKTE_LOKAL\\BILDUNGS\\VelpTEC\\Python3_Kurs")
```

```python
os.path.isdir("c:\\PROJEKTE_LOKAL\\BILDUNGS\\VelpTEC\\Python3_Kurs\\test.txt")
```

### Funktion `path.isfile`

Überprüft, ob der angegebene Parameter eine Datei ist.

```python
os.path.isfile("c:\\PROJEKTE_LOKAL\\BILDUNGS\\VelpTEC\\Python3_Kurs")
```

```python
os.path.isfile("c:\\PROJEKTE_LOKAL\\BILDUNGS\\VelpTEC\\Python3_Kurs\\test.txt")
```

### Funktion `path.join`

Erstellt einen plattformunabhängigen Dateipfad aus den übergebenen Parametern.

```python
os.path.join("uebergeordnetes_verzeichnis", "untergeordnetes_verzeichnis")
```

### Funktion `path.split`

Trennt den letzten Teil des angegebenen Dateipfads ab.

```python
os.path.split("c:\\PROJEKTE_LOKAL\\BILDUNGS\\VelpTEC\\Python3_Kurs\\test.txt")
```

### Funktion `path.splitext`

Trennt den Dateinamen und die Dateiendung (Erweiterung) voneinander.

```python
os.path.splitext("test.txt")
```
