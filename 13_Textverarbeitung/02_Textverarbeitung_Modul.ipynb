## 1. Escape-Sequenzen
Mit dem Backslash `\` können wir spezielle Zeichen darstellen, die sonst im Code schwer zu schreiben wären.

- zur Erstellung von Strings mit ungewöhnlichen Symbolen oder Sonderzeichen
- eine Escape-Sequenz ist eine Folge von Zeichen, die mit einem Backslash \ beginnt
- z.B.: \n - Zeilenumbruch, \" - Anführungszeichen in einem String
- erlaubt Einbau von Schriftzeichen aus anderen Sprachen in einen Text 

Escape-Sequenzen werden verwendet, um nicht druckbare oder spezielle Zeichen in Strings darzustellen.

| Sequenz | Bedeutung |
| :--- | :--- |
| `\\` | Backslash |
| `\n` | Zeilenumbruch |
| `\"` | Anführungszeichen |
| `\N{Name}` | Unicode-Name |
| `\uxxxx` | 4-stelliger Hex-Code |

```python
print("Zeile 1\nZeile 2")        # Zeilenumbruch
print("Das ist ein Backslash: \\") # Backslash
print("Chinesisch: \u99ac")       # Unicode-Hex
```

Mit Escape-Sequenzen kannst du beispielsweise Schriftzeichen aus anderen

Sprachen in einen Text einbauen.

Beispiel:

```python
print('Pferd heißt auf Chinesisch \u99ac.')
```

## 2. String-Methoden
Strings sind unveränderbar (immutable). Jede Methode gibt einen neuen String zurück.
- wichtig! : Strings sind unveränderbare Objekte sind und keine Methode verändert den String 

über die gemeinsamen Operationen der Sequenzen hinaus gibt es noch einige spezielle Operationen für Strings

Hier ist eine Übersicht der wichtigsten Methoden zur Textverarbeitung.

## Übersichtstabelle
| Methode | Erklärung |
| :--- | :--- |
| `s.count(sub)` | Anzahl der Vorkommen von `sub`. |
| `s.endswith(suffix)` | Prüft, ob der String mit `suffix` endet. |
| `s.format(...)` | Der String s enthält Platzhalter mit geschweiften Klammern, z.B. {}. |
| `s.find(sub)` | Sucht die Position von `sub`. |
| `s.join(kollektion)` | Verbindet eine Liste zu einem String. |
| `s.lower()` / `s.upper()` | Case-Konvertierung (Klein/Groß). |
| `s.replace(old, new)` | Ersetzt Teilstrings. |
| `s.split([sep])` | Zerlegt String in eine Liste. |
| `s.startwith(prefix)` | Liefert den Wert True, falls der String mit der Zeichenkette `prefix` beginnt, und sonst `False`. |
| `s.strip()` | Entfernt Whitespace am Rand. |
| `s.isdigit()` | Prüft auf numerische Zeichen. |
| `s.title()` | Formatiert Text als Titel. |
| `s.translate(d)` | Das Argument d ise ein Dictionary mit Zuordnung der Form `nr:zeichenkette`. Die Methode liefert eine Kopie der Strings, in der alle Zeichen mit Unicode-Nummer `nr` durch den String `zeichenkette` ersetz worden sind. |

```python
text = "  Python ist toll!  "
print(f"Strip & Upper: {text.strip().upper()}")
print(f"Replace: {text.replace('toll', 'super')}")
print(f"Split: {'a,b,c'.split(',')}")

print(f"Bereinigt: '{text.strip()}'")
print(f"Groß: '{text.upper()}'")
print(f"Split: {text.strip().split(' ')}")
print(f"Enthält 'mächtig': {'mächtig' in text}")
```

Um sich Klarheit über die genaue Wirkungsweise der String-Methoden zu

verschaffen, kann man in der IDLE-Shell Anweisungen ausprobieren:

```python
'Python'.endswith('on')  # True
```

Die Methode replace() erzeugt einen neuen String, in dem Textteile ersetzt
wurden:

```python
wetter = 'Regenschauer'
wetter.replace('Regen', 'Hagel')  # 'Hagelschauer'
```

Das String-Objekt selbst, dessen Methode aufgerufen worden ist, bleibt
unverändert:

```python
wetter
```

Mit strip() wird jeglicher Leerraum am Anfang und Ende eines Textes entfernt.
Zum Leerraum gehören nicht nur Leerzeichen, sondern auch Zeilenumbrüche.
Im folgenden Beispiel enthält der lange String text vor und hinter dem Wort
PythonD viel Leerraum:

```python
text = '''

       Python

'''
text.strip()  # 'Python'
```

```python
print('Guten\nMorgen')

print('  Guten Morgen \n'.strip())

print('\n  Guten\n  Morgen \n  '.strip())

print('1234'.upper())

print('-'.join(['1', '2', '3']))

d = {ord('ü'): 'ue'}

text = 'üben'
print(text.translate(d))
```
