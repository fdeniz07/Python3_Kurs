## Lernziele

- Profiling als datengetriebenen Prozess anwenden
- CPU- und I/O-gebundene Aufgaben unterscheiden
- Kapazitaetsgrenzen mit Kennzahlen abschaetzen

```python
import time

def heavy_work(n):
    total = 0
    for i in range(n):
        total += i * i
    return total

start = time.perf_counter()
heavy_work(200000)
duration = time.perf_counter() - start
print({'duration_s': round(duration, 4)})
```

## Best Practices

- zuerst messen, dann optimieren
- P95/P99-Latenz statt nur Durchschnitt betrachten
- Lasttests unter realistischen Randbedingungen durchfuehren

## Haeufige Fehler und Debugging-Checkliste

- Sind Annahmen ueber Umgebungswerte oder Testdaten explizit validiert?
- Gibt es reproduzierbare Schritte fuer den Fehlerfall?
- Wurden Logs, Metriken und relevante Konfiguration gemeinsam betrachtet?
- Ist die Fehlerbehebung durch einen neuen Test abgesichert?
- Existiert eine kurze Dokumentation fuer Lessons Learned?

## Pruefungsfragen und Kurzloesungen
