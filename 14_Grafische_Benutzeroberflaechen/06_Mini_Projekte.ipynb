# Projekt: Taschenrechner

## Ziel

Ein einfacher Rechner mit zwei Eingabefeldern.

## Lernziele

- Widgets kombinieren
- Funktionen verwenden
- Ereignisse verarbeiten
- Benutzerinteraktion

## Vollständiges Beispiel


```python
import tkinter as tk

def berechnen():
    a=float(e1.get())
    b=float(e2.get())
    ergebnis.config(text=f"Ergebnis: {a+b}")

root=tk.Tk()
root.title("Taschenrechner")

tk.Label(root,text="Zahl 1").pack()
e1=tk.Entry(root)
e1.pack()

tk.Label(root,text="Zahl 2").pack()
e2=tk.Entry(root)
e2.pack()

tk.Button(root,text="Addieren",command=berechnen).pack(pady=5)

ergebnis=tk.Label(root,text="")
ergebnis.pack()

root.mainloop()
```

### Erweiterungsideen

- Weitere Funktionen hinzufügen
- Fehlerbehandlung verbessern
- Schönere Gestaltung
- ttk-Widgets verwenden

### Übung

Erweitere das Projekt selbst um mindestens eine neue Funktion.


# Projekt: Login

## Ziel

Ein kleines Login-Fenster.

## Lernziele

- Widgets kombinieren
- Funktionen verwenden
- Ereignisse verarbeiten
- Benutzerinteraktion

## Vollständiges Beispiel


```python
import tkinter as tk
from tkinter import messagebox

def anmelden():
    if benutzer.get()=="admin" and passwort.get()=="1234":
        messagebox.showinfo("Login","Anmeldung erfolgreich")
    else:
        messagebox.showerror("Login","Benutzername oder Passwort falsch")

root=tk.Tk()

tk.Label(root,text="Benutzer").pack()
benutzer=tk.Entry(root)
benutzer.pack()

tk.Label(root,text="Passwort").pack()
passwort=tk.Entry(root,show="*")
passwort.pack()

tk.Button(root,text="Login",command=anmelden).pack(pady=5)

root.mainloop()
```

### Erweiterungsideen

- Weitere Funktionen hinzufügen
- Fehlerbehandlung verbessern
- Schönere Gestaltung
- ttk-Widgets verwenden

# Projekt: BMI-Rechner

## Ziel

Berechnet den Body-Mass-Index.

## Lernziele

- Widgets kombinieren
- Funktionen verwenden
- Ereignisse verarbeiten
- Benutzerinteraktion

## Vollständiges Beispiel


```python
import tkinter as tk

def bmi():
    g=float(gewicht.get())
    h=float(groesse.get())/100
    wert=g/(h*h)
    ausgabe.config(text=f"BMI: {wert:.2f}")

root=tk.Tk()

tk.Label(root,text="Gewicht (kg)").pack()
gewicht=tk.Entry(root)
gewicht.pack()

tk.Label(root,text="Größe (cm)").pack()
groesse=tk.Entry(root)
groesse.pack()

tk.Button(root,text="Berechnen",command=bmi).pack()

ausgabe=tk.Label(root)
ausgabe.pack()

root.mainloop()
```

### Erweiterungsideen

- Weitere Funktionen hinzufügen
- Fehlerbehandlung verbessern
- Schönere Gestaltung
- ttk-Widgets verwenden

# Projekt: Passwortgenerator

## Ziel

Erzeugt ein einfaches Zufallspasswort.

## Lernziele

- Widgets kombinieren
- Funktionen verwenden
- Ereignisse verarbeiten
- Benutzerinteraktion

## Vollständiges Beispiel


```python
import tkinter as tk
import random
import string

def erzeugen():
    pw="".join(random.choice(string.ascii_letters+string.digits+string.punctuation) for _ in range(12))
    ausgabe.config(text=pw)

root=tk.Tk()

tk.Button(root,text="Passwort erzeugen",command=erzeugen).pack(pady=10)

ausgabe=tk.Label(root,font=("Arial",14))
ausgabe.pack()

root.mainloop()
```

### Erweiterungsideen

- Weitere Funktionen hinzufügen
- Fehlerbehandlung verbessern
- Schönere Gestaltung
- ttk-Widgets verwenden

# Projekt: Temperaturumrechner

## Ziel

Umrechnung von Celsius in Fahrenheit.

## Lernziele

- Widgets kombinieren
- Funktionen verwenden
- Ereignisse verarbeiten
- Benutzerinteraktion

## Vollständiges Beispiel


```python
import tkinter as tk

def umrechnen():
    c=float(celsius.get())
    f=(c*9/5)+32
    ausgabe.config(text=f"{f:.1f} °F")

root=tk.Tk()

tk.Label(root,text="Celsius").pack()

celsius=tk.Entry(root)
celsius.pack()

tk.Button(root,text="Umrechnen",command=umrechnen).pack()

ausgabe=tk.Label(root)
ausgabe.pack()

root.mainloop()
```

- Fehlerbehandlung verbessern
