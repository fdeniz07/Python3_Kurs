Faustregel:

- CPU-bound: Multiprocessing

# Erklaerung

# Syntax

```python
from multiprocessing import Pool
with Pool() as p:
    result = p.map(func, daten)
```

# Merke

Das schnellste System ist wertlos, wenn es instabil oder unwartbar ist.

````python
# Beispiel 2: Entscheidungsfunktion
def waehle_technik(cpu_bound, viele_io):
    if cpu_bound:
        return 'multiprocessing'
    if viele_io:
        return 'asyncio'
    return 'threading'

print(waehle_technik(cpu_bound=True, viele_io=False))
print(waehle_technik(cpu_bound=False, viele_io=True))
````

````python
# asyncio Timeout-Guard
import asyncio
async def slow_job():
    await asyncio.sleep(2)
    return "done"
async def main():
    try:
        print(await asyncio.wait_for(slow_job(), timeout=1.0))
    except asyncio.TimeoutError:
        print("timeout")
asyncio.run(main())
````
