## DateTime-Modul

Das `datetime`-Modul besteht aus Klassen, die Funktionen und Attribute bereitstellen, mit denen wir Operationen bezüglich Zeit, Stunde und Datum durchführen können.

Innerhalb des `datetime`-Moduls befinden sich die Klassen `date`, `time` und `datetime`.

```python
from datetime import datetime
```

### Die Funktion `now`

Gibt die aktuellen Informationen zu Datum, Uhrzeit und Zeit an.

```python
aktuelle_zeit = datetime.now()
aktuelle_zeit
```

```python
print(aktuelle_zeit.year)
print(aktuelle_zeit.month)
print(aktuelle_zeit.day)
print(aktuelle_zeit.hour)
print(aktuelle_zeit.minute)
print(aktuelle_zeit.second)
print(aktuelle_zeit.microsecond)
```

### Die Funktion `today`

Gibt Informationen über den heutigen Tag an.

```python
heute = datetime.today()
heute
```

```python
print(heute.year)
print(heute.month)
print(heute.day)
print(heute.hour)
print(heute.minute)
print(heute.second)
print(heute.microsecond)
```

### Die Funktion `ctime`

Gibt die aktuellen Datums- und Uhrzeitinformationen in einem lesbaren Format aus.


```python
aktuelle_zeit = datetime.now()
ctime_aktuelle_zeit = datetime.ctime(aktuelle_zeit)
ctime_aktuelle_zeit
```

### Die Funktion `strftime`

Ermöglicht es uns, Datums- und Uhrzeitinformationen zu formatieren:

1. Um die Jahresinformation zu erhalten: `%Y`

2. Für die letzten zwei Ziffern des Jahres: `%y`

3. Um den Monatsnamen zu erhalten: `%B`

4. Um den Wochentagsnamen zu erhalten: `%A`

5. Um die Uhrzeitinformation zu erhalten: `%X`

6. Um das Datum als MM/DD/YY zu erhalten: `%D`

7. Für das vollständige Datum und die Uhrzeit: `%c`


```python
aktuelle_zeit = datetime.now()
strftime_aktuelle_zeit = datetime.strftime(aktuelle_zeit, "%c")
strftime_aktuelle_zeit
```

```python
aktuelle_zeit = datetime.now()
strftime_aktuelle_zeit = datetime.strftime(aktuelle_zeit, "%D")
strftime_aktuelle_zeit
```

### Die Funktion `strptime`

Ermöglicht es uns, Datumsinformationen, die als String vorliegen, zu parsen und zu trennen.

```python
aktuelle_zeit_str = "4. September 2021 13:42:00"
datetime_aktuelle_zeit = datetime.strptime(aktuelle_zeit_str, '%d. %B %Y %H:%M:%S')
datetime_aktuelle_zeit
```

### Die Funktion `timestamp`

Wird verwendet, um ein `datetime`-Objekt in Sekunden (Unix-Zeitstempel) umzuwandeln.

```python
aktuelle_zeit = datetime.now()
sekunden = datetime.timestamp(aktuelle_zeit)
sekunden
```

### Die Funktion `fromtimestamp`

Wird verwendet, um einen Ausdruck in Sekunden wieder in ein `datetime`-Objekt umzuwandeln.

```python
aktuelle_zeit_fts = datetime.fromtimestamp(sekunden)
aktuelle_zeit_fts
```
