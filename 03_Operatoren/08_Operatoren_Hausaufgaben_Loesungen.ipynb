### 1. Frage: Berechnung des Delta-Werts (Diskriminante)

Fragenbeschreibung:

Fordern Sie den Benutzer auf, die Werte für a, b und c einzugeben, und berechnen Sie den Delta-Wert nach der folgenden Formel. Geben Sie anschließend True oder False auf der Konsole aus, je nachdem, ob der Delta-Wert größer oder kleiner als Null ist.

Delta-Berechnungsformel: b^2 - 4 * a * c


```python
a, b, c = map(int, input("Bitte geben Sie die Werte für a, b und c der Reihe nach ein (mit Komma getrennt):").split(","))
deltaWert = b**2 - 4 * a * c
print("Delta-Wert: {}".format(deltaWert))
print("Status (Delta > 0):", deltaWert > 0)
```

### 2. Frage: Durchschnittsberechnung mit Klausur- und Abschlussnote
Fragenbeschreibung:

Fordern Sie den Benutzer auf, die Klausurnote (Vornote) und die Abschlussnote einzugeben, und berechnen Sie den Kursdurchschnitt basierend auf den folgenden prozentualen Gewichtungen. Geben Sie anschließend True auf der Konsole aus, wenn der Kursdurchschnitt größer als 60 ist, andernfalls False.

Prozentuale Gewichtung für die Berechnung: Abschlussnote -> 60%, Klausurnote -> 40%

```python
klausurNote = int(input("Bitte geben Sie Ihre Klausurnote ein:"))
abschlussNote = int(input("Bitte geben Sie Ihre Abschlussnote ein:"))

kursDurchschnitt = klausurNote * 0.4 + abschlussNote * 0.6
print("Ihr Kursdurchschnitt: {} und Status für das Bestehen des Kurses: {}".format(kursDurchschnitt, kursDurchschnitt > 60))
```

### 3. Frage: Berechnung von Kreisfläche und Kreisumfang
Fragenbeschreibung:

Fordern Sie den Benutzer auf, den Radiuswert einzugeben, und berechnen Sie den Umfang sowie die Fläche des Kreises nach den folgenden Formeln.

Kreisflächenformel: pi * r^2

Kreisumfangsformel: 2 * pi * r

```python
PI = 3.1415926535898
radius = float(input("Bitte geben Sie den Radiuswert ein:"))

kreisFlaeche = PI * radius**2
kreisUmfang = 2 * PI * radius

print("Basierend auf dem eingegebenen Radius betraegt der Kreisumfang: {} Einheiten und die Kreisflaeche: {} Einheiten^2".format(kreisUmfang, kreisFlaeche))
```
