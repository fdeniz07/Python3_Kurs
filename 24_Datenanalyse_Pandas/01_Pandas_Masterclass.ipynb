# Rueckgabewert

Pandas-Operationen liefern oft neue DataFrames; Inplace-Aenderungen sparsam und bewusst nutzen.

````python
import pandas as pd

# Beispiel 1
df = pd.DataFrame({
    'kurs': ['Python', 'WSGI', 'Python', 'Pandas'],
    'teilnehmer': [20, 12, 18, 15],
    'bewertung': [4.7, 4.5, 4.8, 4.6]
})
print(df.head())
````

````python
# Beispiel 2
print(df.groupby('kurs').agg({'teilnehmer': 'sum', 'bewertung': 'mean'}))
````

````python
# Beispiel 3: Fehlende Werte
df2 = pd.DataFrame({'a': [1, None, 3], 'b': [None, 2, 3]})
print(df2.isna().sum())
print(df2.fillna(0))
````

# Praxisbeispiel

Analysiere Kursdaten: Top-Kurs nach Teilnehmern, Durchschnittsbewertung je Kategorie, Datenbereinigung bei Missing Values.

# Haeufige Fehler

1. Datentypen nicht pruefen.
2. Missing Values ignorieren.
3. Gruppierung ohne fachliche Plausibilitaet.

- Vor Analyse stets `info()` und `describe()` nutzen.

# Uebung

Erstelle einen DataFrame mit Verkaufsdaten und berechne Umsatz je Produktgruppe.
