#### Inhalt aus einer Datei einlesen und in Großbuchstaben in eine andere Datei übertragen

**Aufgabenstellung:**

Sie müssen alle Buchstaben, die Sie aus einer Datei lesen, in Großbuchstaben umwandeln und in eine zweite Datei übertragen.

```python
datei_lese = open("grossbuchstaben_lesen.txt", "r")
datei_schreibe = open("grossbuchstaben_schreiben.txt", "w")

datei_inhalt = datei_lese.read()
datei_inhalt = datei_inhalt.upper()
datei_schreibe.write(datei_inhalt)

datei_lese.close()
datei_schreibe.close()
```

#### Den ersten Buchstaben von Sätzen in einer Datei groß schreiben und in eine andere Datei übertragen

**Aufgabenstellung:**

Sie müssen alle Sätze aus einer Datei einlesen, den jeweils ersten Buchstaben jedes Satzes in einen Großbuchstaben umwandeln und das Ergebnis in eine andere Datei schreiben.

```python
satz_lese = open("saetze_lesen.txt", "r", encoding="utf-8")
satz_schreibe = open("saetze_schreiben.txt", "w", encoding="utf-8")
datei_inhalt = satz_lese.read()
saetze_liste = datei_inhalt.strip().split(".")
print(saetze_liste)
# Trennen am Punkt
for satz in saetze_liste:
    # Falls durch das letzte Zeichen ein leerer Eintrag entsteht, überspringen
    if not satz.strip():
        continue
    
    # 1. strip() entfernt die Leerzeichen am Anfang und Ende des Satzes
    # 2. capitalize() macht den ersten Buchstaben groß
    formatierter_satz = satz.strip().capitalize()
    
    # Satzzeichen wieder hinzufügen und mit einem sauberen Leerzeichen in die Datei schreiben
    satz_schreibe.write(formatierter_satz + ". ")

satz_lese.close()
satz_schreibe.close()
```

#### Studenten-Notensystem

**Aufgabenstellung:**

In der Datei, die Sie einlesen, befinden sich Vorname, Nachname, die Note der ersten Klausur (Midterm) und die Note der Abschlussklausur (Final) jedes Studenten. Sie müssen diese Daten auswerten, die Gesamtnote berechnen, die entsprechende Buchstabennote (Notenschlüssel) ermitteln und den Erfolgsstatus (Bestanden/Durchgefallen) in einer separaten Ergebnisdatei speichern.

```python
studenten_datei = open("studenten_noten.txt", "r")
ergebnis_datei = open("studenten_ergebnisse.txt", "w")
studenten_daten = studenten_datei.readlines()
print(studenten_daten)
for student in studenten_daten:
    # Berechnungsschritte und Schreiben in die Ergebnisdatei
    vorname,nachname,erste_note,zweite_note = student.split(",")
    erste_note, zweite_note = int(erste_note), int(zweite_note)
    durchschnitt = (erste_note * 0.4 + zweite_note * 0.6) 
    notenbewertung = str()
    notenErfolg= str()
    if durchschnitt >= 90:
        notenbewertung = "A"
        notenErfolg = "Bestanden"
    elif durchschnitt >= 80:
        notenbewertung = "B"
        notenErfolg = "Bestanden"
    elif durchschnitt >= 70:
        notenbewertung = "C"
        notenErfolg = "Bestanden"
    elif durchschnitt >= 60:
        notenbewertung = "D"
        notenErfolg = "Bestanden"
    else:
        notenbewertung = "F"
        notenErfolg = "Nicht Bestanden" 
    einzugebende_daten = (
        f"{f'{vorname} {nachname}':<20} | "
        f"Erste Note: {erste_note:<3} | "
        f"Zweite Note: {zweite_note:<3} | "
        f"Durchschnitt: {durchschnitt:<5.2f} | "
        f"Note: {notenbewertung:<2} | "
        f"Ergebnis: {notenErfolg}\n"
    )
    ergebnis_datei.write(einzugebende_daten)
studenten_datei.close()
ergebnis_datei.close()
```
