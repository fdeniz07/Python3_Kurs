## Haeufige Fehler und Debugging-Checkliste

- Ist das Problem reproduzierbar beschrieben (Input, Zustand, Erwartung)?
- Gibt es messbare Signale (Logs, Tests, Metriken), die die Hypothese stuetzen?
- Wurde die Korrektur durch einen neuen Test oder Guardrail abgesichert?
- Sind Seiteneffekte fuer Betrieb und Wartung bewertet?
- Ist die Dokumentation nach dem Fix aktualisiert?
