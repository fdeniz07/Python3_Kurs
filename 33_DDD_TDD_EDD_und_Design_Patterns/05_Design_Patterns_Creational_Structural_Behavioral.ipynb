Wir betrachten die am haeufigsten verwendeten Pattern-Familien anhand von Python-Beispielen.

## Lernziele

- wissen, welches Pattern zu welchem Problem passt

- Pattern-Unterschiede mit einfachen Python-Beispielen erkennen

```python
# Factory-Method-Minibeispiel
class JsonExporter:
    def export(self, data):
        return f'json:{data}'

class CsvExporter:
    def export(self, data):
        return f'csv:{data}'

def exporter_factory(kind):
    return JsonExporter() if kind == 'json' else CsvExporter()

print(exporter_factory('json').export({'a': 1}))
```

```python
# Adapter-Minibeispiel
class LegacyNotifier:
    def send_sms(self, text):
        return f'SMS:{text}'

class NotifierAdapter:
    def __init__(self, legacy):
        self.legacy = legacy

    def send(self, message):
        return self.legacy.send_sms(message)

print(NotifierAdapter(LegacyNotifier()).send('hello'))
```

```python
# Strategy-Minibeispiel
def normal(price):
    return price

def student(price):
    return price * 0.8

class PricingContext:
    def __init__(self, strategy):
        self.strategy = strategy

    def total(self, price):
        return self.strategy(price)

print(PricingContext(student).total(100))
```

## Leitfaden zur Pattern-Auswahl

- Ist die Objekterzeugung komplex? -> Creational
- Gibt es Schnittstelleninkompatibilitaet? -> Structural
- Aendert sich Verhalten zur Laufzeit? -> Behavioral

## Uebung

Schreibe ein kombiniertes Preisbeispiel mit Factory und Strategy.

```python
# Loesungsskizze
def strategy_factory(kind):
    return student if kind == 'student' else normal

ctx = PricingContext(strategy_factory('student'))
print(ctx.total(250))
```

## Zusammenfassung

Patterns sind Werkzeuge, kein Selbstzweck. Ohne klares Problem ist ein Pattern oft unnoetig.

## Technischer Tiefgang

Begriffe, Risiken, Betriebsaspekte und Qualitaetskriterien werden explizit gemacht, damit Entscheidungen reproduzierbar bleiben.

## Haeufige Fehler und Debugging-Checkliste

- Sind Eingaben, Konfiguration und Randbedingungen explizit validiert?
- Sind reproduzierbare Schritte fuer den Fehler dokumentiert?
- Wurden Logs, Metriken und Tests gemeinsam ausgewertet?
- Ist die Korrektur durch einen neuen Test dauerhaft abgesichert?
- Gibt es eine kurze Lessons-Learned-Notiz fuer das Team?
