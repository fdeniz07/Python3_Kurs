# Aufgabe: Module
Thema: `random`-Modul, `math`-Modul, eigene Funktionen, if-else und for-Schleife

## a) Module importieren

**Aufgabe:** Importiere das Modul `random` und das Modul `math`.

```python
import random
import math
```

## b) Zufällige Ganzzahl speichern

**Aufgabe:** Definiere eine Variable `zahl`, die eine zufällige Ganzzahl zwischen 1 und 100 speichert.

```python
zahl = random.randint(1, 100)
print("Zufallszahl:", zahl)
```

## c) Funktion `quadratwurzel` definieren

**Aufgabe:** Schreibe eine Funktion `quadratwurzel`, die die Quadratwurzel einer Zahl berechnet und zurückgibt. Verwende dafür eine Funktion aus dem Modul `math`.

```python
def quadratwurzel(n):
    return math.sqrt(n)
```

## d) if-else-Kontrollstruktur

**Aufgabe:** Überprüfe, ob `zahl` größer als 50 ist. Wenn ja, rufe `quadratwurzel` mit `zahl` auf und drucke das Ergebnis aus. Andernfalls drucke `"Zahl ist 50 oder kleiner"`.

```python
if zahl > 50:
    ergebnis = quadratwurzel(zahl)
    print("Quadratwurzel von", zahl, "ist", ergebnis)
else:
    print("Zahl ist 50 oder kleiner")
```

## e) for-Schleife mit `quadratwurzel`

**Aufgabe:** Erstelle eine `for`-Schleife, die Zahlen von 1 bis 5 durchläuft, und für jede Zahl die Funktion `quadratwurzel` aufruft und das Ergebnis ausdruckt.

```python
for i in range(1, 6):
    print("Quadratwurzel von", i, "ist", quadratwurzel(i))
```

## Komplettes Programm (alle Teile zusammen)

```python
import random
import math

# b) Zufällige Ganzzahl zwischen 1 und 100
zahl = random.randint(1, 100)
print("Zufallszahl:", zahl)

# c) Funktion zur Berechnung der Quadratwurzel
def quadratwurzel(n):
    return math.sqrt(n)

# d) if-else: Überprüfung ob zahl > 50
if zahl > 50:
    ergebnis = quadratwurzel(zahl)
    print("Quadratwurzel von", zahl, "ist", ergebnis)
else:
    print("Zahl ist 50 oder kleiner")

# e) for-Schleife: Quadratwurzel für Zahlen 1 bis 5
for i in range(1, 6):
    print("Quadratwurzel von", i, "ist", quadratwurzel(i))
```
