# Textverarbeitung in Python

Texte sind die Grundlage unserer Kommunikation. In diesem Kapitel lernen wir, wie Python mit Schriftzeichen umgeht und wie wir diese effizient verarbeiten können.

## 1. Unicode und die Identifikation von Zeichen
- Jedes Zeichen in einem String ist Teil des **Unicode-Standards**. Dieser weist jedem Zeichen eine eindeutige Nummer zu.
- Unicode ist ein internationaler Standard für Schriftzeichen und Symbole
- Unicode ordnet jedem sinntragenden Zeichen eine Nummer und einen eindeutigen Namen zu
- die Nummern werden von den Standardfunktionen ord() und chr() verwendet
- Unicode-Nummern werden meist angegeben als Hexadezimalzahlen (Zahl 16 als Zahlenbasis und die 16 Ziffern 0123456789ABCDEF)

### Die Funktionen `ord()` und `chr()`
- `ord(c)`: Gibt die Unicode-Nummer eines Zeichens `c` zurück.
- `chr(n)`: Gibt das Zeichen zu einer Unicode-Nummer `n` zurück.

### Beispiel: Zeichen-Konvertierung

````python
# Umwandlung in Dezimalzahlen
for buchstabe in 'Abend':
    print(ord(buchstabe), end=' ')

# Umwandlung von Zahl zu Zeichen
print(f"\nDas Zeichen für 65 ist: {chr(65)}")
````

## 2. Hexadezimalzahlen und Unicode
Unicode-Nummern werden häufig hexadezimal (Basis 16) angegeben. In Python erkennt man sie am Präfix `0x`.

Beispiel: `0xFF` entspricht der Dezimalzahl 255.

## 3. Codierung: utf-8
Wenn wir Dateien lesen, müssen wir das `encoding` angeben. `utf-8` ist der Standard für das moderne Web.

```python
# Beispiel für das korrekte Öffnen einer Datei
with open('faust.txt', mode='r', encoding='utf-8') as f:
    inhalt = f.read()
```

## 4. Analyse: Stringvergleiche
**Frage:** Warum ist `'Fisch' < 'angeln'` `True`?

**Erklärung:** In Python werden Strings anhand der Unicode-Werte der Zeichen verglichen (lexikographisch). Großbuchstaben haben im Unicode-Standard (ASCII-kompatibel) niedrigere Nummern als Kleinbuchstaben. Da 'F' (66) kleiner ist als 'a' (97), ist der Vergleich `True`.
