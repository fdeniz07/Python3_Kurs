## Schleifen
Schleifen sind Kontrollstrukturen, die einen Codeblock wiederholt ausführen, solange eine bestimmte Bedingung erfüllt ist.
1. While-Schleife
2. For-Schleife

#### For-Schleife

### Wann verwenden wir sie?

Die For-Schleife wird verwendet, wenn ein Vorgang eine bestimmte Anzahl von Wiederholungen benötigt oder wenn über Elemente von Datenstrukturen wie Listen oder Zeichenketten iteriert werden soll.

### Syntax:

for variable in datensammlung:

    anweisungen_im_schleifenblock

### Erklärung:

Die For-Schleife ermöglicht es, die Elemente einer Datensammlung nacheinander zu durchlaufen. Sobald alle Elemente verarbeitet wurden, endet die Schleife automatisch.


##### Iteration über einen String mit einer Schleife

```python
text="maxmustermann"
for zeichen in text:
    print(zeichen,end=" ")
```

##### Iteration über eine Liste mit einer Schleife

```python
element_liste=[1,2,3,4,5]
for element in element_liste:
    print(element,end=" ")
print("Ausführung wird fortgesetzt")
```

#### Die Funktion range

###### Wofür wird sie verwendet?

Erzeugt Zahlenfolgen anhand der übergebenen Argumente.

##### Verwendung

1. range(endwert): Erzeugt Zahlen von 0 bis endwert-1.
2. range(startwert, endwert): Erzeugt Zahlen von startwert bis endwert-1.
3. range(startwert, endwert, schrittweite): Erzeugt Zahlen von startwert bis endwert-1 mit der angegebenen Schrittweite.


```python
for zahl in range(100):
    print(zahl,end=" ")
```

```python
for zahl in range(50,100):
    print(zahl,end=" ")
```

```python
for zahl in range(10,200,5):
    print(zahl,end=" ")
```
