# pack()

## Beschreibung

Ordnet Widgets automatisch untereinander oder nebeneinander an.

## Beispiel


```python
import tkinter as tk

root = tk.Tk()

tk.Label(root, text="Name").pack()
tk.Entry(root).pack()
tk.Button(root, text="Speichern").pack()

root.mainloop()
```

# grid()

## Beschreibung

Ordnet Widgets in Zeilen und Spalten an.

## Beispiel


```python
import tkinter as tk

root = tk.Tk()

tk.Label(root, text="Name").grid(row=0, column=0, padx=5, pady=5)
tk.Entry(root).grid(row=0, column=1)

tk.Label(root, text="Alter").grid(row=1, column=0)
tk.Entry(root).grid(row=1, column=1)

root.mainloop()
```

# place()

## Beschreibung

Positioniert Widgets mit festen Koordinaten.

## Beispiel


```python
import tkinter as tk

root = tk.Tk()
root.geometry("300x200")

tk.Button(root, text="Button").place(x=80, y=60)

root.mainloop()
```

# Vergleich der Layout-Manager

| Layout | Vorteile | Nachteile |
|---------|----------|-----------|
| pack() | Einfach | Wenig flexibel |
| grid() | Sehr übersichtlich | Nicht mit pack() mischen |
| place() | Exakte Position | Wenig flexibel bei Größenänderungen |

**Tipp:** Für Formulare wird meistens `grid()` verwendet.


# Events mit bind()

Mit `bind()` kann auf Maus- und Tastaturereignisse reagiert werden.


```python
import tkinter as tk

def geklickt(event):
    print("Die Maus wurde geklickt.")

root = tk.Tk()

label = tk.Label(root, text="Hier klicken", bg="lightblue")
label.pack(padx=20, pady=20)

label.bind("<Button-1>", geklickt)

root.mainloop()
```

# Häufig verwendete Events

| Event | Bedeutung |
|-------|-----------|
| `<Button-1>` | Linke Maustaste |
| `<Double-Button-1>` | Doppelklick |
| `<Enter>` | Maus betritt Widget |
| `<Leave>` | Maus verlässt Widget |
| `<Return>` | Enter-Taste |
| `<Escape>` | Escape-Taste |
| `<Key>` | Beliebige Taste |


```python
import tkinter as tk

def taste(event):
    print("Taste:", event.keysym)

root = tk.Tk()

root.bind("<Key>", taste)

root.mainloop()
```

# StringVar

Speichert Zeichenketten.

```python
import tkinter as tk

root=tk.Tk()

name=tk.StringVar()

tk.Entry(root,textvariable=name).pack()

def anzeigen():
    print(name.get())

tk.Button(root,text="Ausgeben",command=anzeigen).pack()

root.mainloop()
```

# IntVar

Speichert Ganzzahlen.

```python
import tkinter as tk

root=tk.Tk()

zahl=tk.IntVar(value=10)

tk.Label(root,textvariable=zahl).pack()

root.mainloop()
```

# BooleanVar

Speichert True oder False.

```python
import tkinter as tk

root=tk.Tk()

wert=tk.BooleanVar()

tk.Checkbutton(root,text="Aktiv",variable=wert).pack()

root.mainloop()
```

# DoubleVar

Speichert Fließkommazahlen.

```python
import tkinter as tk

root=tk.Tk()

zahl=tk.DoubleVar(value=3.14)

tk.Label(root,textvariable=zahl).pack()

root.mainloop()
```

# after()

Mit `after()` kann eine Funktion zeitverzögert ausgeführt werden.


```python
import tkinter as tk

root=tk.Tk()

def hallo():
    print("Hallo nach 3 Sekunden")

root.after(3000,hallo)

root.mainloop()
```

# focus()

Mit `focus()` erhält ein Widget den Tastaturfokus.


```python
import tkinter as tk

root=tk.Tk()

entry=tk.Entry(root)
entry.pack()

entry.focus()

root.mainloop()
```

# destroy()

Schließt ein Fenster.


```python
import tkinter as tk

root=tk.Tk()

tk.Button(root,text="Schließen",command=root.destroy).pack()

root.mainloop()
```
