````python
# Beispiel 1: Zielcodedef addiere(a, b):    return a + b
````

````python
# Beispiel 2: unittest-Grundstrukturimport unittestclass TestAddiere(unittest.TestCase):    def test_positive(self):        self.assertEqual(addiere(2, 3), 5)    def test_null(self):        self.assertEqual(addiere(0, 0), 0)suite = unittest.defaultTestLoader.loadTestsFromTestCase(TestAddiere)result = unittest.TextTestRunner(verbosity=2).run(suite)print('Fehler:', len(result.errors), 'Failures:', len(result.failures))
````

````python
# Beispiel 3: setUp und mehrere Assertionsclass TestMitSetup(unittest.TestCase):    def setUp(self):        self.zahlen = [1, 2, 3]    def test_summe(self):        self.assertEqual(sum(self.zahlen), 6)    def test_laenge(self):        self.assertEqual(len(self.zahlen), 3)suite2 = unittest.defaultTestLoader.loadTestsFromTestCase(TestMitSetup)unittest.TextTestRunner(verbosity=2).run(suite2)
````

# PraxisbeispielWende die Teststrategie auf ein eigenes Modul an und dokumentiere Testfaelle fuer Happy Path und Edge Cases.

# Haeufige Fehler1. Zu grosse Tests mit mehreren Verantwortlichkeiten.2. Abhaengigkeit von Zufall/Uhrzeit/Netzwerk ohne Kontrolle.3. Nur Happy Path testen.

# Best Practice- Testdaten klar benennen.- Pro Test eine klare Erwartung.- Fehler reproduzierbar halten.

# TippSchreibe zuerst den Testnamen als fachlichen Satz, danach den Testcode.

# UebungErstelle mindestens 5 Tests: 3 Erfolgsfaelle, 2 Fehlerfaelle, inklusive Randwerten.

````python
# Loesung: Erweiterter TestCaseclass TestAddiereErweitert(unittest.TestCase):    def test_negative(self):        self.assertEqual(addiere(-2, -3), -5)    def test_mixed(self):        self.assertEqual(addiere(-2, 5), 3)suite3 = unittest.defaultTestLoader.loadTestsFromTestCase(TestAddiereErweitert)unittest.TextTestRunner(verbosity=2).run(suite3)
````
