# Theorie

Eigene Exceptions verbessern Lesbarkeit und Domain-Verstaendnis.
Statt allgemeiner `ValueError` kann z. B. `KursNichtGefundenError` genutzt werden.

# Erklaerung

Eine sinnvolle Fehlerarchitektur hat:
- gemeinsame Basisklasse
- spezialisierte Subklassen
- klare Mapping-Regeln zu HTTP/GUI-Fehlerausgaben

# Syntax

```python
class AppError(Exception):
    pass

class KursNichtGefundenError(AppError):
    pass
```

# Merke

Eigene Exceptions sind besonders wertvoll in groesseren Codebasen.

# Parameter

Exception-Klassen koennen Kontext aufnehmen, z. B. `kurs_id`, `user_id`.

# Rueckgabewert

Exceptions liefern keinen klassischen Rueckgabewert, aber transportieren strukturierte Fehlerinfo.

````python
# Beispiel 1: Eigene Klassen
class AppError(Exception):
    """Basisklasse fuer Anwendungsfehler."""

class ValidierungsError(AppError):
    pass

class KursNichtGefundenError(AppError):
    pass
````

````python
# Beispiel 2: Fachliche Fehler werfen
KURSE = {1: 'Python', 2: 'WSGI'}

def hole_kurs(kurs_id):
    if kurs_id not in KURSE:
        raise KursNichtGefundenError(f'Kurs {kurs_id} existiert nicht')
    return KURSE[kurs_id]

try:
    print(hole_kurs(99))
except KursNichtGefundenError as fehler:
    print('Fachlicher Fehler:', fehler)
````

````python
# Beispiel 3: Mapping in API-Fehlerobjekt
def mappe_fehler_zu_response(fehler):
    if isinstance(fehler, KursNichtGefundenError):
        return {'status': 404, 'code': 'KURS_NICHT_GEFUNDEN', 'nachricht': str(fehler)}
    if isinstance(fehler, ValidierungsError):
        return {'status': 400, 'code': 'VALIDIERUNG', 'nachricht': str(fehler)}
    return {'status': 500, 'code': 'INTERN', 'nachricht': 'Interner Fehler'}

print(mappe_fehler_zu_response(KursNichtGefundenError('x')))
````

# Praxisbeispiel

Definiere fuer dein Kursportal eine Error-Hierarchie mit mindestens 4 spezifischen Klassen.

# Haeufige Fehler

1. Zu viele unstrukturierte Exception-Klassen.
2. Keine gemeinsame Basisklasse.
3. Fehlerklassen ohne klaren Einsatzbereich.

# Best Practice

- Kleine, klare Fehlerhierarchie.
- Konsistentes Mapping auf API/GUI-Ausgaben.

# Tipp

Benenne Exceptions fachlich, nicht technisch (z. B. `AnmeldungGeschlossenError`).

# Uebung

Implementiere `AnmeldungError`, `KapazitaetVollError`, `EingabeFormatError` und mappe sie auf Statuscodes.

# Loesung

Empfehlung: `AnmeldungError` als Basisklasse, darunter die Spezialisierungen mit zentraler Mapping-Funktion.

````python
# Fehlerbehandlung mit Kontext
def parse_int(value):
    try:
        return int(value)
    except ValueError as exc:
        raise ValueError(f"ungueltige Zahl: {value}") from exc
print(parse_int("42"))
````
