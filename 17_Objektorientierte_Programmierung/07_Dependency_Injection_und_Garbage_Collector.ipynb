# Kapitel 17.7 — Dependency Injection und Garbage Collector

## Lernziele
- Dependency Injection verstehen.
- Constructor Injection erklären.
- Den Unterschied zwischen manueller Objektverwaltung und Garbage Collector verstehen.

## Theorie
Dependency Injection bedeutet, dass Abhängigkeiten von außen an ein Objekt übergeben werden.
Das macht Code flexibler und einfacher testbar.
In Python gibt es kein Spring-Framework wie in Java, aber das Prinzip lässt sich sehr gut anwenden.

## Erklärung
- Constructor Injection: Abhängigkeit wird im Konstruktor übergeben
- Setter Injection: Abhängigkeit wird später gesetzt
- Interface/Protocol Injection: über ein gemeinsames Protokoll

In Python entspricht das oft einem sehr einfachen Muster: Die Klasse bekommt ihre Abhängigkeit von außen, statt sie selbst zu erzeugen.
Das ist der Python-Weg für etwas, das in Java oft mit Spring, `@Autowired` oder einem Container beschrieben wird.

Auch `dependency-injector` oder `injector` können in Python verwendet werden, wenn man einen kleinen Container oder eine zentrale Abhängigkeitsverwaltung möchte.
Typische Bausteine sind `providers.Singleton()` für einzelne Instanzen und `providers.Factory()` für neue Objekte bei jedem Bedarf.

```python
class DBService:
    def save(self, data):
        print(f"Speichere {data}")

class NutzerService:
    def __init__(self, db_service):
        self.db_service = db_service

    def erstellen(self, name):
        self.db_service.save(name)

service = NutzerService(DBService())
service.erstellen("Mina")
```

## Beispiel 1
Constructor Injection ist eine sehr häufige Form in Python-Projekten.

```python
class Logger:
    def __init__(self):
        self.messages = []

    def log(self, text):
        self.messages.append(text)

class App:
    def __init__(self, logger):
        self.logger = logger

    def run(self):
        self.logger.log("App gestartet")

app = App(Logger())
app.run()
```

## Beispiel 2
Die Klasse `App` ist nicht direkt an eine konkrete Logger-Implementierung gekoppelt.

```python
import gc

class Beispiel:
    pass

obj = Beispiel()
del obj
print(gc.collect())
```

## Praxisbeispiel
Python verwendet einen Garbage Collector, um nicht mehr benötigte Objekte automatisch freizugeben.

## Häufige Fehler
- Zu viele globale Abhängigkeiten
- Unklare Objektlebenszyklen
- Zu starke Kopplung zwischen Klassen

## Best Practice
Nutze Dependency Injection, wenn ein Objekt flexibel bleiben soll. Beachte aber auch den Lebenszyklus deiner Objekte.

## Übung
Erstelle eine Klasse `BenutzerService`, die ein `Repository` über den Konstruktor erhält.

```python
class Repository:
    def speichern(self, daten):
        print(f"Gespeichert: {daten}")

class BenutzerService:
    def __init__(self, repository):
        self.repository = repository

    def anlegen(self, name):
        self.repository.speichern(name)

service = BenutzerService(Repository())
service.anlegen("Ali")
```

## Zusammenfassung
Dependency Injection sorgt für lose Kopplung, und der Garbage Collector hilft Python dabei, ungenutzte Objekte automatisch zu bereinigen.
