# Kapitel 18.10 — Datenbankdesign und Best Practices

## Lernziele
- Datenbanken richtig designen.
- Normalisierung verstehen.
- Performance und Sicherheit optimieren.

## Theorie

Ein gutes Datenbankdesign ist fundamental für Performanz, Wartbarkeit und Sicherheit.

## Normalisierung

Normalisierung reduziert Redundanz und verbessert Datenintegrität.

- **1. Normalform (1NF)**: Atomare Werte, keine Wiederholungsgruppen
- **2. Normalform (2NF)**: Keine partiellen Abhängigkeiten
- **3. Normalform (3NF)**: Keine transitiven Abhängigkeiten

## Beispiel 1
Schlechtes Design (nicht normalisiert).

```
Tabelle: SchülerKurse
| schüler_id | name  | kurse              |
|------------|-------|--------------------|
| 1          | Anna  | Mathe,Deutsch      |
| 2          | Beni  | Englisch,Mathe     |
```

Problem: `kurse` ist nicht atomar.

## Beispiel 2
Gutes Design (normalisiert).

```
Tabelle: Schüler
| schüler_id | name  |
|------------|-------|
| 1          | Anna  |
| 2          | Beni  |

Tabelle: Kurse
| kurs_id | kursname |
|---------|----------|
| 1       | Mathe    |
| 2       | Deutsch  |
| 3       | Englisch |

Tabelle: SchülerKurse (Junction Table)
| schüler_id | kurs_id |
|------------|----------|
| 1          | 1        |
| 1          | 2        |
| 2          | 3        |
| 2          | 1        |
```

## Performance-Optimierung

- **Indexierung**: Schnellere Suchen
- **Denormalisierung** (selten): Wenn Performance kritisch ist
- **Partitionierung**: Große Tabellen aufteilen
- **Caching**: Häufige Abfragen cachen

````python
import sqlite3

conn = sqlite3.connect('schule.db')
cursor = conn.cursor()

# Index erstellen
cursor.execute("CREATE INDEX idx_klasse ON schüler(klasse)")
cursor.execute("CREATE INDEX idx_name ON schüler(name)")

conn.commit()
conn.close()
````

## Sicherheit

- **SQL-Injection verhindern**: Immer parametrisierte Abfragen verwenden
- **Authentifizierung**: Starke Passwörter
- **Autorisierung**: Granulare Rechte
- **Verschlüsselung**: Sensitive Daten verschlüsseln

````python
import sqlite3

conn = sqlite3.connect('schule.db')
cursor = conn.cursor()

# Richtig: Parametrisierte Abfrage
name = "Anna"
cursor.execute("SELECT * FROM schüler WHERE name = ?", (name,))

# Falsch: String-Konkatenation (SQL-Injection Gefahr)
# cursor.execute(f"SELECT * FROM schüler WHERE name = '{name}'")

conn.close()
````

## Backup und Recovery

```bash
# SQLite Backup
sqlite3 schule.db ".backup '/path/to/backup.db'"

# MySQL Dump
mysqldump -u root -p schule > schule_backup.sql

# PostgreSQL Dump
pg_dump schule > schule_backup.sql
```

## Monitoring und Logging

- Abfrage-Logs überprüfen
- Langsame Abfragen identifizieren
- Regelmäßige Backups
- Disk Space überwachen

## Häufige Fehler
- Fehlende Normalisierung
- SQL-Injection
- Keine Backups
- Falsche Indexierung

## Best Practice
Plane dein Datenbankdesign sorgfältig, bevor du mit der Implementierung beginnst.

## Zusammenfassung
Gutes Datenbankdesign ist das Fundament für zuverlässige und effiziente Anwendungen.
