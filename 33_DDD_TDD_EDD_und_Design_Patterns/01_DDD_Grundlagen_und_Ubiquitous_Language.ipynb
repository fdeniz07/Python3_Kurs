## Theorie

DDD zielt darauf ab, technische Entscheidungen mit Geschaeftsregeln in Einklang zu halten.
Im Zentrum des Modells steht nicht die Datenstruktur, sondern die fachliche Bedeutung.

## Kernbegriffe

- Domain, Subdomain, Bounded Context
- Ubiquitous Language
- Strategic vs Tactical Design

```python
# Einfaches Beispiel fuer Domain-Sprache
class EnrollmentStatus:
    PENDING = 'pending'
    ACTIVE = 'active'
    CANCELLED = 'cancelled'

def can_access_lesson(status):
    return status == EnrollmentStatus.ACTIVE

print(can_access_lesson(EnrollmentStatus.ACTIVE))
```

## Praxisbeispiel

In einer Kursplattform werden Begriffe wie 'Teilnehmer', 'Kurs' und 'Einschreibung' direkt in technischen Klassennamen abgebildet.

## Best Practices

- Ein gemeinsames Begriffslexikon mit dem Fachbereich pflegen
- In Code Reviews auf sprachliche Konsistenz achten
- Grenzen von Bounded Contexts klar dokumentieren

## Haeufige Fehler

- Technische Bezeichnungen ersetzen fachliche Namen
- Das gesamte System in ein einziges Grossmodell pressen
- Kontextgrenzen ignorieren

## Zusammenfassung

DDD ist in grossen Systemen ein zentraler Ansatz fuer semantische Konsistenz und langfristige Wartbarkeit.

## Technischer Tiefgang

Auf fortgeschrittenem Niveau steht nicht nur Funktionalitaet, sondern die technische Nachvollziehbarkeit im Vordergrund.
Begriffe, Risiken, Betriebsaspekte und Qualitaetskriterien werden explizit gemacht, damit Entscheidungen reproduzierbar bleiben.

## Zentrale Fachbegriffe

Design Constraint
Trade-off
Failure Mode
Observability Signal
Quality Gate
Regression Risk

```python
# Zusatzbeispiel: Priorisierung technischer Risiken
risks=[{"name":"regression","score":9},{"name":"operability","score":8},{"name":"complexity","score":7}]
for r in sorted(risks,key=lambda x:x["score"],reverse=True):
    print(r["name"], r["score"])
```

- Sind Eingaben, Konfiguration und Randbedingungen explizit validiert?
- Sind reproduzierbare Schritte fuer den Fehler dokumentiert?
- Wurden Logs, Metriken und Tests gemeinsam ausgewertet?

- Gibt es eine kurze Lessons-Learned-Notiz fuer das Team?
