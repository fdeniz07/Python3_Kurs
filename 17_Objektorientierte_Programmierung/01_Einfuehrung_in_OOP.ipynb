## Theorie
Die objektorientierte Programmierung (OOP) hilft uns, Programme in kleine, gut organisierte Bausteine zu zerlegen.
Statt nur Daten und Funktionen getrennt zu behandeln, bündeln wir sie in Objekten.
Ein Objekt ist eine konkrete Instanz einer Klasse.

## Erklärung
- Klasse: Bauplan für ein Objekt
- Objekt: konkrete, real existierende Instanz dieser Klasse
- Attribut: Daten eines Objekts
- Methode: Funktion, die zu einem Objekt gehört

## Syntax
```python
class KlassenName:
    def __init__(self, parameter):
        self.attribut = parameter

    def methode(self):
        ...
```

## Merke
OOP ist besonders hilfreich, wenn Programme größer werden und viele zusammenhängende Daten verwaltet werden müssen.

````python
class Auto:
    def __init__(self, marke, farbe):
        self.marke = marke
        self.farbe = farbe

    def beschreiben(self):
        print(f"Ich bin ein {self.farbe} {self.marke}.")

mein_auto = Auto("VW", "blau")
mein_auto.beschreiben()
````

## OOP - Objektorientierte Programmierung 

Objektorientierung ist eine sehr mächtige Methode der strukturellen Zerlegung, denn ein umfangreiches Programm kann in relativ große, in sich zusammenhängende Teile gegliedert werden, die ihrerseits noch weiter unterteilt werden können. Deshalb sind moderne, professionelle Programme objektorientiert. Die Grundideen der OOP lassen sich in drei Punkten zusammenfassen:

### 1. Verteilte Zuständigkeiten
- einfache Programme folgen imperativem Programmierstil mit nur einem Akteur, der den Programmlauf kontrolliert
- in objektorientierten Programm gibt es mehrere Akteure: Objekte, die untereinander Botschaften austauschen und gemeinsam eine Aufgabe lösen

### 2.Objekte besitzen Attribute und Methoden
- ein Objekt fasst Daten (Attribute) und Operationen (Methoden) zu einer aktiven Einheit zusammen

### 3. Ein Objeckt ist die Instanz einer Klasse
- zu einer Klasse gehören Objekte mit gleichen Eigenschaften (Attributen) und gleichen Operationen
- eine Klasse ist wie ein Bauplan für Objekte, in dem Attribute und Methoden festgelegt sind

## UML-Klassendiagramm (UML: Unified Modeling Language)
Überlegungen zu Attributen und Methoden können in einem UML-Klassendiagramm (UML: Unified Modeling Language) zusammengefasst werden. Ein UML-Klassendiagramm ist die Grundlage für die Entwicklung von Python-Programmtext.

### Name der Klasse (Flasche)

Ein objektorientiertes Programmierprojekt beginnt mit dem Entwurf einer Klassenstruktur. Eine Klasse definiert einen Typ von Objekten. Klassen sind Modelle von Dingen, Ereignissen oder Ideen der realen Welt. Als Beispiel wird eine Klasse entwickelt, die Flaschen modelliert. Der Name der Klasse ist deshalb Flasche. Klassennamen beginnen üblicherweise mit einem Großbuchstaben.


### Attribute (inhalt, max_inhalt, geöffnet)

Attribute (oder Objektvariablen) sind die Merkmale oder Eigenschaften eines Objekts. Sie werden mit Werten belegt. Ein Objekt befindet sich immer in einem bestimmten Zustand. Der momentane Zustand wird durch die Belegungen der Attribute definiert.

Das erste Attribut der Flasche wird inhalt genannt. Es ist das Volumen einer Flüssigkeit in cm³, mit der die Flasche gefüllt ist. Eine Flasche hat ein bestimmtes Fassungsvermögen, also einen maximalen Inhalt, der nicht überschritten werden kann. Dieses Attribut wird max_inhalt genannt. Nun kann eine Flasche geöffnet oder geschlossen sein. Diese Facette ihres Zustands soll durch ein boolesches Attribut namens geöffnet modelliert werden. Wenn geöffnet den Wert True trägt, ist die Flasche geöffnet, ansonsten ist sie geschlossen.


### Methoden (öffnen(), schließen(), füllen(), leeren())

Die Operationen, die Objekte ausführen können, nennt man Methoden.

Welche Methoden beherrscht eine Flasche? Sie kann sich öffnen und schließen. Sie kann sich mit einem bestimmten Volumen füllen, sofern sie geöffnet ist. Und sie kann sich leeren, sofern sie geöffnet ist. Im Unterschied zur Realität sind in der objektorientierten Denkweise die Objekte eigenaktiv und ändern selbst ihren Zustand. Es wird sich vorgestellt, dass ein Objekt der Klasse Flasche sich selbst öffnet, wenn es den Auftrag dazu erhält.

## Beispiel 1
Hier wird sichtbar, dass die Daten und das Verhalten eines Autos zusammengehören.

````python
class Hund:
    def __init__(self, name):
        self.name = name

    def bellen(self):
        print(self.name + " sagt: Wuff!")

hund = Hund("Bello")
hund.bellen()
````

## Beispiel 2
Ein Objekt kann sich wie ein reales Ding verhalten.

````python
class Person:
    def __init__(self, vorname, nachname):
        self.vorname = vorname
        self.nachname = nachname

    def vorstellen(self):
        print(f"Hallo, ich heiße {self.vorname} {self.nachname}.")

p = Person("Anna", "Müller")
p.vorstellen()
````

## Praxisbeispiel
Wir modellieren ein Buch als Objekt.

````python
class Buch:
    def __init__(self, titel, autor):
        self.titel = titel
        self.autor = autor

    def info(self):
        print(f"{self.titel} von {self.autor}")

buch1 = Buch("Python lernen", "Max")
buch1.info()
````

## Häufige Fehler
- `self` vergessen
- Attribute außerhalb der Klasse falsch verwenden
- Objekt vor dem Erstellen einer Klasse ansprechen

## Best Practice
Namen von Klassen sollten mit Großbuchstaben beginnen.

## Tipp
Stell dir eine Klasse wie einen Bauplan und ein Objekt wie ein fertiges Produkt vor.

## Übung
Erstelle eine Klasse `Lampe` mit einem Attribut `zustand` und einer Methode `einschalten()`.

````python
class Lampe:
    def __init__(self, zustand="aus"):
        self.zustand = zustand

    def einschalten(self):
        self.zustand = "an"
        print("Die Lampe ist jetzt an.")

lampe = Lampe()
lampe.einschalten()
````
