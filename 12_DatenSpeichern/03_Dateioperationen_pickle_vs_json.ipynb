# Datenstrukturen speichern und laden: pickle vs. json

Um komplexe Datenstrukturen (wie Listen oder Dictionaries) dauerhaft zu speichern, nutzen wir in Python hauptsächlich zwei Module: `pickle` und `json`.

### 1. Das Pickle-Modul
`pickle` ist ein Python-spezifisches Modul. Es wandelt (serialisiert) Python-Objekte in einen Byte-Strom um. Das bedeutet, man kann fast jedes Python-Objekt speichern, aber diese Datei ist nur mit Python lesbar.

```python
import pickle

daten = {"Name": "Max", "Alter": 25, "Hobbys": ["Fußball", "Lesen"]}

# Speichern mit pickle (Binär-Modus 'wb')
with open("daten.pkl", "wb") as f:
    pickle.dump(daten, f)

# Laden mit pickle (Binär-Modus 'rb')
with open("daten.pkl", "rb") as f:
    geladene_daten = pickle.load(f)
    print("Pickle-Daten:", geladene_daten)
```

### 2. Das JSON-Modul
`json` (JavaScript Object Notation) ist ein plattformunabhängiges Format. Es ist menschenlesbar (Textdatei) und kann von vielen anderen Programmiersprachen gelesen werden. Es unterstützt jedoch nur Basis-Datentypen (String, int, list, dict).

```python
import json

daten = {"Name": "Max", "Alter": 25, "Hobbys": ["Fußball", "Lesen"]}

# Speichern mit json (Text-Modus 'w')
with open("daten.json", "w", encoding="utf-8") as f:
    json.dump(daten, f, indent=4)

# Laden mit json (Text-Modus 'r')
with open("daten.json", "r", encoding="utf-8") as f:
    geladene_json = json.load(f)
    print("JSON-Daten:", geladene_json)
```
