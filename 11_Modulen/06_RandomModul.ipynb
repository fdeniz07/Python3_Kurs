## Random-Modul

Ein Modul, das verschiedene Attribute und Funktionen enthält, um Zufallszahlen zu generieren.

```python
import random
```

### Funktion `random`

Ermöglicht es uns, eine zufällige Fließkommazahl (Dezimalzahl) zwischen 0 und 1 zu generieren.

```python
random.random()
```

### Funktion `uniform`

Ermöglicht es uns, eine zufällige Fließkommazahl (Dezimalzahl) innerhalb eines selbst definierten Bereichs zu generieren.

```python
random.uniform(0.5, 2.5)
```

### Funktion `randint`

Ermöglicht es uns, eine ganzzahlige Zufallszahl (Integer) innerhalb eines selbst definierten Bereichs (Grenzen inklusive) zu generieren.

```python
random.randint(1, 100)
```

### Funktion `randrange`

Ermöglicht es uns, eine ganzzahlige Zufallszahl innerhalb eines definierten Bereichs mit optionaler Schrittweite zu generieren.

```python
random.randrange(100)
```

```python
random.randrange(50, 200)
```

### Funktion `choice`

Ermöglicht die Auswahl eines zufälligen Elements aus einer übergebenen Liste oder Sequenz.

```python
programmiersprachen = ["C", "C++", "Python", "Ruby", "Java"]
random.choice(programmiersprachen)
```

```python
random.choice([i for i in range(50)])
```

### Funktion `shuffle`

Sorgt dafür, dass eine übergebene Liste zufällig durchgemischt wird (In-Place-Operation).

```python
meine_liste = [i for i in range(20)]
meine_liste
```

```python
random.shuffle(meine_liste)
meine_liste
```

### Funktion `sample`

Erstellt eine zufällige Teilliste (Stichprobe) basierend auf einer übergebenen Liste und der gewünschten Anzahl von Elementen.

```python
große_liste = [i for i in range(100)]
große_liste
```

```python
random.sample(große_liste, 10)
```
