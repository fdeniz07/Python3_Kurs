# Kapitel 18.5 — SQLite in Python

## Lernziele
- SQLite mit Python verwenden.
- Verbindungen aufbauen und schließen.
- Daten einfügen, abrufen und ändern.

## Theorie

SQLite ist eine einfache, dateibasierte Datenbank, die perfekt für kleine Projekte ist. Sie benötigt keinen separaten Server und wird mit Python bereits mitgeliefert (Modul `sqlite3`).

```python
import sqlite3

# Verbindung erstellen oder öffnen
conn = sqlite3.connect('schule.db')
cursor = conn.cursor()

# Tabelle erstellen
cursor.execute('''
    CREATE TABLE IF NOT EXISTS schüler (
        id INTEGER PRIMARY KEY,
        name TEXT NOT NULL,
        klasse TEXT
    )
''')

conn.commit()
conn.close()
```

## Beispiel 1
Daten einfügen.

```python
import sqlite3

conn = sqlite3.connect('schule.db')
cursor = conn.cursor()

# Einzelnen Datensatz einfügen
cursor.execute("INSERT INTO schüler (name, klasse) VALUES (?, ?)", ('Anna', '10A'))

# Mehrere Datensätze einfügen
daten = [
    ('Beni', '10B'),
    ('Carla', '10A'),
    ('David', '11A')
]
cursor.executemany("INSERT INTO schüler (name, klasse) VALUES (?, ?)", daten)

conn.commit()
conn.close()
```

## Beispiel 2
Daten abrufen.

```python
import sqlite3

conn = sqlite3.connect('schule.db')
cursor = conn.cursor()

# Alle Datensätze abrufen
cursor.execute("SELECT * FROM schüler")
alle_schüler = cursor.fetchall()
print(alle_schüler)

# Mit Bedingung abrufen
cursor.execute("SELECT * FROM schüler WHERE klasse = ?", ('10A',))
schüler_10a = cursor.fetchall()
print(schüler_10a)

conn.close()
```

## Beispiel 3
Daten aktualisieren und löschen.

```python
import sqlite3

conn = sqlite3.connect('schule.db')
cursor = conn.cursor()

# Daten aktualisieren
cursor.execute("UPDATE schüler SET klasse = ? WHERE name = ?", ('10B', 'Anna'))

# Daten löschen
cursor.execute("DELETE FROM schüler WHERE id = ?", (4,))

conn.commit()
conn.close()
```

## Praxisbeispiel
Eine einfache Klasse zum Verwalten von Schülern.

```python
import sqlite3

class SchülerVerwaltung:
    def __init__(self, db_name='schule.db'):
        self.conn = sqlite3.connect(db_name)
        self.cursor = self.conn.cursor()
    
    def schüler_hinzufügen(self, name, klasse):
        self.cursor.execute("INSERT INTO schüler (name, klasse) VALUES (?, ?)", (name, klasse))
        self.conn.commit()
    
    def alle_schüler(self):
        self.cursor.execute("SELECT * FROM schüler")
        return self.cursor.fetchall()
    
    def close(self):
        self.conn.close()

verwaltung = SchülerVerwaltung()
verwaltung.schüler_hinzufügen('Eva', '11B')
print(verwaltung.alle_schüler())
verwaltung.close()
```

## Häufige Fehler
- `conn.commit()` vergessen
- Keine Parameterersetzung verwenden (SQL-Injection)
- Verbindung nicht schließen

## Best Practice
Verwende immer `?` für Parameter und rufe `commit()` nach Änderungen auf.

## Zusammenfassung
SQLite ist eine praktische Wahl für Python-Projekte mit kleinen bis mittleren Datenmengen.
