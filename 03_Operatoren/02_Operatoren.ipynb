## Was sind Operatoren?

##### Arithmetische Operatoren: Addition, Subtraktion, Multiplikation, Ganzzahldivision, Gleitkomma-Division, Modulo, Potenzierung
##### Operatoren für Vergleichsoperationen: ==, !=, >, <, >=, <=
##### Logische Operatoren: AND, OR, NOT
##### Operatoren für Bit-Operationen: &, |, ^, ~, <<, >>
##### Extras: in, not in, is, is not

### Operatoren für Vergleichsoperationen

```python
a,b,c=5,10,15

print(a==b) # Wenn zwei Ausdrücke gleich sind True, sonst False
print(a==c)

print(a!=b) # Wenn zwei Ausdrücke nicht gleich sind True, sonst False
print(a!=c)

print(a>b) # Wenn der erste Ausdruck größer als der zweite ist True, sonst False
print(a>c)

print(a<b) # Wenn der erste Ausdruck kleiner als der zweite ist True, sonst False
print(a<c)

print(a>=b) # Wenn der erste Ausdruck größer oder gleich dem zweiten ist True, sonst False
print(a>=c)

print(a<=b) # Wenn der erste Ausdruck kleiner oder gleich dem zweiten ist True, sonst False
print(a<=c)
```

#### Was passiert bei mehreren Abfragen?

```python
print(a<b<c) # a < b UND b < c
```
