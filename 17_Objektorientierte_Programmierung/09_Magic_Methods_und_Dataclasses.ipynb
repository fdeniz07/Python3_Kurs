# Kapitel 17.9 - Magic Methods und Dataclasses

## Lernziele
- Verstehen, was Magic Methods sind und wann Python sie automatisch aufruft.
- `__str__`, `__repr__`, `__len__`, `__eq__` und `__add__` praktisch anwenden.
- `@dataclass` nutzen, um Datenklassen kompakt und lesbar zu schreiben.
- Eigene Klassen durch magische Methoden wie eingebaute Typen bedienbar machen.

## Theorie
Magic Methods (auch Dunder-Methoden genannt) sind spezielle Methoden mit doppelten Unterstrichen, z. B. `__str__`.
Sie werden nicht direkt von dir, sondern oft automatisch durch Python aufgerufen.

## Erklaerung
Wenn du `print(objekt)` schreibst, prueft Python zuerst, ob `objekt.__str__()` existiert.
Wenn du `len(objekt)` verwendest, ruft Python intern `objekt.__len__()` auf.
Bei `obj1 == obj2` wird `__eq__` genutzt.

## Syntax
```python
class MeineKlasse:
    def __str__(self):
        return "Lesbare Darstellung"

    def __len__(self):
        return 42
```

## Merke
Magic Methods machen Objekte benutzerfreundlich. Du steuerst damit, wie sich eigene Klassen bei Standardoperationen verhalten.

````python
class Buch:
    def __init__(self, titel, autor):
        self.titel = titel
        self.autor = autor

    def __str__(self):
        return f"'{self.titel}' von {self.autor}"

    def __len__(self):
        return len(self.titel)


b = Buch("Python fuer Einsteiger", "Fatih Deniz")
print(b)
print("Titel-Laenge:", len(b))
````

## Parameter und Rueckgabewerte (wichtig)
- `__str__(self) -> str`: Gibt eine **menschlich lesbare** Darstellung zurueck.
- `__repr__(self) -> str`: Gibt eine **entwicklerfreundliche** Darstellung zurueck.
- `__len__(self) -> int`: Muss eine Ganzzahl (>= 0) liefern.
- `__eq__(self, other) -> bool`: Gibt `True` oder `False` zurueck.
- `__add__(self, other) -> Objekt`: Definiert das Verhalten von `+`.

## Beispiel 1 - `__str__` und `__len__`
`__str__` macht Ausgaben lesbar, `__len__` liefert eine sinnvolle Laenge.

## Kurztabelle: Haeufige Magic Methods

| Methode | Wird aufgerufen bei | Rueckgabe | Typischer Zweck |
|---|---|---|---|
| `__str__(self)` | `print(obj)` / `str(obj)` | `str` | Lesbare Ausgabe fuer Nutzer |
| `__repr__(self)` | `repr(obj)` / Debug-Kontext | `str` | Eindeutige Entwickler-Darstellung |
| `__len__(self)` | `len(obj)` | `int` | Anzahl/Laenge eines Objekts |
| `__eq__(self, other)` | `obj1 == obj2` | `bool`/`NotImplemented` | Fachlich sinnvoll vergleichen |
| `__add__(self, other)` | `obj1 + obj2` | neues Objekt/`NotImplemented` | Objekte kombinieren |

### Weitere nuetzliche Magic Methods (Kurzueberblick)
- `__init__(self, ...)`: Initialisiert ein Objekt nach der Erstellung.
- `__lt__(self, other)`: Vergleich mit `<` (kleiner als).
- `__le__(self, other)`: Vergleich mit `<=`.
- `__gt__(self, other)`: Vergleich mit `>`.
- `__ge__(self, other)`: Vergleich mit `>=`.
- `__contains__(self, item)`: Wird bei `item in obj` genutzt.
- `__getitem__(self, key)`: Zugriff mit `obj[key]`.
- `__setitem__(self, key, value)`: Setzen mit `obj[key] = value`.
- `__iter__(self)`: Macht ein Objekt iterierbar (z. B. in `for`-Schleifen).
- `__call__(self, ...)`: Macht ein Objekt wie eine Funktion aufrufbar.

Merke: Implementiere nur Methoden, die fuer dein Dominenmodell wirklich sinnvoll sind.

````python
class Punkt:
    def __init__(self, x, y):
        self.x = x
        self.y = y

    def __repr__(self):
        return f"Punkt(x={self.x}, y={self.y})"

    def __str__(self):
        return f"({self.x}, {self.y})"

    def __eq__(self, other):
        if not isinstance(other, Punkt):
            return NotImplemented
        return self.x == other.x and self.y == other.y


p1 = Punkt(2, 3)
p2 = Punkt(2, 3)
p3 = Punkt(5, 1)

print("str:", p1)
print("repr:", repr(p1))
print("p1 == p2:", p1 == p2)
print("p1 == p3:", p1 == p3)
````

## Beispiel 2 - Praxisbeispiel Warenkorb
Wir modellieren einen Warenkorb. Mit `__add__` koennen zwei Warenkoerbe addiert werden, mit `__len__` zaehlen wir Produkte.

## Beispiel 3 - Dataclass in der Praxis
Dataclasses reduzieren Boilerplate-Code: Konstruktor, Vergleich und oft auch Darstellung werden automatisch erzeugt.

````python
class Warenkorb:
    def __init__(self, produkte=None):
        self.produkte = produkte or []

    def __len__(self):
        return len(self.produkte)

    def __add__(self, other):
        if not isinstance(other, Warenkorb):
            return NotImplemented
        return Warenkorb(self.produkte + other.produkte)

    def __str__(self):
        if not self.produkte:
            return "Warenkorb ist leer."
        return "Warenkorb: " + ", ".join(self.produkte)


wk1 = Warenkorb(["Buch", "Stift"])
wk2 = Warenkorb(["Heft"])
wk3 = wk1 + wk2
print(wk1)
print(wk2)
print(wk3)
print("Anzahl Produkte:", len(wk3))
````

Kurze Beobachtung: Bei `@dataclass` ist `__init__`, `__repr__` und (standardmaessig) `__eq__` bereits automatisch vorhanden.

````python
from dataclasses import dataclass

@dataclass
class Student:
    name: str
    note: float


s1 = Student("Lina", 1.7)
s2 = Student("Lina", 1.7)
print(s1)
print("s1 == s2:", s1 == s2)
````

## Haeufige Fehler
- `__str__` gibt keinen String zurueck (muss immer `str` sein).
- `__len__` liefert negative Werte oder keinen `int`.
- In `__eq__` wird `other` nicht auf den richtigen Typ geprueft.
- `__add__` veraendert bestehende Objekte unerwartet statt ein neues Objekt zurueckzugeben.
- Dataclass wird genutzt, obwohl komplexe Logik und Seiteneffekte notwendig sind.

## Best Practice
- Implementiere nur die Magic Methods, die fachlich sinnvoll sind.
- Halte das Verhalten intuitiv: `len(obj)` sollte wirklich eine Menge/Laenge beschreiben.
- Nutze `NotImplemented` bei Typkonflikten (`__eq__`, `__add__`).
- Verwende Dataclasses fuer reine Datenobjekte mit wenig Logik.

## Tipp
Starte mit `__str__` und `__repr__`. Das verbessert Debugging und Ausgabe sofort spuerbar.

## Uebung
1. Erstelle eine `@dataclass` `Produkt` mit `name` und `preis`.
2. Erstelle eine Klasse `Einkaufsliste` mit Attribut `produkte` (Liste von `Produkt`).
3. Implementiere:
   - `__len__`, damit `len(einkaufsliste)` die Anzahl der Produkte liefert.
   - `__str__`, damit eine lesbare Ausgabe entsteht.
   - Optional `__add__`, um zwei Einkaufslisten zusammenzufuehren.

````python
from dataclasses import dataclass

@dataclass
class Produkt:
    name: str
    preis: float


class Einkaufsliste:
    def __init__(self, produkte=None):
        self.produkte = produkte or []

    def __len__(self):
        return len(self.produkte)

    def __str__(self):
        if not self.produkte:
            return "Einkaufsliste ist leer."
        eintraege = [f"{p.name} ({p.preis:.2f} EUR)" for p in self.produkte]
        return "Einkaufsliste: " + ", ".join(eintraege)

    def __add__(self, other):
        if not isinstance(other, Einkaufsliste):
            return NotImplemented
        return Einkaufsliste(self.produkte + other.produkte)


liste1 = Einkaufsliste([Produkt("Milch", 1.49), Produkt("Brot", 2.19)])
liste2 = Einkaufsliste([Produkt("Apfel", 0.79)])
gesamt = liste1 + liste2

print(liste1)
print(gesamt)
print("Anzahl Produkte:", len(gesamt))
````

## Zusammenfassung
- Magic Methods steuern, wie sich eigene Objekte bei Standardoperationen verhalten.
- Besonders haeufig: `__str__`, `__repr__`, `__len__`, `__eq__`, `__add__`.
- Dataclasses sparen viel Schreibarbeit bei einfachen Datenobjekten.
- Gute Klassen wirken fuer Nutzende wie eingebaute Python-Typen.
