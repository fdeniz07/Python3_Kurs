# Rueckgabewert

Tooling liefert oft Exit-Codes und Reports (z. B. Test-Report, Lint-Fehlerliste).

````python
# Beispiel 1: Kleine Service-Funktion mit klaren Verantwortlichkeiten
def berechne_rabatt(preis, prozent):
    if preis < 0:
        raise ValueError('preis darf nicht negativ sein')
    if not (0 <= prozent <= 100):
        raise ValueError('prozent muss zwischen 0 und 100 liegen')
    return preis * (1 - prozent / 100)

print(berechne_rabatt(100, 15))
````

````python
# Beispiel 2: Mini-Tests ohne externe Bibliothek
def test_berechne_rabatt():
    assert berechne_rabatt(200, 10) == 180
    try:
        berechne_rabatt(-1, 10)
    except ValueError:
        pass
    else:
        raise AssertionError('ValueError erwartet')

test_berechne_rabatt()
print('Tests erfolgreich')
````

````python
# Beispiel 3: Dokumentationsnahe API-Signatur mit Type Hints
def erstelle_benutzer(name: str, rolle: str = 'user') -> dict:
    if not name.strip():
        raise ValueError('name darf nicht leer sein')
    return {'name': name.strip(), 'rolle': rolle}

print(erstelle_benutzer('  Ada  ', 'admin'))
````

## Haeufige Fehler und Debugging-Checkliste

- Ist der Fehler reproduzierbar und sauber eingegrenzt?
- Sind Messsignale (Tests, Logs, Kennzahlen) vorhanden?
- Wurde eine Hypothese formuliert und mit Gegenbeweisen geprueft?
- Ist die Korrektur durch Tests dauerhaft abgesichert?
- Wurde die Ursache im Team dokumentiert (Lessons Learned)?
