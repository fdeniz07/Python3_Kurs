## Time-Modul

Ein Modul, das nützliche Attribute und Funktionen für zeit- und stundenbezogene Operationen bereitstellt.

```python
import time
```

### Funktion `gmtime`

Gibt die Startzeit (EPOCH) in Form einer Struktur (`struct_time`) zurück und enthält Attribute wie `tm_year`, `tm_mon`, `tm_mday`, `tm_sec`, `tm_wday`, `tm_yday`.

```python
epoch = time.gmtime(0)
epoch
```

```python
epoch.tm_year
```

### Funktion `time`

Gibt die seit der Startzeit (EPOCH) bis zu diesem Moment vergangene Gesamtzeit in Sekunden zurück.

```python
time.time()
```

### Funktion `localtime`

Gibt die lokale Zeitinformation in Form einer Struktur (`struct_time`) zurück und enthält Attribute wie `tm_year`, `tm_mon`, `tm_mday`.

```python
time.localtime()
```

### Funktion `sleep`

Sorgt dafür, dass der geschriebene Code für eine bestimmte Zeitspanne pausiert (schläft).

```python
print("Gleich wird sleep ausgeführt und das Programm pausiert")
time.sleep(3)
print("Gerade eben wurde sleep ausgeführt und ich habe pausiert")
```
