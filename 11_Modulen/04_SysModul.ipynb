## Sys-Modul

Ermöglicht es Ihnen, Informationen über die von Ihnen verwendete Python-Version zu erhalten und verschiedene Operationen mit der Python-Laufzeitumgebung durchzuführen.

```python
import sys
```

### Attribut `argv`

Hält die Parameter, die beim Ausführen des Programms übergeben wurden, in Form einer Liste.

```python
sys.argv
```

### Attribut `executable`

Wird verwendet, um den Namen und den Pfad der ausführbaren Python-Datei (Interpreter) zu erfahren.

```python
sys.executable
```

### Attribut `platform`

Wird verwendet, um Informationen über das Betriebssystem zu erhalten, auf dem unser Code läuft.

GNU/Linux: Gibt 'linux' aus
Windows: Gibt 'win32' aus
macOS: Gibt 'darwin' aus

```python
sys.platform
```

### Attribut `version`

Wird verwendet, um die installierte Python-Version zu erfahren.

```python
sys.version
```

### Funktion `exit`

Wird verwendet, um den Programmfluss und die Ausführung vorzeitig zu beenden.

```python
sys.exit()
```

### Standard-Streams (Standard-Kanäle)

Die folgenden Standard-Datenströme werden verwendet, um Ausgaben zu erzeugen und Werte vom Benutzer entgegenzunehmen.

1. Standard-Ausgabekanal - `stdout`

2. Standard-Fehlerkanal - `stderr`

3. Standard-Eingabekanal - `stdin`

```python
sys.stderr.write("Fehler\n")
```

```python
sys.stdout.write("Normal\n")
```

```python
print("Normal")
```
