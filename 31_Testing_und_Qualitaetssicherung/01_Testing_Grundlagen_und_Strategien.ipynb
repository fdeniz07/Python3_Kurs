# Rueckgabewert

Tests liefern meist keinen Fachwert, sondern Bestehen/Fehlschlag mit Diagnose.

````python
# Beispiel 1: Kleine Fachfunktion

def rabatt(preis, prozent):
    if preis < 0:
        raise ValueError('preis darf nicht negativ sein')
    if not (0 <= prozent <= 100):
        raise ValueError('prozent muss zwischen 0 und 100 liegen')
    return preis * (1 - prozent/100)

print(rabatt(100, 10))
````

````python
# Beispiel 2: Sehr einfacher Teststil (ohne Framework)

def test_rabatt_grundfall():
    assert rabatt(200, 10) == 180

test_rabatt_grundfall()
print('test_rabatt_grundfall OK')
````

````python
# Beispiel 3: Fehlerfall-Test

def test_rabatt_fehlerfall():
    try:
        rabatt(-1, 10)
    except ValueError:
        return
    raise AssertionError('ValueError wurde erwartet')

test_rabatt_fehlerfall()
print('test_rabatt_fehlerfall OK')
````

# Praxisbeispiel

Wende die Teststrategie auf ein eigenes Modul an und dokumentiere Testfaelle fuer Happy Path und Edge Cases.

# Haeufige Fehler

1. Zu grosse Tests mit mehreren Verantwortlichkeiten.
2. Abhaengigkeit von Zufall/Uhrzeit/Netzwerk ohne Kontrolle.
3. Nur Happy Path testen.

# Best Practice

- Testdaten klar benennen.
- Pro Test eine klare Erwartung.
- Fehler reproduzierbar halten.

# Tipp

Schreibe zuerst den Testnamen als fachlichen Satz, danach den Testcode.

# Uebung

Erstelle mindestens 5 Tests: 3 Erfolgsfaelle, 2 Fehlerfaelle, inklusive Randwerten.

````python
# Loesung: Mini-Testset

def testset():
    assert rabatt(100, 0) == 100
    assert rabatt(100, 100) == 0
    try:
        rabatt(10, 101)
    except ValueError:
        pass
    else:
        raise AssertionError('ValueError fuer prozent>100 erwartet')


testset()
print('Testset erfolgreich')
````

## Haeufige Fehler und Debugging-Checkliste

- Sind Eingaben, Konfiguration und Randbedingungen explizit validiert?
- Sind reproduzierbare Schritte fuer den Fehler dokumentiert?
- Wurden Logs, Metriken und Tests gemeinsam ausgewertet?

- Gibt es eine kurze Lessons-Learned-Notiz fuer das Team?
