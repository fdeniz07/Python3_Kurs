# Kapitel 35.5 - Testdaten, Fixtures und Determinismus

Stabile Tests benoetigen reproduzierbare Daten und klaren Fixture-Lifecycle.

```python
from random import Random

rng = Random(42)
values = [rng.randint(1, 10) for _ in range(5)]
print(values)
assert values == [2, 1, 5, 4, 4]
```

## Best Practices

- Seed fuer Zufall explizit setzen
- Fixture-Scope bewusst waehlen
- Testdaten auf fachlich relevante Variationen fokussieren
