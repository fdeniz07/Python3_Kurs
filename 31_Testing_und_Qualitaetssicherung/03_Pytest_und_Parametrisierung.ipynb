# Merke- Ein Test prueft eine fachliche Aussage.- Fehlermeldungen sollen Ursache schnell sichtbar machen.

# ParameterTypische Testparameter: Eingabe, erwartetes Ergebnis, Fehlertyp, Randfaelle.

# RueckgabewertTests liefern meist keinen Fachwert, sondern Bestehen/Fehlschlag mit Diagnose.

````python
# Beispiel 1: Testfaelle als Datendef multipliziere(a, b):    return a * bfaelle = [    (2, 3, 6),    (0, 5, 0),    (-2, 4, -8)]for a, b, erwartet in faelle:    assert multipliziere(a, b) == erwartetprint('parametrisierte Logik erfolgreich')
````

````python
# Beispiel 2: Eigene einfache Parametrisierungsschleifedef pruefe_faelle(funktion, faelle):    for index, (a, b, erwartet) in enumerate(faelle, start=1):        ergebnis = funktion(a, b)        assert ergebnis == erwartet, f'Fall {index} fehlgeschlagen: {ergebnis} != {erwartet}'pruefe_faelle(multipliziere, faelle)print('Alle Faelle gruen')
````

````python
# Beispiel 3: Fehlerfall mit Exception-Pruefungdef teile(a, b):    return a / btry:    teile(5, 0)except ZeroDivisionError:    print('ZeroDivisionError korrekt erkannt')
````

# PraxisbeispielWende die Teststrategie auf ein eigenes Modul an und dokumentiere Testfaelle fuer Happy Path und Edge Cases.

# Haeufige Fehler1. Zu grosse Tests mit mehreren Verantwortlichkeiten.2. Abhaengigkeit von Zufall/Uhrzeit/Netzwerk ohne Kontrolle.3. Nur Happy Path testen.

# Best Practice- Testdaten klar benennen.- Pro Test eine klare Erwartung.- Fehler reproduzierbar halten.

# TippSchreibe zuerst den Testnamen als fachlichen Satz, danach den Testcode.

# UebungErstelle mindestens 5 Tests: 3 Erfolgsfaelle, 2 Fehlerfaelle, inklusive Randwerten.

````python
# Loesung: Kombinierter Testlaufdef run_tests():    pruefe_faelle(multipliziere, faelle)    try:        teile(1, 0)    except ZeroDivisionError:        pass    else:        raise AssertionError('ZeroDivisionError erwartet')run_tests()print('Kombinierter Testlauf erfolgreich')
````
