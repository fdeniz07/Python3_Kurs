## Die Parameter `sep` und `end`

#### Der Parameter `end`
Mit dem Parameter `end` kann festgelegt werden, womit die Ausgabe der Funktion `print()` endet.

```python
print("Max Mustermann", end="\n:)")
print("Hallo Welt!", end="!")
```

#### Der Parameter `sep`
Mit dem Parameter `sep` kann festgelegt werden, durch welches Zeichen oder welchen Text mehrere Ausgaben getrennt werden.

```python
print("Hallo", "Welt", "Max", sep=" ")
```

##### Auf diese Weise können auch einzelne Zeichen getrennt ausgegeben werden

```python
print(*"HalloWeltMaxMustermann", sep=" ")
```
