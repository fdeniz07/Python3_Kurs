# Daten im JSON-Format speichern

### Was ist JSON?
JSON (JavaScript Object Notation) ist ein leichtgewichtiges Datenformat, das zur Speicherung und Übertragung von Daten verwendet wird. Es ist menschenlesbar und wird von nahezu allen modernen Programmiersprachen unterstützt. In Python nutzen wir das integrierte Modul `json`.

### Warum JSON nutzen?
Im Gegensatz zu binären Formaten wie `pickle` ist JSON eine reine Textdatei. Das bedeutet:
1. **Interoperabilität:** Daten können leicht zwischen verschiedenen Programmen (z.B. Python und JavaScript) ausgetauscht werden.
2. **Transparenz:** Man kann die gespeicherten Daten direkt in einem Texteditor lesen und bearbeiten.

## JSON Modul (Daten speichern)

Für die Arbeit mit JSON nutzen wir zwei Hauptfunktionen:
* `json.dump(daten, datei)`: Schreibt Daten in die Datei.
* `json.load(datei)`: Liest Daten aus der Datei.

JSON ist essenziell für die Serialisierung von Daten.

| Funktion | Erklärung |
| :--- | :--- |
| `json.dump()` | Schreibt Daten in eine Datei |
| `json.load()` | Liest Daten aus einer Datei |
| `json.dumps()` | Konvertiert Daten in einen String |

```python
import json

# Datenstruktur (ein Dictionary)
mein_planer = {
    "Termin": "24.06.2026",
    "Aktivität": "Python lernen",
    "Status": "In Bearbeitung"
}

# 1. Speichern (mit 'with' für Sicherheit)
with open("planer_daten.json", "w", encoding="utf-8") as f:
    # indent=4 sorgt für eine schöne, lesbare Formatierung
    json.dump(mein_planer, f, indent=4)

# 2. Laden
with open("planer_daten.json", "r", encoding="utf-8") as f:
    geladene_daten = json.load(f)
    print("Geladene Daten:", geladene_daten)
```

## Arbeiten mit JSON
JSON ist der Standard für Datenaustausch.

```python
import json

daten = {"Name": "Max", "Hobbys": ["Programmieren", "Wandern"]}

# In String konvertieren
json_string = json.dumps(daten, indent=4)
print(json_string)
```

## Rezept-Reader
Ein kleines System, um Rezepte aus einem JSON-ähnlichen Dictionary zu lesen.

```python
rezept_db = {
    "r": {"name": "Rich Bread", "zutaten": ["8 cups flour", "2 packs yeast"]},
    "a": {"name": "Arcadian Chili", "zutaten": ["8 beans", "chili powder"]}
}

auswahl = "r"
if auswahl in rezept_db:
    rezept = rezept_db[auswahl]
    print(f"Rezept: {rezept['name']}")
    for zutat in rezept['zutaten']:
        print(f"- {zutat}")
```
