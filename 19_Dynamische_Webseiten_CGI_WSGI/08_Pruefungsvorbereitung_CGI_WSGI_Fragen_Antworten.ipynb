# Syntax

Typische WSGI-Grundstruktur:

```python
def application(environ, start_response):
    status = '200 OK'
    headers = [('Content-Type', 'text/plain; charset=utf-8')]
    start_response(status, headers)
    return [b'Hallo']
```

# Merke

- Fachlich sauber antworten heisst: korrekt + praezise + begruendet.
- Bei HTTP/WSGI immer Statuscode, Header und Body mitdenken.

# Parameter

Pruefungsrelevante Parameter und Begriffe:

- `environ`
- `start_response`
- `REQUEST_METHOD`
- `PATH_INFO`
- `QUERY_STRING`

# Rueckgabewert

WSGI verlangt ein iterierbares Objekt mit `bytes`.
Eine haeufige korrekte Form ist `return [text.encode('utf-8')]`.

# Beispiel 1 - Theoriefragen

## Frage 1
Erklaere den Unterschied zwischen CGI und WSGI in 4-6 Saetzen.

## Musterantwort
CGI startet pro Request ein eigenes Skript als separaten Prozess. Das ist leicht zu verstehen, aber bei hoher Last ineffizient. WSGI definiert stattdessen eine standardisierte Python-Schnittstelle zwischen Server und Anwendung, sodass die App dauerhaft laufen kann. Dadurch sinkt der Overhead pro Anfrage deutlich. CGI ist didaktisch sehr gut fuer den Einstieg, WSGI ist die Grundlage vieler produktiver Python-Webanwendungen.

# Beispiel 2 - Codeverstaendnis

## Frage
Was ist am folgenden Code falsch?

```python
def app(environ, start_response):
    start_response('200 OK', [('Content-Type', 'text/plain')])
    return ['Hallo']
```

## Musterantwort
Der Rueckgabewert ist falsch, weil WSGI `bytes` statt `str` erwartet. Korrekt waere z. B. `return [b'Hallo']` oder `return ['Hallo'.encode('utf-8')]`.

# Beispiel 3 - Transferfrage

## Frage
Ein Endpoint erhaelt `?id=abc`. Welcher Statuscode ist sinnvoll und warum?

## Musterantwort
Sinnvoll ist `400 Bad Request`, weil der Client eine formal ungueltige Eingabe sendet (id ist nicht numerisch). `404` waere nur korrekt, wenn die Eingabe formal gueltig ist, aber die Ressource nicht existiert.

# Praxisbeispiel

## Kompaktfall
Entwirf fuer eine Route `/api/teilnehmer?id=...` drei Antwortfaelle:

1. Erfolg (200)
2. Ungueltige ID (400)
3. Nicht gefunden (404)

Formuliere pro Fall ein JSON-Beispiel.

````python
# Musterloesung als Datenstruktur
antworten = {
    '200': {'ok': True, 'teilnehmer': {'id': 12, 'name': 'Mina'}},
    '400': {'ok': False, 'code': 'INVALID_ID', 'nachricht': 'id muss numerisch sein'},
    '404': {'ok': False, 'code': 'NOT_FOUND', 'nachricht': 'Teilnehmer nicht gefunden'}
}

for status, payload in antworten.items():
    print(status, payload)
````

# Haeufige Fehler

1. Begriffe nur oberflaechlich und ohne Ursache-Wirkung erklaeren.
2. 400 und 404 verwechseln.
3. Bei WSGI den bytes-Rueckgabewert vergessen.
4. Musterantwort blind auswendig lernen statt zu verstehen.

# Best Practice

- Antworte in der Reihenfolge: Definition -> Mechanismus -> Konsequenz -> Beispiel.
- Begruende Statuscodes immer fachlich.
- Nutze kleine, saubere Codefragmente fuer Erklaerungen.

````python
# WSGI-Minibeispiel mit Statuscode
def app(environ, start_response):
    path = environ.get("PATH_INFO", "/")
    if path == "/health":
        start_response("200 OK", [("Content-Type", "text/plain")])
        return [b"ok"]
    start_response("404 Not Found", [("Content-Type", "text/plain")])
    return [b"not found"]
````
