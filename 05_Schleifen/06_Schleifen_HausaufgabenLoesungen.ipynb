### Aufgabe 1: Dreiecke mit Schleifen zeichnen

Aufgabenbeschreibung: Zeichnen Sie abhängig von einer eingegebenen Dreiecksgröße die folgenden Dreiecksformen.

Beispiel: Wenn die Dreiecksgröße 6 beträgt:

```python
x            xxxxxx         xxxxxxxxxxx             x

xx           xxxxx          xxxxxxxxx              xxx

xxx          xxxx           xxxxxxx               xxxxx  

xxxx         xxx            xxxxx                xxxxxxx 

xxxxx        xx             xxx                 xxxxxxxxx

xxxxxx       x              x                  xxxxxxxxxxx
```

```python
dreiecksGroesse=int(input("Bitte geben Sie die Dreiecksgröße ein:"))
for i in range(1,dreiecksGroesse+1):
    for j in range(0,i):
        print("x",end="")
    print(end="\n")
```

```python
dreiecksGroesse=int(input("Bitte geben Sie die Dreiecksgröße ein:"))
for i in range(0,dreiecksGroesse):
    for j in range(dreiecksGroesse-i,0,-1):
        print("x",end="")
    print(end="\n")
```

### Aufgabe 2: Passwortüberprüfung

Aufgabenbeschreibung: Überprüfen Sie, ob ein Passwort alle Anforderungen erfüllt.

Bedingung 1: Groß- und Kleinbuchstaben müssen enthalten sein.

Bedingung 2: Mindestens eine Zahl muss enthalten sein.

Bedingung 3: Das Passwort muss länger als 8 Zeichen sein.

Bedingung 4: Das Passwort muss kürzer als 16 Zeichen sein.

```python
kleinbuchstabeVorhanden=False
grossbuchstabeVorhanden=False
zahlVorhanden=False
anzahlZeichen=0

passwort=input("Bitte geben Sie passwort ein:")
for ch in passwort:
    if grossbuchstabeVorhanden==False and ch>="A" and ch<="Z":
        grossbuchstabeVorhanden=True
    elif kleinbuchstabeVorhanden==False and ch>='a' and ch<='z':
        kleinbuchstabeVorhanden=True
    elif zahlVorhanden==False and ch>='0' and ch<='9':
        zahlVorhanden=True
    anzahlZeichen+=1
if grossbuchstabeVorhanden==False:
    print("Mindestens ein Großbuchstabe ist erforderlich.")
elif kleinbuchstabeVorhanden==False:
    print("Mindestens ein Kleinbuchstabe ist erforderlich.")
elif zahlVorhanden==False:
    print("Mindestens eine Zahl ist erforderlich.")
elif anzahlZeichen<8:
    print("Das Passwort muss mindestens 8 Zeichen enthalten.")
elif anzahlZeichen>16:
    print("Das Passwort darf höchstens 16 Zeichen enthalten.")
else:
    print("Passwort ist gültig!")
        
    
```

### Aufgabe 3: Perfekte Zahl prüfen

Eine Zahl ist perfekt, wenn die Summe ihrer positiven Teiler ohne die Zahl selbst wieder die Zahl ergibt.

Beispiel: 6 ist perfekt, da 1+2+3=6.

Zusatzaufgabe: Geben Sie alle perfekten Zahlen zwischen 1 und 1000 aus.

```python
summe=0
zuPruefendeZahl=int(input("Bitte geben Sie die zu prüfende Zahl ein:"))
for i in range(1,zuPruefendeZahl):
    if zuPruefendeZahl%i==0:
        summe+=i
if summe==zuPruefendeZahl:
    print("{} ist eine perfekte Zahl.".format(zuPruefendeZahl))
else:
    print("{} ist keine perfekte Zahl.".format(zuPruefendeZahl))
```

```python
for i in range(2,1000):
    summe=0
    for j in range(1,i):
        if i%j==0:
            summe+=j
    if i==summe:
        print("{} ist eine perfekte Zahl.".format(i))
```

### Aufgabe 5: Armstrong-Zahl prüfen

Eine Armstrong-Zahl ist eine Zahl, deren Ziffern hoch drei addiert wieder die Zahl ergeben.

```python
zahl = int(input("Bitte geben Sie die zu prüfende Zahl ein: "))
temporaereZahl = zahl
summe = 0

while temporaereZahl != 0:
    ziffer = temporaereZahl % 10
    summe += ziffer ** 3
    temporaereZahl //= 10

if summe == zahl:
    print("{} ist eine Armstrong-Zahl".format(zahl))
else:
    print("{} ist keine Armstrong-Zahl".format(zahl))
```

```python
for i in range(1,1000):
    temporaereZahl=i
    summe=0
    while temporaereZahl!=0:
        ziffer=temporaereZahl%10
        summe+=ziffer**3
        temporaereZahl//=10
    if i==summe:
        print("{} ist eine Armstrong-Zahl".format(i))
```

### Aufgabe 6: Perfekte Zahl mit break und continue prüfen

```python
zuPruefendeZahl=int(input("Bitte geben Sie die zu prüfende Zahl ein:"))
zaehler=1
summe=0
while True:
    if zaehler==zuPruefendeZahl:
        break
    if zuPruefendeZahl%zaehler==0:
        summe+=zaehler
    zaehler+=1
if summe==zuPruefendeZahl:
    print("{} ist eine perfekte Zahl".format(zuPruefendeZahl))
else:
    print("{} ist keine perfekte Zahl".format(zuPruefendeZahl))
```
