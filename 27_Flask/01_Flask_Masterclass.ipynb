# Theorie

Flask ist ein leichtgewichtiges Webframework auf WSGI-Basis.
Es ist flexibel und ideal fuer Lern- und Produktionsprojekte mit klarer Architektur.

# Erklaerung

Kernbausteine:
- App-Instanz
- Routen
- Request/Response
- Templates (Jinja2)
- Blueprints fuer modulare Struktur

# Syntax

```python
from flask import Flask
app = Flask(__name__)

@app.get('/')
def home():
    return 'Hallo Flask'
```

# Merke

Flask bleibt klein, aber Architekturentscheidungen bleiben trotzdem entscheidend.

# Parameter

- Route-Parameter (`/kurs/<int:id>`)
- Query-Parameter (`?seite=2`)
- Request-Body bei POST

# Rueckgabewert

Flask-Handler geben Text, HTML, JSON oder Response-Objekte zurueck.

````python
# Beispiel 1: Reines Python-Pattern fuer Routing-Denke (ohne Serverstart im Notebook)
def route_home():
    return 'Willkommen'

def route_kurs(kurs_id):
    return {'id': kurs_id, 'titel': 'Python Kurs'}

print(route_home())
print(route_kurs(3))
````

````python
# Beispiel 2: Form-Validierungslogik
def validiere_form(name, email):
    fehler = []
    if not name.strip():
        fehler.append('Name fehlt')
    if '@' not in email:
        fehler.append('E-Mail ungueltig')
    return fehler

print(validiere_form('Ada', 'ada@mail.de'))
print(validiere_form('', 'falsch'))
````

````python
# Beispiel 3: API-Response-Struktur
def api_response(ok, daten=None, fehler=None):
    if ok:
        return {'ok': True, 'daten': daten}
    return {'ok': False, 'fehler': fehler}

print(api_response(True, daten={'id': 1}))
print(api_response(False, fehler='nicht gefunden'))
````
