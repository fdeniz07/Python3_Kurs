## Iteratoren und Generatoren
In diesem Kapitel lernst du, wie Python Schritt für Schritt Werte aus einer Sammlung verarbeitet. Iteratoren liefern die Elemente nacheinander. Generatoren erzeugen Werte bei Bedarf.

### Iterierbare Objekte und Iteratoren

Viele Python-Objekte sind iterierbar. Das bedeutet: Sie können nacheinander durchlaufen werden. Beispiele sind Listen, Tupel und Strings.

```python
farben = ["rot", "grün", "blau"]
for farbe in farben:
    print(farbe)
```

Mit `iter()` erzeugen wir einen Iterator. `next()` liefert das nächste Element des Iterators.

#### Einen Iterator für eine Liste erstellen

```python
farben = ["rot", "grün", "blau"]
iterator = iter(farben)
print(next(iterator))
print(next(iterator))
print(next(iterator))

try:
    print(next(iterator))
except StopIteration:
    print("Ende der Iteration erreicht")
```

#### Einen Iterator für einen String erstellen

```python
myStr="Python"
myIt=iter(myStr)

print(next(myIt))
print(next(myIt))
print(next(myIt))
print(next(myIt))
print(next(myIt))
print(next(myIt))
```

```python
print(next(myIt))
```

#### Verwendung eines Iterators in einer Schleife

```python
myList=["C","C++","Python"]
myIt=iter(myList)

while True:
    try:
        print(next(myIt))
    except:
        break
```

### Eigene Iteratoren

Ein eigener Iterator verwendet `__iter__()` und `__next__()`. So kann Python die Elemente nacheinander abrufen.

```python
class Zaehler:
    def __init__(self, start, ende):
        self.aktuell = start
        self.ende = ende

    def __iter__(self):
        return self

    def __next__(self):
        if self.aktuell > self.ende:
            raise StopIteration
        wert = self.aktuell
        self.aktuell += 1
        return wert

for zahl in Zaehler(3, 6):
    print(zahl)
```

### Was ist ein Generator? Warum brauchen wir ihn?

Im Gegensatz zu normalen Funktionen geben Generatorfunktionen einen Iterator zurück. Generatorfunktionen müssen mindestens eine `yield`-Anweisung enthalten, die anstelle der `return`-Anweisung verwendet wird.

Die Methoden `__iter__()`, `__next__()` sowie die Ausnahme `StopIteration` werden dabei automatisch vom Python-Interpreter bereitgestellt.

#### Generator mit () definieren

```python
zahlListe=[i for i in range(20)]
print(zahlListe)
```

```python
generator=(i for i in range(20))
print(generator)
```

```python
for i in generator:
    print(i)
```

### Generatoren mit `yield`

Generatoren sind einfacher als eigene Iterator-Klassen. Jede Funktion mit `yield` liefert einen Iterator zurück.

```python
def quadratzahlen(limit):
    zahl = 1
    while zahl <= limit:
        yield zahl * zahl
        zahl += 1

for wert in quadratzahlen(5):
    print(wert)
```

Generatoren sind besonders praktisch, wenn viele Werte erzeugt werden sollen, aber nicht alle auf einmal im Speicher stehen müssen.

```python
def fibonacci_generator(anzahl):
    a, b = 0, 1
    for _ in range(anzahl):
        yield a
        a, b = b, a + b

for zahl in fibonacci_generator(7):
    print(zahl)
```

#### Generator, der Zeichenketten zurückgibt

```python
def kuben_ausgeben():
    zahlListe=[i**3 for i in range(50)]
    for i,j in enumerate(zahlListe):
        yield "{}^3:{}".format(i,j)
```

```python
kuben_ausgeben_iter=kuben_ausgeben()
print(next(kuben_ausgeben_iter))
```

```python
for i in kuben_ausgeben():
    print(i)
```
