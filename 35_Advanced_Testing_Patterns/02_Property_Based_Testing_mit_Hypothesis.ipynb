# Kapitel 35.2 - Property-Based Testing mit Hypothesis

Property-Based Testing prueft allgemeine Eigenschaften statt nur einzelner Beispiele.

```python
def normalize_name(name: str) -> str:
    return ' '.join(name.strip().split())

samples = ['  Max  Mustermann ', 'Anna', '  A   B  ' ]
for s in samples:
    out = normalize_name(s)
    assert '  ' not in out
print('basic property checks passed')
```

## Best Practices

- Invarianten zuerst definieren
- Shrinking-Ergebnisse dokumentieren
- kritische Parser und Berechnungen bevorzugt property-basiert testen

## Vertiefung

Dieses Kapitel wird in realen Projekten wirksam, wenn Entscheidungen explizit dokumentiert, Risiken priorisiert und kontinuierlich ueberprueft werden.

```python
# Zusatzanalyse
risks=[{"name":"stability","score":9},{"name":"maintainability","score":8},{"name":"complexity","score":6}]
for r in sorted(risks,key=lambda x:x["score"],reverse=True):
    print(r["name"],r["score"])
```
