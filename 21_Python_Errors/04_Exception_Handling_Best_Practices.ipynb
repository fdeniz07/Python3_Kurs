# Erklaerung

Eine gute Ausnahmebehandlung verhindert Datenverlust und verbessert Wartbarkeit.
Wichtig ist die Entscheidung: behandeln oder weiterwerfen.

# Syntax

```python
try:
    ...
except ValueError as fehler:
    raise
finally:
    cleanup()
```

# Merke

`finally` ist ideal fuer Ressourcenbereinigung.

# Parameter

- Fehlerobjekt
- Kontextdaten (z. B. user_id, dateiname)

# Rueckgabewert

Definiere klar: Rueckgabeobjekt mit Fehlerstatus oder Exception-Propagation.

````python
# Beispiel 1: Spezifische Behandlung
def parse_alter(text):
    try:
        alter = int(text)
        if alter < 0:
            raise ValueError('Alter darf nicht negativ sein')
        return alter
    except ValueError as fehler:
        print('Eingabefehler:', fehler)
        return None

print(parse_alter('20'))
print(parse_alter('-3'))
print(parse_alter('abc'))
````

````python
# Beispiel 2: Ressourcen sauber schliessen
datei = None
try:
    datei = open('beispiel.txt', 'w', encoding='utf-8')
    datei.write('Test')
finally:
    if datei is not None:
        datei.close()
        print('Datei geschlossen')
````

````python
# Beispiel 3: Exception weiterwerfen mit Kontext
def lade_konfiguration(pfad):
    try:
        with open(pfad, 'r', encoding='utf-8') as datei:
            return datei.read()
    except FileNotFoundError as fehler:
        raise FileNotFoundError(f'Konfigurationsdatei fehlt: {pfad}') from fehler
````

````python
# Fehlerbehandlung mit Kontext
def parse_int(value):
    try:
        return int(value)
    except ValueError as exc:
        raise ValueError(f"ungueltige Zahl: {value}") from exc
print(parse_int("42"))
````
