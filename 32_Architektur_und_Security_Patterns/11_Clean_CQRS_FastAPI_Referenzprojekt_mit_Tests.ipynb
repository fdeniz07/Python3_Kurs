# Rueckgabewert

````python
# Beispiel 1: Command-/Query-Ablauf
store = {}
def create_course(store, cid, title):
    store[cid] = {'id': cid, 'title': title}
    return cid
def get_course(store, cid):
    return store.get(cid)

new_id = create_course(store, 1, 'Clean CQRS')
print(get_course(store, new_id))
````

````python
# Beispiel 2: Einfaches globales Error-Mapping
def map_exc(exc):
    if isinstance(exc, ValueError):
        return {'status': 400, 'error': str(exc)}
    return {'status': 500, 'error': 'internal'}

print(map_exc(ValueError('bad input')))
````
