# Kapitel 8 – Widget-Methoden und Eigenschaften

In diesem Kapitel lernst du die wichtigsten Methoden kennen, die fast jedes Tkinter-Widget besitzt.


## Warum sind Methoden wichtig?

Ein Widget wird nicht nur erstellt. Es kann später verändert, gelöscht oder abgefragt werden.

Beispiele:

- Text ändern
- Größe ändern
- Fokus setzen
- Widget löschen
- Position ermitteln


# config() / configure()

## Beschreibung

Eigenschaften eines Widgets ändern.

## Syntax

```python
widget.config()(...)
```

## Beispiel


````python
import tkinter as tk

root=tk.Tk()

label=tk.Label(root,text="Hallo")
label.pack()

label.config(text="Neuer Text",fg="blue",font=("Arial",16))

root.mainloop()
````

### Tipp

Probiere den Code aus und ändere verschiedene Parameter.

### Kleine Übung

Ändere das Beispiel so, dass ein anderer Text oder eine andere Farbe verwendet wird.


# cget()

## Beschreibung

Eine Eigenschaft auslesen.

## Syntax

```python
widget.cget()(...)
```

## Beispiel


````python
import tkinter as tk

root=tk.Tk()

label=tk.Label(root,text="Hallo",fg="red")
label.pack()

print(label.cget("text"))
print(label.cget("fg"))

root.mainloop()
````

### Tipp

Probiere den Code aus und ändere verschiedene Parameter.

### Kleine Übung

Ändere das Beispiel so, dass ein anderer Text oder eine andere Farbe verwendet wird.


# destroy()

## Beschreibung

Widget oder Fenster schließen.

## Syntax

```python
widget.destroy()(...)
```

## Beispiel


````python
import tkinter as tk

root=tk.Tk()

label=tk.Label(root,text="Ich verschwinde")
label.pack()

tk.Button(root,text="Löschen",command=label.destroy).pack()

root.mainloop()
````

### Tipp

Probiere den Code aus und ändere verschiedene Parameter.

### Kleine Übung

Ändere das Beispiel so, dass ein anderer Text oder eine andere Farbe verwendet wird.


# focus_set()

````python
import tkinter as tk

root=tk.Tk()

entry=tk.Entry(root)
entry.pack()

entry.focus_set()

root.mainloop()
````

# update()

## Beschreibung

Fenster sofort aktualisieren.

## Syntax

```python
widget.update()(...)
```

## Beispiel


````python
import tkinter as tk
import time

root=tk.Tk()

label=tk.Label(root,text="Start")
label.pack()

root.update()

time.sleep(2)

label.config(text="Fertig")

root.update()

root.mainloop()
````

### Tipp

Probiere den Code aus und ändere verschiedene Parameter.

### Kleine Übung

Ändere das Beispiel so, dass ein anderer Text oder eine andere Farbe verwendet wird.


# winfo_width() / winfo_height()

## Beschreibung

Größe eines Widgets ermitteln.

## Syntax

```python
widget.winfo_width()(...)
```

## Beispiel


````python
import tkinter as tk

root=tk.Tk()

button=tk.Button(root,text="Button")
button.pack()

root.update()

print(button.winfo_width())
print(button.winfo_height())

root.mainloop()
````

### Kleine Übung

# winfo_x() / winfo_y()

## Beschreibung

Position eines Widgets ermitteln.

## Syntax

```python
widget.winfo_x()(...)
```

## Beispiel


````python
import tkinter as tk

root=tk.Tk()

label=tk.Label(root,text="Hallo")
label.pack(padx=20,pady=20)

root.update()

print(label.winfo_x())
print(label.winfo_y())

root.mainloop()
````

### Tipp

Probiere den Code aus und ändere verschiedene Parameter.

### Kleine Übung

# Häufig verwendete Widget-Methoden

| Methode | Beschreibung |
|---|---|
| config() | Eigenschaften ändern |
| configure() | Eigenschaften ändern |
| cget() | Eigenschaften lesen |
| destroy() | Widget löschen |

| winfo_x() | X-Position |
| winfo_y() | Y-Position |
| winfo_width() | Breite |
| winfo_height() | Höhe |
