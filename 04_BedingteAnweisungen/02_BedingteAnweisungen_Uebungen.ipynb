#### Einen Taschenrechner erstellen

Wir erstellen einen Taschenrechner, mit dem der Benutzer grundlegende Rechenoperationen ausführen kann. Die Ergebnisse von Addition, Subtraktion, Multiplikation, Division, Potenzierung und Modulo-Berechnung sollen berechnet werden können.

Bedingung 1 -> Der eingegebene Operator muss einer der folgenden sein: (+,-,*,/,^,%)

```python
menueText="""
		 Willkommen beim Taschenrechner
		 1. Addition: '+'
		 2. Subtraktion: '-'
		 3. Multiplikation: '*'
		 4. Division: '/'
		 5. Potenzierung: '^'
		 6. Modulo: '%'
"""
print(menueText)

operation=input("Bitte geben Sie den Operator ein:")

if operation=='+' or operation=='-' or operation=='*' or operation=='/' or operation=='^' or operation=='%':
    zahl1,zahl2=map(int,input("Bitte geben Sie zwei Zahlen nacheinander ein:").split(","))
    if operation=='+':
        print("Ergebnis:",zahl1+zahl2)
    elif operation=='-':
        print("Ergebnis:",zahl1-zahl2)
    elif operation=='*':
        print("Ergebnis:",zahl1*zahl2)
    elif operation=='/':
        print("Ergebnis:",zahl1/zahl2)
    elif operation=='^':
        print("Ergebnis:",zahl1**zahl2)
    elif operation=='%':
        print("Ergebnis:",zahl1%zahl2)
```

#### Ein Geldautomatenprogramm erstellen

Wir erstellen ein Geldautomatenprogramm mit einem bestimmten Kontostand und einem täglichen Auszahlungslimit. Der Benutzer kann Geld abheben, einzahlen und den Kontostand anzeigen.

Bedingung 1 -> Der gewünschte Auszahlungsbetrag darf den Kontostand nicht überschreiten.

Bedingung 2 -> Der gewünschte Auszahlungsbetrag darf das tägliche Auszahlungslimit nicht überschreiten.

```python
kontostand,tageslimit=1250.0,500.0
menueText="""
		 Willkommen am Geldautomaten
		 1. Geld abheben: 1
		 2. Geld einzahlen: 2
		 3. Kontostand anzeigen: 3
"""
print(menueText)

benutzerEingabe=int(input("Bitte wählen Sie eine Aktion: "))

if benutzerEingabe==1:
    print("Auszahlungsmenü")
    betrag=int(input("Bitte geben Sie den gewünschten Auszahlungsbetrag ein:"))
    if betrag>kontostand:
        print("Der Auszahlungsbetrag darf den Kontostand nicht überschreiten!")
    elif betrag>tageslimit:
        print("Der Auszahlungsbetrag darf das Tageslimit nicht überschreiten!")
    else:
        kontostand-=betrag
    print("Aktueller Kontostand:",kontostand)
elif benutzerEingabe==2:
    print("Einzahlungsmenü")
    betrag=int(input("Bitte geben Sie den Einzahlungsbetrag ein:"))
    if betrag<=0:
        print("Der Einzahlungsbetrag muss größer als 0 sein!")
    else:
        kontostand+=betrag
    print("Aktueller Kontostand:",kontostand)
elif benutzerEingabe==3:
    print("Aktueller Kontostand:",kontostand)
```
