````python
# Beispiel 1: Einfache Coverage-Idee durch Branchesdef klassifiziere_punktzahl(punkte):    if punkte < 0:        raise ValueError('ungueltig')    if punkte < 50:        return 'nicht bestanden'    if punkte < 75:        return 'bestanden'    return 'gut'print(klassifiziere_punktzahl(80))
````

````python
# Beispiel 2: Branch-Abdeckung ueber Testfaellefaelle = [    (10, 'nicht bestanden'),    (60, 'bestanden'),    (90, 'gut')]for punkte, erwartet in faelle:    assert klassifiziere_punktzahl(punkte) == erwartetprint('Branch-Faelle abgedeckt')
````
