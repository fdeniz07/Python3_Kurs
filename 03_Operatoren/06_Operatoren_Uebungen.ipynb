## Aufteilen eines eingegebenen Geldbetrags in Banknoten

Eingegebener Betrag: 1480

Ursprünglich eingegebener Betrag: 1480 € = Es gibt 7 Stück 200 €.

Verbleibender Betrag: 80 € = Es gibt 0 Stück 100 €.

Verbleibender Betrag: 80 € = Es gibt 1 Stück 50 €.

Verbleibender Betrag: 30 € = Es gibt 1 Stück 20 €.

Verbleibender Betrag: 10 € = Es gibt 1 Stück 10 €.

Verbleibender Betrag: 0 € = Es gibt 0 Stück 5 €.

```python
eingegebenerGeldbetrag = int(input("Bitte geben Sie einen Geldbetrag ein:"))
print("{} Stueck 200 € sind vorhanden".format(eingegebenerGeldbetrag // 200))
eingegebenerGeldbetrag %= 200 # eingegebenerGeldbetrag = eingegebenerGeldbetrag % 200
print("{} Stueck 100 € sind vorhanden".format(eingegebenerGeldbetrag // 100))
eingegebenerGeldbetrag %= 100
print("{} Stueck 50 € sind vorhanden".format(eingegebenerGeldbetrag // 50))
eingegebenerGeldbetrag %= 50
print("{} Stueck 20 € sind vorhanden".format(eingegebenerGeldbetrag // 20))
eingegebenerGeldbetrag %= 20
print("{} Stueck 10 € sind vorhanden".format(eingegebenerGeldbetrag // 10))
eingegebenerGeldbetrag %= 10
print("{} Stueck 5 € sind vorhanden".format(eingegebenerGeldbetrag // 5))
eingegebenerGeldbetrag %= 5
```

## Umkehren einer eingegebenen 3-stelligen Zahl


Eingegebene Zahl: 123
    
Zahl in der ersten Iteration: 123, gedruckte Ziffer: 3
        
Zahl in der zweiten Iteration: 12, gedruckte Ziffer: 2
        
Zahl in der dritten Iteration: 1, gedruckte Ziffer: 1

```python
eingegebeneZahl = int(input("Bitte geben Sie die Zahl ein, die umgekehrt ausgegeben werden soll:"))
temporaereZahl = eingegebeneZahl # Einer temporaeren Variable zugewiesen

# Um die Ziffer 3 auszugeben
print(temporaereZahl % 10, end="")
temporaereZahl = temporaereZahl // 10 # temporaereZahl //= 10

# Um die Ziffer 2 auszugeben
print(temporaereZahl % 10, end="")
temporaereZahl = temporaereZahl // 10 

# Um die Ziffer 1 auszugeben
print(temporaereZahl % 10, end="\n")
temporaereZahl = temporaereZahl // 10 

print(temporaereZahl)
print(eingegebeneZahl)
```
