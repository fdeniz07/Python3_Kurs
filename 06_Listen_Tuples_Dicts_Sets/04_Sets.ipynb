## Mengen (Sets)

#### Warum brauchen wir Mengen?
Mengen sind das Äquivalent zu den mathematischen Mengen in der Programmiersprache Python. Ähnlich wie Listen und Tupel ermöglichen sie das gemeinsame Speichern mehrerer Elemente. Im Gegensatz zu Tupeln sind Mengen jedoch veränderbare Strukturen.

#### Wie werden Mengen definiert?
1. Mit dem Schlüsselwort `set()`
2. Mit geschweiften Klammern `{}`. Hierbei müssen jedoch direkt Werte zugewiesen werden, da andernfalls eine Dictionary-Struktur (Sözlük) entsteht.

##### Mengen definieren

```python
mengeA = set()
mengeB = {} # Dies erstellt ein leeres Dictionary, keine Menge!
mengeC = {1, 2, 3, 4, 5}
mengeD = set([1, 2, 3, 4, 5, 6, 7])
mengeE = {1, 1, 1, 1, 2, 3, 4, 5} # Jedes Element kann in einer Menge nur einmal vorkommen

print(type(mengeA))
print(type(mengeB))
print(type(mengeC))
print(type(mengeD))
print(type(mengeE))
print(mengeD)
```

#### WICHTIG
Mengen besitzen keine Indexierung! Die Elemente werden ungeordnet im Speicher abgelegt. Daher kann auf die Elemente einer Menge nicht über einen Index (wie z. B. `menge[0]`) zugegriffen werden.

```python
testMenge = {1, 2, 3, 4, 5}
# print(testMenge[0]) # Dies wirft einen Fehler (TypeError)
for element in testMenge:
    print(element)
```

#### Methoden für Mengen

1. `add()`-Methode: Wird verwendet, um ein neues Element zur Menge hinzuzufügen
2. `discard()`-Methode: Wird verwendet, um ein Element aus der Menge zu löschen. Wenn das Element nicht existiert, passiert nichts (kein Fehler)
3. `remove()`-Methode: Wird verwendet, um ein Element aus der Menge zu löschen. Wenn das Element nicht existiert, wird ein Fehler geworfen

```python
meineMenge = {1, 2, 3, 4, 5}
print(meineMenge)
meineMenge.add(10)
print(meineMenge)
meineMenge.discard(10)
meineMenge.discard(25) # Verursacht keinen Fehler, obwohl 25 nicht existiert
print(meineMenge)
# meineMenge.remove(25) # Dies wuerde einen KeyError werfen
```

#### Mathematische Mengenoperationen in Python

1. `intersection()`-Methode: Wird verwendet, um die Schnittmenge (gemeinsame Elemente) zweier Mengen zu finden
2. `isdisjoint()`-Methode: Überprüft, ob zwei Mengen disjunkt sind (d. h. ob ihre Schnittmenge leer ist)
3. `issubset()`-Methode: Überprüft, ob eine Menge eine Teilmenge einer anderen Menge ist
4. `issuperset()`-Methode: Überprüft, ob eine Menge eine Obermenge einer anderen Menge ist
5. `union()`-Methode: Wird verwendet, um die Vereinigungsmenge zweier Mengen zu bilden

```python
menge1 = {1, 2, 3, 4}
menge2 = {1, 2, 3, 4, 5, 6}

print(menge1.intersection(menge2))
print(menge2.intersection(menge1))

print(menge1.isdisjoint(menge2))
print(menge2.isdisjoint(menge1))

print(menge1.issubset(menge2))
print(menge2.issubset(menge1))

print(menge1.issuperset(menge2))
print(menge2.issuperset(menge1))

print(menge1.union(menge2))
print(menge2.union(menge1))
print(menge1 | menge2) # Alternatives Zeichen fuer die Vereinigungsmenge
```
