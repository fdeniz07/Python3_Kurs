### Aufgabe 1: Anzahl der Stellen berechnen

Aufgabenbeschreibung:

Berechnen Sie die Anzahl der Stellen einer zufällig erzeugten Zahl mit maximal 5 Stellen.


```python
import random
erzeugteZahl=random.randint(1,99999)

if erzeugteZahl<10:
    print("{} hat 1 Stelle.".format(erzeugteZahl))
elif erzeugteZahl<100:
    print("{} hat 2 Stellen.".format(erzeugteZahl))
elif erzeugteZahl<1000:
    print("{} hat 3 Stellen.".format(erzeugteZahl))
elif erzeugteZahl<10000:
    print("{} hat 4 Stellen.".format(erzeugteZahl))
else:
    print("{} hat 5 Stellen.".format(erzeugteZahl))
```

### Aufgabe 2: Einkaufs- und Preisberechnung

Aufgabenbeschreibung:

Passen Sie den Preis abhängig davon an, ob der Benutzer eine Ratenzahlung wünscht und ob eine Kundenkarte vorhanden ist.

Bedingung 1 -> Bei 3 Raten +3 %, bei 6 Raten +6 %, bei 9 Raten +9 %.

Bedingung 2 -> Mit Kundenkarte wird ein Rabatt von 10 % gewährt.


```python
import random

einkaufsbetrag = random.randint(1, 10000)
print("Einkaufsbetrag: {}€".format(einkaufsbetrag))

ratenzahlung = input("Möchten Sie eine Ratenzahlung? (J-N) ")

if ratenzahlung == 'J' or ratenzahlung == 'j':
    print("Ratenzahlungsmenü")

    anzahlRaten = int(input("Wie viele Raten möchten Sie (3-6-9)? "))

    if anzahlRaten == 3:
        print("3 Raten werden angewendet...")
        einkaufsbetrag += einkaufsbetrag * 0.03
        print("Betrag nach Ratenzuschlag: {}€".format(einkaufsbetrag))

    elif anzahlRaten == 6:
        print("6 Raten werden angewendet...")
        einkaufsbetrag += einkaufsbetrag * 0.06
        print("Betrag nach Ratenzuschlag: {}€".format(einkaufsbetrag))

    elif anzahlRaten == 9:
        print("9 Raten werden angewendet...")
        einkaufsbetrag += einkaufsbetrag * 0.09
        print("Betrag nach Ratenzuschlag: {}€".format(einkaufsbetrag))

    else:
        print("Ungültige Anzahl von Raten!")

else:
    print("Keine Ratenzahlung wird angewendet!")

kundenkarte = input("Besitzen Sie eine Kundenkarte? (J-N) ")

if kundenkarte == 'J' or kundenkarte == 'j':
    print("Rabattmenü")
    einkaufsbetrag -= einkaufsbetrag * 0.10
    print("Betrag nach Rabatt: {}€".format(einkaufsbetrag))
else:
    print("Es wurde kein Rabatt angewendet!")

print("Endbetrag: {}€".format(einkaufsbetrag))
```

### Aufgabe 3: Prüfung der Dreiecksbedingung

Aufgabenbeschreibung:

Lesen Sie die drei Seitenlängen eines Dreiecks ein und prüfen Sie, ob mit diesen Werten ein Dreieck gebildet werden kann.

Bedingung 1 -> Für x: |y-z| < x < y+z

Bedingung 2 -> Für y: |x-z| < y < x+z

Bedingung 3 -> Für z: |x-y| < z < x+y


```python
x,y,z=map(int,input("Bitte geben Sie die drei Seitenlängen durch Kommas getrennt ein: ").split(","))

if abs(y-z)>=x or x>=y+z:
    print("Die Seite x erfüllt die Dreiecksbedingung nicht!")
elif abs(x-z)>=y or y>=x+z:
    print("Die Seite y erfüllt die Dreiecksbedingung nicht!")
elif abs(x-y)>=z or z>=x+y:
    print("Die Seite z erfüllt die Dreiecksbedingung nicht!")
else:
    print("Ein Dreieck kann gebildet werden!")
```
