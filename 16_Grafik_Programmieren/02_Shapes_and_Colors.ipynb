# Formen und Farben
Dieses Notebook zeigt, wie man verschiedene Formen auf einem Canvas zeichnet und Farben verwendet.

```python
import tkinter as tk

fenster = tk.Tk()
fenster.title('Formen und Farben')

canvas = tk.Canvas(fenster, width=500, height=350, bg='white')
canvas.pack()

# Rechteck mit Füllfarbe
canvas.create_rectangle(30, 30, 180, 140, outline='black', fill='#88ccee', width=2)

# Oval (Ellipse)
canvas.create_oval(220, 30, 380, 140, outline='darkred', fill='pink', width=2)

# Polygon (Dreieck)
canvas.create_polygon(50, 200, 150, 250, 20, 300, outline='green', fill='lightgreen')

# Bogen
canvas.create_arc(220, 170, 380, 300, start=0, extent=180, fill='orange')

fenster.mainloop()
```
