## Tupel (Tuples)

#### Warum brauchen wir Tupel?

Wir verwenden Tupel, um eine große Anzahl von Variablen oder Werten zusammen zu speichern und einfacher auf sie zugreifen zu können. Es ist eine Struktur, die es uns ermöglicht, mehrere Werte desselben oder unterschiedlicher Typen zusammenzuhalten.

### WICHTIG
Der wichtigste Unterschied zu Listen besteht darin, dass Tupel nach ihrer Erstellung unveränderlich (immutable) sind. Das bedeutet, es können keine neuen Elemente hinzugefügt, aktualisiert oder gelöscht werden.

#### Wie werden Tupel definiert?
1. Mit dem Schlüsselwort `tuple()`
2. Mit runden Klammern `()`

##### Ein leeres Tupel definieren

```python
tupelA = tuple()
tupelB = ()
print(type(tupelA))
print(type(tupelB))
```

##### Ein Integer-Tupel definieren

```python
zahlenTupel = (1, 2, 3)
print(type(zahlenTupel))
print(zahlenTupel)
```

##### Ein String-Tupel definieren

```python
textTupel = ("Python", "C", "C++", "Perl", "Java")
print(type(textTupel))
print(textTupel)
```

##### Ein gemischtes Tupel definieren

```python
gemischtesTupel = (1, 2, "Python", "C", 3, "Perl", 3.5, False, True, "Java", [1, 2, 3])
print(gemischtesTupel)
```

##### Einzelner Zugriff auf Tupelelemente

```python
meinTupel = (1, 2, 3, 4, 5)
print(type(meinTupel))

print(meinTupel[0])
print(meinTupel[1])
print(meinTupel[2])
print(meinTupel[-2])
print(meinTupel[-1])
```

##### Zugriff auf Tupelelemente mit einer Schleife

```python
for element in meinTupel:
    print(element**2)
```

##### Die Anzahl der Elemente eines Tupels ermitteln

```python
print(len(meinTupel))
```

##### Zugriff auf die Indizes der Tupelelemente

```python
for i in range(len(meinTupel)):
    print(meinTupel[i])
```

##### Slicing (Teilen) eines Tupels
`tupel[start_index:end_index:schrittweite]`

```python
basisTupel = (1, 2, 3, 4, 5)
teilTupelA = basisTupel[3:]
teilTupelB = basisTupel[:3]
teilTupelC = basisTupel[::2]
teilTupelD = basisTupel[::-1]
print(basisTupel)
print(teilTupelA)
print(teilTupelB)
print(teilTupelC)
print(teilTupelD)
```

##### Überprüfung von Elementen in einem Tupel
1. `in`-Operator: Gibt True zurück, wenn das Element im Tupel existiert, andernfalls False.

2. `not in`-Operator: Gibt False zurück, wenn das Element im Tupel existiert, andernfalls True.

3. `index`-Methode: Wenn das gesuchte Element im Tupel existiert, wird der entsprechende Index zurückgegeben. Existiert das Element nicht, wird ein ValueError geworfen.

```python
suchTupel = (10, 20, 30, 40, 50)
print(10 in suchTupel)
print(25 in suchTupel)

print(10 not in suchTupel)
print(25 not in suchTupel)

print(suchTupel.index(30))
if 60 in suchTupel:
    print(suchTupel.index(60))
```

##### Das kleinste, größte Element und die Summe eines Tupels ermitteln

`min`-Funktion: Gibt das kleinste Element des Tupels zurück

`max`-Funktion: Gibt das größte Element des Tupels zurück

`sum`-Funktion: Gibt die Gesamtsumme aller Elemente im Tupel zurück

```python
werteTupel = (1, 5, 96, 32, 543, 65)
print(min(werteTupel))
print(max(werteTupel))
print(sum(werteTupel))
```
