## Übung
Erstelle eine kleine Factory-Funktion für Fahrzeuge.

```python
def create_fahrzeug(typ):
    if typ == "auto":
        return "Auto"
    if typ == "bike":
        return "Bike"
    return "Unbekannt"

print(create_fahrzeug("bike"))
```
