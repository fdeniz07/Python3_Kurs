# Kapitel 16 – Grafik Programmieren
In diesem Kapitel lernst du, wie Python einfache Grafiken mit einem Canvas zeichnet. Wir erstellen Linien, Formen und Elemente, die sich in einem Fenster anzeigen lassen.

## Einfache Grafiken mit Tkinter
Tkinter ist gut geeignet, um grundlegende Grafiken zu zeichnen. Das zentrale Element dafür ist das `Canvas`-Widget.

```python
import tkinter as tk

fenster = tk.Tk()
fenster.title('Grafik Programmieren')

canvas = tk.Canvas(fenster, width=400, height=300, bg='white')
canvas.pack()

canvas.create_line(50, 50, 350, 50, fill='blue', width=3)
canvas.create_rectangle(50, 80, 200, 180, outline='green', width=2)
canvas.create_oval(220, 80, 340, 180, outline='red', width=2)

fenster.mainloop()
```

## Linien und Formen
Mit `create_line`, `create_rectangle` und `create_oval` zeichnest du einfache Formen auf das Canvas.

```python
import tkinter as tk

fenster = tk.Tk()
fenster.title('Linien und Formen')

canvas = tk.Canvas(fenster, width=400, height=300, bg='lightyellow')
canvas.pack()

canvas.create_line(20, 20, 380, 20, fill='black', width=2)
canvas.create_rectangle(50, 60, 180, 140, outline='blue', fill='lightblue')
canvas.create_oval(220, 60, 340, 140, outline='purple', fill='lavender')
canvas.create_arc(50, 170, 180, 250, start=0, extent=150, fill='orange')

fenster.mainloop()
```

## Text und Farben
Mit `create_text` und Farboptionen kannst du Beschriftungen und farbige Elemente hinzufügen.

```python
import tkinter as tk

fenster = tk.Tk()
fenster.title('Text und Farben')

canvas = tk.Canvas(fenster, width=400, height=300, bg='white')
canvas.pack()

canvas.create_rectangle(30, 30, 150, 120, fill='skyblue', outline='blue')
canvas.create_text(90, 75, text='Hallo', font=('Arial', 18, 'bold'), fill='darkblue')
canvas.create_text(250, 150, text='Grafik mit Python', font=('Arial', 16), fill='green')
canvas.create_oval(220, 190, 320, 260, fill='red', outline='black')

fenster.mainloop()
```

## Animation mit wiederholtem Zeichnen
Du kannst einfache Animationen erreichen, indem du das Canvas mehrfach aktualisierst.

```python
import tkinter as tk

fenster = tk.Tk()
fenster.title('Einfache Animation')

canvas = tk.Canvas(fenster, width=400, height=300, bg='white')
canvas.pack()

kreis = canvas.create_oval(10, 130, 50, 170, fill='red')

def bewegen():
    canvas.move(kreis, 5, 0)
    pos = canvas.coords(kreis)
    if pos[2] < 400:
        fenster.after(50, bewegen)

bewegen()
fenster.mainloop()
```
