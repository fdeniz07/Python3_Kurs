# Kapitel 17.3 — Vererbung (Inheritance)

## Lernziele
- Vererbung verstehen.
- Eine Unterklasse von einer Oberklasse ableiten.
- Methoden überschreiben.

## Theorie
Vererbung bedeutet: Eine neue Klasse übernimmt Eigenschaften und Methoden einer bestehenden Klasse.
So können wir allgemeine Merkmale einmal definieren und in speziellen Klassen wiederverwenden.

In Python ist Vererbung oft sehr flexibel. Eine Unterklasse kann Methoden der Oberklasse übernehmen oder überschreiben.

## Duck Typing in Python
Ein typischer Python-Gedanke ist Duck Typing: Es ist nicht wichtig, von welcher Klasse ein Objekt ist, sondern ob es die benötigten Methoden besitzt.

```python
class Druckbar:
    def drucken(self):
        raise NotImplementedError

class PapierDrucker:
    def drucken(self):
        print("Drucke auf Papier")

class BildschirmDrucker:
    def drucken(self):
        print("Drucke auf dem Bildschirm")

for ziel in [PapierDrucker(), BildschirmDrucker()]:
    ziel.drucken()
```

````python
class Fahrzeug:
    def __init__(self, marke):
        self.marke = marke

    def beschreiben(self):
        print(f"Das Fahrzeug ist eine {self.marke}.")

class Auto(Fahrzeug):
    def __init__(self, marke, tueren):
        super().__init__(marke)
        self.tueren = tueren

    def beschreiben(self):
        print(f"Das Auto {self.marke} hat {self.tueren} Türen.")

a = Auto("VW", 4)
a.beschreiben()
````

## Beispiel 1
Die Unterklasse nutzt die Oberklasse und ergänzt sie.

````python
class Tier:
    def __init__(self, name):
        self.name = name

    def laut(self):
        print("Das Tier macht ein Geräusch.")

class Katze(Tier):
    def laut(self):
        print("Miau!")

katze = Katze("Minka")
katze.laut()
````

## Praxisbeispiel
Eine allgemeine Klasse `Mitarbeiter` und eine spezielle Klasse `Manager`.

````python
class Mitarbeiter:
    def __init__(self, name):
        self.name = name

    def arbeite(self):
        print(f"{self.name} arbeitet.")

class Manager(Mitarbeiter):
    def arbeite(self):
        print(f"{self.name} leitet das Team.")

m = Manager("Sofia")
m.arbeite()
````

## Häufige Fehler
- `super()` vergessen
- Methoden nicht korrekt überschreiben
- Falsche Vererbungshierarchie

## Best Practice
Vererbung sollte nur genutzt werden, wenn ein „ist ein“-Verhältnis besteht.

## Übung
Lege eine Oberklasse `Form` und eine Unterklasse `Kreis` an.

````python
class Form:
    def flaeche(self):
        return 0

class Kreis(Form):
    def __init__(self, radius):
        self.radius = radius

    def flaeche(self):
        return 3.14 * self.radius ** 2

k = Kreis(3)
print(k.flaeche())
````

## Zusammenfassung
Vererbung erlaubt es, gemeinsame Merkmale an Unterklassen weiterzugeben und sie gezielt zu erweitern.
