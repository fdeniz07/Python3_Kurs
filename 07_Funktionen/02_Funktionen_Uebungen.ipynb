#### Funktion zum Ersetzen eines bestimmten Zeichens in einem Satz
Funktionsbeschreibung:

Die Funktion nimmt die Parameter `satz`, `altesZeichen` und `neuesZeichen` entgegen. Sie ersetzt jedes Vorkommen des alten Zeichens im Satz durch das neue Zeichen und gibt den aktualisierten Satz (als Liste) zurück.

Beispiel:

- Satz = "Bücher", altes Zeichen = "ue", neues Zeichen = "-"
- Neuer Liste = "['B', 'ue', 'c', 'h', 'e', 'r']"
- Neuer Satz = "Buecher"

```python
def ersetzeZeichen(satz, altesZeichen, neuesZeichen):
    satz = list(satz) # In eine Liste umwandeln
    for index in range(len(satz)):
        if satz[index] == altesZeichen:
            satz[index] = neuesZeichen
    return satz # Gibt eine Liste zurück
```

```python
def schreibeTextAus(datenListe):
    # Liste in einen String umwandeln und ausgeben
    liste_str = str(datenListe)
    #print(f"Neuer Satz = \"{liste_str}\"")
    
    # Die Liste in einen String umwandeln und zurückgeben
    return "".join(datenListe)
```

```python
eingebenerSatz = input("Bitte geben Sie den zu ändernden Satz ein:")
eingabe = input("Bitte geben Sie das alte und neue Zeichen (mit Komma getrennt) ein: ")
altesBuchstabe, neuesBuchstabe = eingabe.split(",") # Eingabe in altes und neues Zeichen aufteilen

ergebnis = ersetzeZeichen(eingebenerSatz, altesBuchstabe, neuesBuchstabe)
print(f"Neuer Liste = \"{str(ergebnis)}\"")
print(f"Neuer Satz = \"{schreibeTextAus(ergebnis)}\"")
```

#### Funktion zur Großschreibung des jeweils ersten Buchstabens eines Satzes

Funktionsbeschreibung: Die Funktion nimmt einen Text von außen entgegen, wandelt den jeweils ersten Buchstaben nach einem Satzanfang (oder nach einem Punkt) in einen Großbuchstaben um und gibt das Ergebnis zurück.

Beispiel:

Der Satz: "fatih deniz python kursuna katiliyor.lutfen dikkatli izle"

wird umgewandelt in: "Fatih deniz python kursuna katiliyor.Lutfen dikkatli izle"

```python
def konvertiereZuGrossbuchstaben(satz):
    satz = list(satz) # In eine Liste umwandeln
    satz[0] = satz[0].upper()
    for index in range(len(satz)):
        if satz[index] == '.':
            if index + 1 == len(satz):
                return satz
            index += 1
            satz[index] = satz[index].upper()
    return satz
```

```python
eingebenerSatz = input("Bitte geben Sie die Sätze ein, die korrigiert werden sollen:")
eingebenerSatz = str(konvertiereZuGrossbuchstaben(eingebenerSatz))
print(eingebenerSatz)
```

#### Anagramm-Prüffunktion

Funktionsbeschreibung:

Die Funktion nimmt zwei Wörter von außen entgegen und prüft, ob diese Anagramme voneinander sind. Wenn es ein Anagramm ist, gibt sie `True` zurück, andernfalls `False`.

Anagramm: Ein Wort, das durch Umstellung der Buchstaben eines anderen Wortes entsteht (z. B. "asya" und "yasa").

```python
def istAnagramm(wort1, wort2):
    if len(wort1) != len(wort2):
        return False
    wort1Liste, wort2Liste = list(wort1), list(wort2)
    wort1Liste.sort()
    wort2Liste.sort()
    print(wort1Liste)
    print(wort2Liste)
    for index in range(len(wort1Liste)):
        if wort1Liste[index] != wort2Liste[index]:
            return False
    return True
```

```python
print(istAnagramm("asya", "yasa"))
print("======================")
print(istAnagramm("irak", "raki"))
```

#### Palindrom-Prüffunktion

Funktionsbeschreibung:

Die Funktion nimmt eine Zeichenkette (String) von außen entgegen und prüft, ob es sich um ein Palindrom handelt. Wenn es ein Palindrom ist, wird `True` zurückgegeben, andernfalls `False`.

Palindrom: Ein Wort oder eine Zeichenkette, die vorwärts wie rückwärts gelesen dasselbe ergibt (z. B. "lagerregal" oder "abccba").

```python
def istPalindrom(text):
    textInvertiert = text[::-1]
    if text == textInvertiert:
        return True
    return False
```

```python
print(istPalindrom("kabak"))
```

#### Rekursive Fibonacci-Funktion

Funktionsbeschreibung:

Die Funktion nimmt eine Zahl entgegen und gibt den entsprechenden Wert der Fibonacci-Folge an dieser Stelle zurück.

Beispiel für die Zahl 4:

Zuerst wird die Funktion mit der Zahl 4 aufgerufen, sodass die Variable `zahl` den Wert 4 erhält.

Funktionsaufruf -> return recursiveFibonacci(3) + recursiveFibonacci(2)    

Funktionsaufruf -> return recursiveFibonacci(2) + recursiveFibonacci(1)

Funktionsaufruf -> return recursiveFibonacci(1) + recursiveFibonacci(0)


Das bedeutet:

recursiveFibonacci(0) Ergebnis -> 0

recursiveFibonacci(1) Ergebnis -> 1

recursiveFibonacci(2) Ergebnis -> recursiveFibonacci(1) + recursiveFibonacci(0) = 1

recursiveFibonacci(3) Ergebnis -> recursiveFibonacci(2) + recursiveFibonacci(1) = 2

recursiveFibonacci(4) Ergebnis -> recursiveFibonacci(3) + recursiveFibonacci(2) = 3

```python
def recursiveFibonacci(zahl):
    if zahl == 0 or zahl == 1:
        return zahl
    return recursiveFibonacci(zahl - 1) + recursiveFibonacci(zahl - 2)
```

```python
for i in range(10):
    print(recursiveFibonacci(i))
```
