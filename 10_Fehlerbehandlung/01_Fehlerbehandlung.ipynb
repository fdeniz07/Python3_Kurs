## Fehlerbehandlung

#### Warum brauchen wir das?

Damit der von uns geschriebene Code reibungslos läuft, müssen wir alle potenziellen Fehler, die im Code auftreten können, im Voraus bedenken und entsprechende Maßnahmen ergreifen. Wenn beispielsweise in einem Projekt, bei dem wir Daten aus dem Internet abrufen, die Internetverbindung abbricht, wird die Ausführung des Programms unterbrochen. Auf solche Situationen müssen wir im Vorfeld vorbereitet sein und Lösungen anbieten.

#### Fehlerarten

1. Syntaxfehler (Syntax Error): Tritt auf, wenn wir Code falsch schreiben. In diesen Fällen gibt uns der Compiler direkt einen Fehler aus, den wir korrigieren können.

2. Logischer Fehler (Logic Error): Dies sind Fehler im Ablauf des Codes. Da der Compiler uns hierbei keine Warnung ausgibt, sind diese Fehler schwieriger zu finden.

#### Ausnahmetypen (Exception Types):

AssertionError, AttributeError, EOFError, FloatingPointError, GeneratorExit, ImportError, IndexError, KeyError, KeyboardInterrupt, MemoryError, NameError,
NotImplementedError, OSError, OverflowError, ReferenceError, RuntimeError, StopIteration, SyntaxError, IndentationError, TabError, SystemError, SystemExit,
TypeError, UnboundLocalError, UnicodeError, UnicodeEncodeError, UnicodeDecodeError, UnicodeTranslateError, ValueError, ZeroDivisionError


#### Fehler abfangen (Try-Except-Struktur)

Die Struktur, bei der eine Fehlerwahrscheinlichkeit besteht, wird in den Try-Block geschrieben. Der Ausdruck, der ausgeführt wird, wenn ein Fehler auftritt, wird in den Except-Block geschrieben.

```python
try:
    sayi1 = int(input("Bitte geben Sie die erste Zahl ein:"))
    sayi2 = int(input("Bitte geben Sie die zweite Zahl ein:"))
    print(sayi1 / sayi2)
except:
    print("Ungültiger Wert eingegeben!")
```

```python
try:
    sayi1 = int(input("Bitte geben Sie die erste Zahl ein:"))
    sayi2 = int(input("Bitte geben Sie die zweite Zahl ein:"))
    print(sayi1 / sayi2)
except TypeError:
    print("Divisionsoperation ist nicht geeignet!")
except ValueError:
    print("Sie müssen eine Zahl eingeben!")
except ZeroDivisionError:
    print("Die zweite Zahl darf nicht 0 sein!")
```

```python
try:
    sayi1 = int(input("Bitte geben Sie die erste Zahl ein:"))
    sayi2 = int(input("Bitte geben Sie die zweite Zahl ein:"))
    print(sayi1 / sayi2)
except TypeError:
    print("Divisionsoperation ist nicht geeignet!")
except ValueError:
    print("Sie müssen eine Zahl eingeben!")
except:
    print("Ein Fehler ist aufgetreten!")
```

### ACHTUNG
Der allgemeine `except:`-Block muss am Ende stehen! Wenn Sie ihn an den Anfang setzen, erhalten Sie einen Syntaxfehler!

```python
try:
    sayi1 = int(input("Bitte geben Sie die erste Zahl ein:"))
    sayi2 = int(input("Bitte geben Sie die zweite Zahl ein:"))
    print(sayi1 / sayi2)
except:
    print("Ein Fehler ist aufgetreten!")
except TypeError:
    print("Divisionsoperation ist nicht geeignet!")
except ValueError:
    print("Sie müssen eine Zahl eingeben!")
```

#### Finally-Ausdruck
Dies ist der Codeblock, der unabhängig davon, ob während der Programmausführung unvorhergesehene Fehler auftreten veya nicht, definitiv ausgeführt wird.

```python
try:
    sayi1 = int(input("Bitte geben Sie die erste Zahl ein:"))
    sayi2 = int(input("Bitte geben Sie die zweite Zahl ein:"))
    print(sayi1 / sayi2)
except TypeError:
    print("Divisionsoperation ist nicht geeignet!")
except ValueError:
    print("Sie müssen eine Zahl eingeben!")
except:
    print("Ein Fehler ist aufgetreten!")
finally:
    print("Das Programm wird beendet..")
```

#### Eigene Ausnahmen definieren (Benutzerdefinierte Fehler)

In einigen Fällen reichen uns die vordefinierten Fehlermeldungen möglicherweise nicht aus. Daher müssen wir in unserem Code eventuell eigene Fehler erzeugen. Wir können dies auf zwei verschiedene Arten tun;
1. Mit raise Exception(): Wird direkt im allgemeinen except-Block abgefangen.
2. Mit raise vordefinierte_fehlerklasse(): Da vordefinierte Fehlerklassen verwendet werden, wird der Fehler spezifisch abgefangen. Der Fehler, den Sie erstellen, sollte mit der vordefinierten Fehlerklasse geschrieben werden, die am besten dazu passt.

```python
try:
    sayi1 = int(input("Bitte geben Sie eine positive Zahl ein:"))
    if sayi1 < 0:
        raise Exception()
except:
    print("Die eingegebene Zahl darf nicht negativ sein!")
```

```python
try:
    sayi2 = int(input("Bitte geben Sie eine positive Zahl ein:"))
    if sayi2 < 0:
        raise ValueError("Die eingegebene Zahl darf nicht negativ sein!")
except TypeError:
    print("Ungültiger Typ")
except ValueError:
    print("Die eingegebene Zahl darf nicht negativ sein!")
```

#### Assertion-Ausdruck

Mit dem Assert-Ausdruck können wir Bedingungen definieren, bei denen wir uns über die Richtigkeit veya Falschheit nicht sicher sind. Wenn die Bedingung falsch ist, erhalten wir einen AssertionError. Wenn sie wahr ist, läuft das Programm ohne Fehler weiter. Die Fehlermeldung, die im AssertionError ausgegeben wird, können wir ebenfalls selbst bestimmen.

```python
sayi1 = int(input("Bitte geben Sie eine positive Zahl ein:"))
assert sayi1 > 0
```

```python
sayi1 = int(input("Bitte geben Sie eine positive Zahl ein:"))
assert sayi1 > 0, "Zahl darf nicht negativ sein!"
```

```python
try:
    sayi1 = int(input("Bitte geben Sie eine positive Zahl ein:"))
    assert sayi1 > 0
except:
    print("Zahl darf nicht negativ sein")
```

```python
try:
    sayi1 = int(input("Bitte geben Sie eine positive Zahl ein:"))
    assert sayi1 > 0
except AssertionError:
    print("Zahl darf nicht negativ sein")
except ValueError:
    print("Sie müssen eine Zahl eingeben!")
```
