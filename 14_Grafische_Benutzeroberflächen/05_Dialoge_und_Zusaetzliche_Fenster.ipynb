## messagebox.showinfo

**Beschreibung:** Information anzeigen

```python
import tkinter as tk
from tkinter import messagebox

root=tk.Tk()

def info():
    messagebox.showinfo("Information","Speichern erfolgreich!")

tk.Button(root,text="Info",command=info).pack(pady=20)

root.mainloop()
```

## messagebox.askyesno

**Beschreibung:** Ja/Nein-Abfrage

```python
import tkinter as tk
from tkinter import messagebox

root=tk.Tk()

def frage():
    antwort=messagebox.askyesno("Frage","Programm beenden?")
    print(antwort)

tk.Button(root,text="Fragen",command=frage).pack()

root.mainloop()
```

## filedialog.askopenfilename

**Beschreibung:** Datei auswählen

```python
import tkinter as tk
from tkinter import filedialog

root=tk.Tk()

datei=filedialog.askopenfilename()
print(datei)

root.mainloop()
```

## colorchooser.askcolor

**Beschreibung:** Farbe auswählen

```python
import tkinter as tk
from tkinter import colorchooser

root=tk.Tk()

farbe=colorchooser.askcolor()
print(farbe)

root.mainloop()
```

## simpledialog.askstring

**Beschreibung:** Text eingeben

```python
import tkinter as tk
from tkinter import simpledialog

root=tk.Tk()

name=simpledialog.askstring("Name","Wie heißen Sie?")
print(name)

root.mainloop()
```

# Toplevel

Mit `Toplevel` wird ein zweites Fenster geöffnet.


```python
import tkinter as tk

root=tk.Tk()

def neues_fenster():
    fenster=tk.Toplevel(root)
    fenster.title("Zweites Fenster")
    tk.Label(fenster,text="Hallo!").pack(padx=20,pady=20)

tk.Button(root,text="Fenster öffnen",command=neues_fenster).pack()

root.mainloop()
```

# Menü erstellen

Ein Menü gehört zu fast jeder Desktop-Anwendung.


```python
import tkinter as tk

root=tk.Tk()

menueleiste=tk.Menu(root)

datei=tk.Menu(menueleiste,tearoff=0)
datei.add_command(label="Neu")
datei.add_command(label="Öffnen")
datei.add_separator()
datei.add_command(label="Beenden",command=root.destroy)

menueleiste.add_cascade(label="Datei",menu=datei)

root.config(menu=menueleiste)

root.mainloop()
```

# Merke

- Dialoge liefern häufig einen Rückgabewert.
- `askyesno()` gibt `True` oder `False` zurück.
- `askstring()` liefert einen String.
- `askopenfilename()` liefert den Dateipfad.
