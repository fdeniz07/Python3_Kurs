## Vordefinierte Funktionen - Mathematische Funktionen

#### Absolutwert berechnen
1. abs() Befehl: Wird verwendet, um den Absolutwert einer Zahl zu berechnen.

```python
x = -5
y = 10
print(abs(x))
print(abs(y))
```

#### Zahlen runden
1. ceil() Befehl: Wird verwendet, um Zahlen mit Nachkommastellen auf die nächste ganze Zahl aufzurunden.
2. floor() Befehl: Wird verwendet, um Zahlen mit Nachkommastellen auf die vorherige ganze Zahl abzurunden.
3. round() Befehl: Wird verwendet, um Zahlen mit Nachkommastellen zu runden; wenn der Nachkommateil größer als 0.5 ist, wird aufgerundet, wenn er kleiner als 0.5 ist, wird abgerundet.

```python
import math

x = 1.35
y = 3.95
print(math.ceil(x), math.ceil(y))
print(math.floor(x), math.floor(y))
print(round(x), round(y))
```

#### Potenzierung
1. pow() Befehl: Wird für Potenzierungsoperationen verwendet; der erste Parameter gibt die Basis und der zweite Parameter den Exponenten an.

```python
print(math.pow(3, 5))
print(math.pow(5, 3))
print(3**5)
print(5**3)
```

#### Quadratwurzel ziehen
1. sqrt() Befehl: Wird zur Berechnung der Quadratwurzel verwendet.

```python
print(math.sqrt(81))
print(math.sqrt(20))
```

#### Logarithmus-Operationen
1. log() Befehl: Entspricht der Logarithmusfunktion in der Mathematik. Er berechnet, die wievielte Potenz einer Zahl eine andere Zahl ist.

```python
print(math.log(81, 9))
print(math.log(27, 3))
print(math.log(25, 3))
```

#### Trigonometrische Operationen
1. sin() Befehl: Wird zur Berechnung des Sinuswerts verwendet.
2. cos() Befehl: Wird zur Berechnung des Kosinuswerts verwendet.
3. tan() Befehl: Wird zur Berechnung des Tangenswerts verwendet.
4. radians() Befehl: Um die oben genannten Funktionen nutzen zu können, müssen wir die Gradzahlen zuerst in Bogenmaß (Radians) umrechnen.

```python
grad = math.radians(90)
print(grad)
print(math.sin(grad))
print(math.cos(grad))
print(math.tan(grad))
```

#### In das Binärsystem umwandeln
1. bin() Befehl: Wird verwendet, um eine Zahl in ihre Entsprechung im Binärsystem umzuwandeln.

```python
print(bin(128))
print(bin(4))
print(bin(156))
```

#### Ganzzahldivision und Rest (Modulo) finden
1. divmod() Befehl: Gibt den Quotienten und den Rest einer Division einer Zahl durch eine andere Zahl zurück.

```python
print(divmod(8, 5))
```
