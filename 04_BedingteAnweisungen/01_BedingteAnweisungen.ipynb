## If-Else-Elif-Strukturen

Wenn das Ergebnis einer überprüften Bedingung True ist (nicht 0 oder null), wird der entsprechende if- oder elif-Block ausgeführt.

Falls keine der Bedingungen in den if- oder elif-Blöcken erfüllt ist, wird der else-Block ausgeführt.

#### Syntax:

if bedingung:

    code_im_if_block

elif bedingung:

    code_im_elif_block

else:

    code_im_else_block


```python
if 2>3:
    print("Der If-Block wurde ausgeführt.")
elif 6>7:
    print("Der Elif-Block wurde ausgeführt.")
else:
    print("Der Else-Block wurde ausgeführt.")
```

#### Prüfung auf gerade oder ungerade Zahlen

```python
zahl=int(input("Bitte geben Sie die zu prüfende Zahl ein: "))
if zahl%2:
    print(zahl,"ist ungerade")
else:
    print(zahl,"ist gerade")
```

#### Zwei Zahlen vergleichen

```python
import random
zahl1=random.randint(1,5)
zahl2=random.randint(1,5)

print("Erste Zahl:{} und zweite Zahl:{}".format(zahl1,zahl2))

if zahl1>zahl2:
    print("{} ist größer als {}".format(zahl1,zahl2))
elif zahl2>zahl1:
    print("{} ist kleiner als {}".format(zahl1,zahl2))
else:
    print("{} ist gleich {}".format(zahl1,zahl2))
```
