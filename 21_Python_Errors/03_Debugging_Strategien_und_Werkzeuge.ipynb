# Rueckgabewert

Debugging-Funktionen liefern oft Diagnosedaten statt Fachresultate.

````python
# Beispiel 1: Logikfehler finden
def durchschnitt(werte):
    # Absichtlicher Fehler: Division durch len(werte)-1
    return sum(werte) / (len(werte) - 1)

werte = [10, 20, 30]
print('Ergebnis:', durchschnitt(werte))
````

````python
# Beispiel 2: Logging
import logging
logging.basicConfig(level=logging.INFO, format='%(levelname)s: %(message)s')

def berechne_preis(menge, einzelpreis):
    logging.info('berechne_preis aufgerufen mit menge=%s einzelpreis=%s', menge, einzelpreis)
    return menge * einzelpreis

print(berechne_preis(3, 19.99))
````

````python
# Beispiel 3: Kleine Diagnosefunktion
def validiere_daten(datensatz):
    fehler = []
    if 'name' not in datensatz:
        fehler.append('name fehlt')
    if 'alter' in datensatz and not isinstance(datensatz['alter'], int):
        fehler.append('alter muss int sein')
    return {'ok': len(fehler) == 0, 'fehler': fehler}

print(validiere_daten({'name': 'Ada', 'alter': '12'}))
````

````python
# Fehlerbehandlung mit Kontext
def parse_int(value):
    try:
        return int(value)
    except ValueError as exc:
        raise ValueError(f"ungueltige Zahl: {value}") from exc
print(parse_int("42"))
````

- Ist das Problem reproduzierbar mit klaren Schritten?
