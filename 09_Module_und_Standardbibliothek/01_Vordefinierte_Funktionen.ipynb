### Zip-Funktion

Eine Funktion, mit der wir zwei oder mehr Listen oder Dictionaries kombinieren können, um ein einzelnes Zip-Objekt zu erhalten.

#### Zippen von zwei Listen

```python
liste1 = [i for i in range(10)]
liste2 = [j for j in range(10, 20)]

liste3 = zip(liste1, liste2)
print(liste3)
```

```python
liste1
```

```python
liste2
```

```python
list(zip(liste1, liste2))
```

```python
for i, j in liste3:
    print("Werte der ersten Liste:{}, Werte der zweiten Liste:{}".format(i, j))
```

#### Zippen von drei Listen

```python
liste1 = [i for i in range(10)]
liste2 = [j for j in range(10, 20)]
liste3 = [k for k in range(20, 30)]

liste4 = list(zip(liste1, liste2, liste3))
print(liste4)
```

#### Zippen von String- und Integer-Listen

```python
liste1 = [i for i in range(5)]
liste2 = ['A', 'B', 'C', 'D', 'E']

for i, j in zip(liste1, liste2):
    print(i, j)
```

```python
liste1
```

```python
liste2
```

#### Zippen von Dictionaries

```python
dict1 = {"A": 1, "B": 2, "C": 3}
dict2 = {"D": 4, "E": 5, "F": 6}

dict3 = zip(dict1, dict2)
print(dict3)
```

```python
list(zip(dict1.keys(), dict2.keys()))
```

```python
list(zip(dict1.values(), dict2.values()))
```

```python
list(zip(dict1.items(), dict2.items()))
```

### Map-Funktion

Der Zweck dieser Funktion ist es, dieselbe Operation auf alle Elemente einer iterierbaren Struktur anzuwenden.

map(Funktionsname, Liste)

```python
liste1 = [i for i in range(20)]
liste1
```

```python
def kubik_berechnen(x):
    return x**3
```

```python
list(map(kubik_berechnen, liste1))
```

#### Map-Funktion auf einer einzelnen Liste

```python
for i in map(kubik_berechnen, liste1):
    print(i)
```

#### Map-Funktion auf zwei oder mehr Listen

```python
def kubik_berechnen_zwei(x, y):
    return x**3, y**3
```

```python
for i, j in map(kubik_berechnen_zwei, [i for i in range(10)], [j for j in range(10, 20)]):
    print(i, j)
```

### Filter-Funktion

Die Filter-Funktion erstellt eine separate Liste für die Elemente, für die eine Funktion den Wert True zurückgibt.

filter(Funktionsname, Liste)

```python
def ist_gerade(x):
    if x % 2 == 0:
        return True
    else:
        return False
```

```python
liste1 = [i for i in range(20)]
for i in liste1:
    print(ist_gerade(i))
```

```python
filter(ist_gerade, liste1)
```

```python
gefilterte_liste = list(filter(ist_gerade, liste1))
```

```python
gefilterte_liste
```

### Enumerate-Funktion

Diese Funktion kombiniert sowohl die Werte als auch die Indizes einer übergebenen Liste und gibt sie als einzelnen kombinierten Wert zurück.

```python
liste1 = [i**2 for i in range(20)]
liste1
```

```python
enumerate(liste1)
```

```python
list(enumerate(liste1))
```

```python
for i, j in enumerate(liste1):
    print("Index:{}, Wert:{}".format(i, j))
```
