## Theorie

Event-Driven Design lockert die Kopplung zwischen Services hinsichtlich Zeit und Protokoll.

```python
# Einfaches Event-Modell
event = {
    'name': 'course.created',
    'course_id': 1001,
    'title': 'Python Advanced'
}
print(event)
```

```python
# Producer/Consumer-Simulation
queue = []

def publish(evt):
    queue.append(evt)

def consume():
    if queue:
        return queue.pop(0)
    return None

publish({'name': 'user.registered', 'user_id': 1})
print(consume())
```

## Haeufige Fehler

- Event surum bilgisini tutmamak
- Nicht darauf vorbereitet sein, dass dasselbe Event mehrfach verarbeitet wird
- Event ile command kavramini karistirmak

## Best Practices

- Event adlandirmasini gecmis zamanda yap
- Idempotent consumer yaz
- Retry und dead-letter stratejisi belirle

## Uebung

Erstelle ein idempotentes Consumer-Grundgeruest fuer das Event 'payment.received'.

```python
# Loesung
processed_ids = set()

def handle_payment(event_id, payload):
    if event_id in processed_ids:
        return 'already-processed'
    processed_ids.add(event_id)
    return f"processed {payload['amount']}"

print(handle_payment('e1', {'amount': 99}))
print(handle_payment('e1', {'amount': 99}))
```

## Zusammenfassung

EDD wird mit klaren Event-Vertraegen und zuverlaessigen Zustellmustern erst wirklich robust.
