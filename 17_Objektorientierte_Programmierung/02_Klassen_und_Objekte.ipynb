## Theorie
Eine Klasse beschreibt, welche Eigenschaften und Funktionen ein Objekt haben soll.
Ein Objekt ist eine konkrete Ausprägung dieser Klasse.

#### `Klassendefinition`
Die Klassendefinition beginnt mit dem Schlüsselwort class. Danach folgt der Name der Klasse und am Ende der Zeile ein Doppelpunkt. Wie immer steht nach einem Doppelpunkt ein neuer Block, der eingerückt ist. In der ersten Zeile dieses Blocks kann – wie bei Funktionsdefinitionen – ein Docstring stehen, der die Klasse kurz erklärt

#### `Klasse` 
Der Begriff Klasse bezieht sich auf die Implementierung des Typs in einer Klassendefinition.

#### `Typ`
Der Begriff Typ bezeichnet die Außensicht einer Klasse. 

#### `__init__()`
Diese Methode wirdmauch Initialisierungsmethode genannt. Sie wird automatisch aufgerufen, wenn ein Objekt der Klasse erzeugt wird (dazu später mehr). In der Initialisierungsmethode werden die Attribute der Objekte der Klasse definiert und mit Anfangswerten belegt.

Im Python-Stil ist ein Konstruktor besonders wichtig: `__init__()` wird aufgerufen, wenn ein neues Objekt erzeugt wird.
Er initialisiert die Anfangswerte eines Objekts.

#### `Instanziierung`
Eine Klasse kann als Bauplan für Objekte betrachtet werden. Um ein Objekt der Klasse zu erzeugen, musst du die Klasse aufrufen. Diesen Vorgang nennt man instanziieren und das neue Objekt nennt man auch Instanz der Klasse. Beim Instanziieren wird die Methode __init__() aufgerufen, die die Anfangswerte der Attribute festlegt.

#### `Klassenattribut`
Ein Klassenattribut beschreibt eine Eigenschaft, die für alle Objekte einer Klasse gilt.

## Vergleich mit Java
In Java spricht man oft von einem Konstruktor. In Python verwendet man dafür `__init__()`. Der Unterschied ist vor allem syntaktisch, nicht logisch: Beide initialisieren den Zustand eines neuen Objekts.

```python
class Benutzer:
    def __init__(self, name, alter):
        self.name = name
        self.alter = alter

    def beschreiben(self):
        print(f"{self.name} ist {self.alter} Jahre alt")

b = Benutzer("Mina", 21)
b.beschreiben()
```

## Syntax
```python
class Klasse:
    def __init__(self, wert):
        self.wert = wert

    def methode(self):
        ...
```

````python
class Konto:
    def __init__(self, inhaber, saldo=0):
        self.inhaber = inhaber
        self.saldo = saldo

    def einzahlen(self, betrag):
        self.saldo += betrag
        return self.saldo

    def abheben(self, betrag):
        if betrag <= self.saldo:
            self.saldo -= betrag
            return self.saldo
        return "Nicht genug Geld"

k1 = Konto("Anna", 100)
print(k1.einzahlen(50))
print(k1.abheben(80))
````

## Beispiel 1
Jedes Objekt hat seinen eigenen Zustand.

````python
class Tier:
    def __init__(self, name, alter):
        self.name = name
        self.alter = alter

tier1 = Tier("Mia", 3)
tier2 = Tier("Balu", 5)
print(tier1.name, tier1.alter)
print(tier2.name, tier2.alter)
````

## Beispiel 2
Objekte sind unabhängig voneinander.

````python
class Rechner:
    def addieren(self, a, b):
        return a + b

r = Rechner()
print(r.addieren(5, 7))
````

## Praxisbeispiel
Eine Klasse für Schüler mit Namen und Note.

````python
class Schueler:
    def __init__(self, name, note):
        self.name = name
        self.note = note

    def hat_bestanden(self):
        return self.note >= 4

s1 = Schueler("Lina", 5)
s2 = Schueler("Tom", 3)
print(s1.hat_bestanden())
print(s2.hat_bestanden())
````

## Häufige Fehler
- `self` wird nicht übergeben
- Attribute werden ohne `self` verwendet
- Methoden werden nicht mit `()` aufgerufen

## Best Practice
Klassen sollten klar benannt und logisch aufgebaut sein.

## Übung
Erstelle eine Klasse `Produkt` mit `name` und `preis` und eine Methode `preis_mit_steuer()`.

````python
class Produkt:
    def __init__(self, name, preis):
        self.name = name
        self.preis = preis

    def preis_mit_steuer(self, steuersatz=0.19):
        return round(self.preis * (1 + steuersatz), 2)

p = Produkt("Laptop", 1000)
print(p.preis_mit_steuer())
````
