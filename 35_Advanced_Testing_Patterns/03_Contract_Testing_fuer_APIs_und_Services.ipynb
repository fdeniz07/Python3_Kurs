# Kapitel 35.3 - Contract Testing fuer APIs und Services

Contract Tests sichern die Schnittstelle zwischen Consumer und Provider gegen ungewollte Aenderungen.

```python
provider_response = {'id': 1, 'title': 'Python', 'active': True}
required_keys = {'id', 'title', 'active'}
assert required_keys.issubset(provider_response.keys())
print('contract ok')
```
