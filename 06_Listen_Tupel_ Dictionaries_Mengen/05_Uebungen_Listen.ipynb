```python
zahlen = [17, 23, 5, 12, 9]
zahlen.sort()
print(zahlen)

zahlen.reverse()
print(zahlen)

namen = ['Anke', 'Bernd', 'Claudia', 'Dieter', 'Eva', 'Frank', 'Gisela']
namen.sort()
print(namen)

personen = [(23, 'Max'), (45, 'Susi'), (12, 'Tom'), (34, 'Lisa')]
personen.sort()
print(personen) # Default sorting: Sortieren nach dem ersten Element (Alter)

personen.sort(key=lambda x: x[0])  # Sortieren nach dem ersten Element (Alter)
print(personen)

personen.sort(key=lambda x: x[1])  # Sortieren nach dem zweiten Element (Name)
print(personen)


```

```python
wort = list('Python')
print(wort)  # ['P', 'y', 't', 'h', 'o', 'n']

zahlen = list(range(5))
print(zahlen)  # [0, 1, 2, 3, 4]

quadratzahlen = [n**2 for n in range(10)]  # [0, 1, 4, 9, 16, 25, 36, 49, 64, 81]
print(quadratzahlen)

gerade_quadratzahlen = [x**2 for x in range(10) if x % 2 == 0]  # [0, 4, 16, 36, 64]
print(gerade_quadratzahlen)
```
