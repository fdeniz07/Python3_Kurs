# Kapitel 33.3 - TDD: Red Green Refactor in Python

Wir wenden den Test-Driven-Development-Zyklus Schritt fuer Schritt an.

## Lernziele

- den Zyklus Red Green Refactor verstehen
- in kleinen Testschritten vorgehen
- Verhalten waehrend des Refactorings stabil halten

## Theorie

TDD nutzt Tests nicht nur zur Verifikation, sondern als Entwurfswerkzeug.

```python
# Red: zuerst den Test formulieren
def add(a, b):
    return a + b

assert add(2, 3) == 5
```

```python
# Green: minimale Implementierung
def discount(price, percent):
    return price * (1 - percent / 100)

assert round(discount(100, 10), 2) == 90.0
```

```python
# Refactor: Lesbarkeit verbessern
def discount_v2(price, percent):
    ratio = percent / 100
    return price * (1 - ratio)

assert round(discount_v2(200, 25), 2) == 150.0
```

## Best Practices

- Ein Verhalten, ein Test
- Testnamen verhaltensorientiert formulieren
- Refactor-Schritt nicht ueberspringen

## Uebung

Schreibe zuerst einen fehlschlagenden Test und danach die Implementierung fuer eine Kapazitaetspruefung.

```python
# Loesungsbeispiel
def can_enroll(current, max_seats):
    return current < max_seats

assert can_enroll(2, 3) is True
assert can_enroll(3, 3) is False
```

## Zusammenfassung

TDD senkt Aenderungskosten und macht Architektur besser testbar.

## Technischer Tiefgang

Auf fortgeschrittenem Niveau steht nicht nur Funktionalitaet, sondern die technische Nachvollziehbarkeit im Vordergrund.
Begriffe, Risiken, Betriebsaspekte und Qualitaetskriterien werden explizit gemacht, damit Entscheidungen reproduzierbar bleiben.

## Haeufige Fehler und Debugging-Checkliste

- Sind Eingaben, Konfiguration und Randbedingungen explizit validiert?
- Sind reproduzierbare Schritte fuer den Fehler dokumentiert?
- Wurden Logs, Metriken und Tests gemeinsam ausgewertet?
- Ist die Korrektur durch einen neuen Test dauerhaft abgesichert?
- Gibt es eine kurze Lessons-Learned-Notiz fuer das Team?
