## Protokolle mit `typing.Protocol`
In Python kann man das sehr elegant mit `Protocol` aus dem `typing`-Modul ausdrücken. Ein Protokoll beschreibt, welche Methoden ein Objekt haben muss, ohne eine konkrete Oberklasse zu erzwingen.

```python
from typing import Protocol

class Lautbar(Protocol):
    def laut(self) -> str:
        ...

class Hund:
    def laut(self) -> str:
        return "Wuff"

class Katze:
    def laut(self) -> str:
        return "Miau"

for tier in [Hund(), Katze()]:
    print(tier.laut())
```

````python
from typing import Protocol

class Lautbar(Protocol):
    def laut(self) -> str:
        ...

class Hund:
    def laut(self) -> str:
        return "Wuff"

class Katze:
    def laut(self) -> str:
        return "Miau"

for tier in [Hund(), Katze()]:
    print(tier.laut())
````

````python
class Tier:
    def laut(self):
        raise NotImplementedError("Diese Methode muss überschrieben werden")

class Hund(Tier):
    def laut(self):
        return "Wuff"

class Katze(Tier):
    def laut(self):
        return "Miau"

tiere = [Hund(), Katze()]
for tier in tiere:
    print(tier.laut())
````

## Polymorphie
Die Technik, für denselben Operator in unterschiedlichen Klassen verschiedene Methoden zu definieren, nennt man Polymorphie.

## Beispiel 1
Die gleiche Methode `laut()` wird von verschiedenen Klassen unterschiedlich umgesetzt.

````python
class Form:
    def flaeche(self):
        pass

class Rechteck(Form):
    def __init__(self, breite, hoehe):
        self.breite = breite
        self.hoehe = hoehe

    def flaeche(self):
        return self.breite * self.hoehe

class Kreis(Form):
    def __init__(self, radius):
        self.radius = radius

    def flaeche(self):
        return 3.14 * self.radius ** 2

formen = [Rechteck(4, 5), Kreis(3)]
for form in formen:
    print(form.flaeche())
````

## Praxisbeispiel
Ein kleines Spiel mit Figuren.

````python
class Figur:
    def bewegen(self):
        print("Bewege mich")

class Ritter(Figur):
    def bewegen(self):
        print("Der Ritter bewegt sich nach vorne.")

class Magier(Figur):
    def bewegen(self):
        print("Der Magier teleportiert sich.")

figuren = [Ritter(), Magier()]
for figur in figuren:
    figur.bewegen()
````

## Häufige Fehler
- Abstrakte Methoden werden nicht überschrieben
- Zu starke Vererbung statt Komposition
- Polymorphie ohne gemeinsame Schnittstelle

## Best Practice
Nutze Vererbung sparsam und halte Schnittstellen klar und konsistent.

## Übung
Erstelle ein kleines OOP-Projekt mit `Schüler`, `Lehrer` und `Person`.

````python
class Person:
    def __init__(self, name):
        self.name = name

    def vorstellen(self):
        print(f"Ich bin {self.name}")

class Schueler(Person):
    def vorstellen(self):
        print(f"Ich bin {self.name} und ich lerne.")

class Lehrer(Person):
    def vorstellen(self):
        print(f"Ich bin {self.name} und ich unterrichte.")

personen = [Schueler("Mina"), Lehrer("Herr Klein")]
for person in personen:
    person.vorstellen()
````
