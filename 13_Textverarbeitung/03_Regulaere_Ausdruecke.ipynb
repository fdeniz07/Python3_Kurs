## Was ist ein regulärer Ausdruck?

Ein regulärer Ausdruck (engl. regular expression oder kurz regex) ist eine **Zeichenkette**, die ein abstraktes Muster für Texte definiert. Durch einen regulären Ausdruck reg wird eine Menge von Strings definiert, die zu reg "passen". Im einfachsten Fall besteht ein regulärer Ausdruck aus einer Zeichenfolge, die in jedem passenden String vorkommen muss.

Ein regulärer Ausdruck (engl. *regular expression*, kurz **Regex**) ist eine Zeichenkette, die ein Suchmuster beschreibt. Mit Regex lassen sich Informationen aus Texten finden oder Formate überprüfen.

Typische Anwendungsfälle:

- Telefonnummern finden
- Datumsangaben erkennen
- E-Mail-Adressen suchen
- Postleitzahlen extrahieren
- Wörter oder Zahlen filtern


Definition des Formats eines bestimmten Typs von Texten
Herausfiltern von Passagen in einem langen Text, die einem bestimmten Muster entsprechen

## Einfache Beispiele

| Regulärer Ausdruck | Bedeutung | Beispiel |
|---|---|---|
| `\d+` | eine oder mehrere Dezimalziffern | `1001`, `0` |
| `\d+ EUR` | Zahl gefolgt von Leerzeichen und EUR | `1 EUR`, `1000 EUR` |
| `\d{4}-\d{2}-\d{2}` | Datum im Format JJJJ-MM-TT | `2022-05-12` |


## Aufbau eines regulären Ausdrucks

- einfache Zeichen, die für sich selbst stehen
- Platzhalter für Zeichen aus einer Zeichenmenge
- Operatoren, die aus regulären Ausdrücken neue reguläre Ausdrücke erzeugen
- Klammern zur Gruppierung von Zeichen

Ein regulärer Ausdruck besteht aus:

- **Zeichen** (`a`, `b`, `3`)
- **Platzhaltern** (`\d`, `\w`, `.`)
- **Operatoren** (`+`, `*`, `?`, `|`)
- **Gruppierungen** (`(...)`)


## Wichtige Sonderzeichen

| Zeichen | Bedeutung |
|---|---|
| `.` | beliebiges Zeichen außer Zeilenumbruch |
| `^` | Anfang eines Strings |
| `$` bzw. `\Z` | Ende eines Strings |
| `[abc]` | genau eines der Zeichen a,b,c |
| `[^abc]` | alle Zeichen außer a,b,c |
| `*` | 0 oder mehr Wiederholungen |
| `+` | 1 oder mehr Wiederholungen |
| `?` | 0 oder 1 Wiederholung |
| `{m}` | genau m-mal |
| `{m,n}` | mindestens m-, höchstens n-mal |
| `\d` | Ziffer |
| `\D` | keine Ziffer |
| `\s` | Leerzeichen, Tab usw. |
| `\S` | kein Leerzeichen |
| `\w` | Buchstaben, Zahlen oder `_` |
| `\W` | kein alphanumerisches Zeichen |
| `|` | Oder-Verknüpfung |
| `(?:...)` | Gruppe ohne Speicherung |


# Arbeiten mit `findall()`

Die Funktion

```python
findall(pattern, string)
```

liefert **alle längsten, nicht überlappenden Treffer** als Liste zurück.


````python
from re import findall 

text = "1001 Nacht. Ali Baba und die 40 Räuber."

zahlen = findall(r'\d+', text)

print(zahlen)
````

**Ausgabe**

```text
['1001', '40']
```

`\d+` bedeutet:

- `\d` → eine Ziffer
- `+` → mindestens einmal


````python
from re import findall

text = "Ein Geist, huhuhu!"

print(findall("huhu", text))
````

## Raw-Strings

Regex enthalten viele Backslashes. Deshalb verwendet man fast immer Raw-Strings.

Statt

```python
'\\d+'
```

schreibt man

```python
r'\d+'
```

Das macht den Code übersichtlicher und verhindert Probleme mit Escape-Sequenzen.


````python
print("Erste Zeile\nZweite Zeile")

print(r"Erste Zeile\nZweite Zeile")
````

## Zeichenklassen

- `[abc]` → a oder b oder c
- `[a-z]` → Kleinbuchstaben
- `[A-Z]` → Großbuchstaben
- `[0-9]` → Ziffern
- `[^0-9]` → alles außer Ziffern


````python
from re import findall

text = "Witten 58452 Berlin 10115"

plz = findall(r'\d{5}', text)

print(plz)
````

## Reguläre Ausdrücke kombinieren

Mehrere Muster lassen sich verbinden.

Beispiele:

- `EUR|€`
- `Hund|Katze`
- `\d{2}|\d{4}`

Außerdem können Gruppen verwendet werden:

```python
(?:ha)+!
```

passt auf

- `ha!`
- `haha!`
- `hahaha!`


````python
from re import findall

text = "Python 3.10, Python 3.11 und Python 3.9"

versionen = findall(r"Python 3\.\d+", text)

print(versionen)
````

---

## Merke

- Verwende möglichst **Raw-Strings (`r''`)**.
- `findall()` liefert **alle längsten, nicht überlappenden Treffer**.
- Reguläre Ausdrücke eignen sich hervorragend zur Analyse großer Textmengen.


## Beispiele für reguläre Ausdrücke

| Textklasse                                                | Regulärer Ausdruck          | Beispieltext                                   | Ausgabe von `findall()`                        |
| --------------------------------------------------------- | --------------------------- | ---------------------------------------------- | ---------------------------------------------- |
| Positive Gleitkommazahlen als Dezimalbrüche               | `\d+\.\d+`                  | `15.23 0.1234`                                 | `['15.23', '0.1234']`                          |
| Fünfstellige positive oder negative Zahlen                | `-?\d{5}`                   | `58454 -38812`                                 | `['58454', '-38812']`                          |
| Anrede                                                    | `Frau\|Herr`                | `Frau Müller und Herr Meier`                   | `['Frau', 'Herr']`                             |
| Kleingeschriebene Wörter, die auf **en** enden            | `[a-zäöüß]+en`              | `laufen hören spielen Baum`                    | `['laufen', 'hören', 'spielen']`               |
| Gleitkommazahlen als Dezimalbrüche (positiv oder negativ) | `-?\d+\.\d+`                | `-15.23 0.1234`                                | `['-15.23', '0.1234']`                         |
| Gleitkommazahlen in Exponentialschreibweise               | `-?\d+\.\d+[eE]-?\d+`       | `-1.23e4 1.2345E-19`                           | `['-1.23e4', '1.2345E-19']`                    |
| Inlandstelefonnummer mit Vorwahl (DIN 5008)               | `0\d{2,4}[ /]?\d+(?:-\d+)?` | `0201 89775`<br>`0201/89775`<br>`031 78455-53` | `['0201 89775', '0201/89775', '031 78455-53']` |

### Hinweise

* `?` bedeutet **optional** (0 oder 1-mal).
* `{5}` bedeutet **genau fünf Wiederholungen**.
* `[eE]` akzeptiert sowohl ein kleines **e** als auch ein großes **E**.
* `(?:...)` ist eine **nicht speichernde Gruppe** (non-capturing group).
* `findall()` liefert immer **alle längsten, nicht überlappenden Treffer** zurück.
