## Ein einfaches Menü

```python
import tkinter as tk

root=tk.Tk()

menueleiste=tk.Menu(root)

datei=tk.Menu(menueleiste,tearoff=0)
datei.add_command(label="Neu")
datei.add_command(label="Öffnen")
datei.add_separator()
datei.add_command(label="Beenden",command=root.destroy)

menueleiste.add_cascade(label="Datei",menu=datei)

root.config(menu=menueleiste)

root.mainloop()
```

## Popup-Menü (Kontextmenü)

```python
import tkinter as tk

root=tk.Tk()

popup=tk.Menu(root,tearoff=0)
popup.add_command(label="Kopieren")
popup.add_command(label="Einfügen")

def anzeigen(event):
    popup.post(event.x_root,event.y_root)

root.bind("<Button-3>",anzeigen)

root.mainloop()
```

## showinfo()

**Beschreibung:** Information anzeigen

```python
import tkinter as tk
from tkinter import messagebox

root=tk.Tk()

tk.Button(root,text="showinfo()",command=lambda:messagebox.showinfo('Info','Alles wurde gespeichert.')).pack(padx=20,pady=20)

root.mainloop()
```

## showwarning()

**Beschreibung:** Warnung anzeigen

```python
import tkinter as tk
from tkinter import messagebox

root=tk.Tk()

tk.Button(root,text="showwarning()",command=lambda:messagebox.showwarning('Warnung','Bitte Eingabe prüfen.')).pack(padx=20,pady=20)

root.mainloop()
```

## showerror()

**Beschreibung:** Fehler anzeigen

```python
import tkinter as tk
from tkinter import messagebox

root=tk.Tk()

tk.Button(root,text="showerror()",command=lambda:messagebox.showerror('Fehler','Datei nicht gefunden.')).pack(padx=20,pady=20)

root.mainloop()
```

## askyesno()

**Beschreibung:** Ja/Nein-Frage

```python
import tkinter as tk
from tkinter import messagebox

root=tk.Tk()

tk.Button(root,text="askyesno()",command=lambda:antwort=messagebox.askyesno('Frage','Fortfahren?')
print(antwort)).pack(padx=20,pady=20)

root.mainloop()
```

## askokcancel()

**Beschreibung:** OK/Abbrechen

```python
import tkinter as tk
from tkinter import messagebox

root=tk.Tk()

tk.Button(root,text="askokcancel()",command=lambda:antwort=messagebox.askokcancel('Frage','Programm schließen?')
print(antwort)).pack(padx=20,pady=20)

root.mainloop()
```

## Datei öffnen

```python
import tkinter as tk
from tkinter import filedialog

root=tk.Tk()

datei=filedialog.askopenfilename(
    title="Datei auswählen",
    filetypes=[("Textdateien","*.txt"),("Alle Dateien","*.*")]
)

print(datei)

root.mainloop()
```

## Datei speichern

```python
import tkinter as tk
from tkinter import filedialog

root=tk.Tk()

datei=filedialog.asksaveasfilename(
    defaultextension=".txt"
)

print(datei)

root.mainloop()
```

## Verzeichnis auswählen

```python
import tkinter as tk
from tkinter import filedialog

root=tk.Tk()

ordner=filedialog.askdirectory()

print(ordner)

root.mainloop()
```

## Farbe auswählen

```python
import tkinter as tk
from tkinter import colorchooser

root=tk.Tk()

farbe=colorchooser.askcolor()

print(farbe)

root.mainloop()
```

## Benutzereingabe mit simpledialog

```python
import tkinter as tk
from tkinter import simpledialog

root=tk.Tk()

name=simpledialog.askstring("Name","Wie heißen Sie?")

print(name)

root.mainloop()
```
