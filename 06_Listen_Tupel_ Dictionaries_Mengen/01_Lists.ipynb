## Listen

#### Warum brauchen wir Listen?

Wir verwenden Listen, um eine große Anzahl von Variablen oder Werten zusammen zu speichern ve einfacher auf sie zugreifen zu können. Es ist eine Struktur, die es uns ermöglicht, mehrere Werte desselben oder unterschiedlicher Typen zusammenzuhalten ve Operationen wie das Hinzufügen, Löschen ve Aktualisieren von Werten durchzuführen.

Ist es beispielsweise sinnvoll, die Noten von 50 verschiedenen Personen in einer Klasse in 50 verschiedenen Variablen zu speichern, oder ist es besser, nur eine einzige Liste zu definieren ve sie darin aufzubewahren?

### WICHTIG
Die Elemente einer Liste befinden sich sequentiell (hintereinander) im Speicher. Der Name der Liste repräsentiert gleichzeitig die Startadresse.

Adresse Element

0000  1. Element

0001  2. Element

0002  3. Element

0003  4. Element

#### Wie werden Listen definiert?
1. Mit dem Schlüsselwort `list()`
2. Mit eckigen Klammern `[]`

##### Eine leere Liste definieren

```python
listeA = list()
listeB = []
print(type(listeA))
print(listeA)
print(type(listeB))
print(listeB)
```

##### Eine Integer-Liste definieren

```python
zahlenListe = [1, 2, 3, 4, 5]
print(type(zahlenListe))
print(zahlenListe)
```

##### Eine String-Liste definieren

```python
textListe = ["Python", "C", "C++", "Perl", "Java"]
print(type(textListe))
print(textListe)
```

##### Einzelner Zugriff auf Listenelemente
Die Indizes von Listenelementen beginnen bei 0.

```python
zahlenListe = [1, 10, 33, 54, 75]
print(zahlenListe[0])
print(zahlenListe[1])
print(zahlenListe[-3])
print(zahlenListe[-2])
print(zahlenListe[-1])
```

##### Zugriff auf Listenelemente mit einer Schleife

```python
for element in zahlenListe:
    print(element**3)
```

##### Die Anzahl der Elemente einer Liste ermitteln
Der Befehl `len()` gibt die Anzahl der Elemente in einer Liste zurück.

```python
anzahlElemente = len(zahlenListe)
print("Anzahl der Elemente in der Liste:", anzahlElemente)
```

##### Zugriff auf die Indizes der Listenelemente

```python
for i in range(len(zahlenListe)):
    print(i)
```

##### Werte von Elementen in einer Liste ändern

```python
zahlenListe = [1, 2, 3, 4, 5]
print(zahlenListe)
zahlenListe[0] = 0
zahlenListe[1] = 1
print(zahlenListe)
testListe = [1, 5, 33, 4, 2, 6]
for i in range(len(testListe)):
    testListe[i] = i**2
print(testListe)
```

##### Slicing (Teilen) einer Liste
`liste[start_index:end_index:schrittweite]`

```python
zahlenListe = [1, 2, 3, 4, 5]
teilListeA = zahlenListe[3:]
teilListeB = zahlenListe[:3]
teilListeC = zahlenListe[::2]
teilListeD = zahlenListe[::-1]
print(teilListeA)
print(teilListeB)
print(teilListeC)
print(teilListeD)
```

#### Methoden zum Hinzufügen von Elementen
1. `append`-Methode: Wird verwendet, um ein Element am Ende der Liste hinzuzufügen.
2. `insert`-Methode: Wird verwendet, um ein Element an einem bestimmten Index der Liste hinzuzufügen.

```python
meineListe = [1, 2, 3, 4, 5]
print(meineListe)
meineListe.append(10)
meineListe.append("Max")
meineListe.append(3.5)
meineListe.append(True)
print(meineListe)
```

```python
meineListe = [1, 2, 3, 4, 5]
print(meineListe)
meineListe.insert(0, 10)
print(meineListe)
meineListe.insert(3, "Max")
print(meineListe)
```

#### Elemente aus einer Liste löschen
1. `remove`-Methode: Wird verwendet, um einen bestimmten Wert aus der Liste zu entfernen.
2. `del`-Befehl: Wird verwendet, um einen Wert an einem bestimmten Index zu löschen oder die gesamte Liste zu löschen.
3. `pop`-Methode: Wird verwendet, um das letzte Element oder ein Element an einem bestimmten Index zu löschen ve das gelöschte Element einer Variablen zuzuweisen.

```python
loeschListe = [1, 2, 3, 4, 5, 5, 5, 5]
print(loeschListe)
loeschListe.remove(5)
loeschListe.remove(5)
loeschListe.remove(5)
print(loeschListe)
del(loeschListe[0])
print(loeschListe)
geloeschterWert = loeschListe.pop(1)
print(loeschListe)
print(geloeschterWert)
del loeschListe
#print(loeschListe)
```

#### Überprüfung von Elementen in einer Liste
1. `in`-Operator: Gibt True zurück, wenn das Element in der Liste existiert, andernfalls False.
2. `not in`-Operator: Gibt False zurück, wenn das Element in der Liste existiert, andernfalls True.
3. `index`-Methode: Wenn das gesuchte Element in der Liste existiert, wird der entsprechende Index zurückgegeben. Existiert das Element nicht, wird ein ValueError geworfen.

```python
suchListe = [10, 20, 30, 40, 50]

print(10 in suchListe)
print(25 in suchListe)

print(10 not in suchListe)
print(25 not in suchListe)

print(suchListe.index(10))

if 25 in suchListe:
    print(suchListe.index(25))
```

#### Listen sortieren ve umdrehen
1. `sort`-Methode: Sortiert die Elemente der Liste in aufsteigender Reihenfolge (klein nach groß).
2. `reverse`-Methode: Kehrt die Reihenfolge der Elemente in der Liste um (das erste wird zum letzten, das letzte zum ersten).

```python
sortierListe = [25, 43, 21, 54, 23]
print(sortierListe)
sortierListe.reverse()
print(sortierListe)
sprachenListe = ["python", "c", "ruby", "perl", "java"]
print(sortierListe)
sortierListe.sort()
print(sortierListe)
print(sprachenListe)
sprachenListe.sort()
print(sprachenListe)

sortierListe.reverse()
print(sortierListe)
```

##### Zählen, wie oft ein Element in einer Liste vorkommt
`count`-Methode: Gibt an, wie oft das übergebene Element in der Liste enthalten ist.

```python
vorkommenListe = [1, 1, 1, 2, 3, 1, 5, 5, 6]
print(vorkommenListe.count(1))
```

##### Das kleinste, größte Element ve die Summe einer Liste ermitteln

`min`-Funktion: Gibt das kleinste Element der Liste zurück

`max`-Funktion: Gibt das größte Element der Liste zurück

`sum`-Funktion: Gibt die Gesamtsumme aller Elemente in der Liste zurück

```python
werteListe = [1, 5, 96, 32, 543, 65]
print(min(werteListe))
print(max(werteListe))
print(sum(werteListe))

werteListe.sort()
print(werteListe[0])
print(werteListe[-1])
```

#### Methoden zur Generierung von Listenelementen (List Comprehensions)

```python
generierteListeA = [i for i in range(5)]
print(generierteListeA)
```

```python
generierteListeB = [i**2 for i in range(5)]
print(generierteListeB)
```

```python
generierteListeC = [i**2 for i in range(10)]
print(generierteListeC)

generierteListeC = [i**2 for i in range(10) if i % 2 == 0]
print(generierteListeC)
```
