## Was sind Operatoren?

##### Arithmetische Operatoren: Addition, Subtraktion, Multiplikation, Ganzzahldivision, Gleitkommadivision, Modulo, Potenzierung
##### Vergleichsoperatoren: ==, !=, >, <, >=, <=
##### Logische Operatoren: AND, OR, NOT
##### Bitweise Operatoren: &, |, ^, ~, <<, >>
##### Extra: in, not in, is, is not

### Logische Operatoren

```python
x = True
y = False

print(x and y) # Wenn beide Ausdrücke True sind, wird True zurückgegeben. Andernfalls wird False zurückgegeben.
print(not (7 > 5) and (8 > 6))

print(x or y) # Wenn mindestens einer der beiden Ausdrücke True ist, wird True zurückgegeben. Andernfalls wird False zurückgegeben.
print((7 < 5) or (8 < 6))

print(not x) # Kehrt den booleschen Wert um, vor dem es steht.
print(not y)
```
