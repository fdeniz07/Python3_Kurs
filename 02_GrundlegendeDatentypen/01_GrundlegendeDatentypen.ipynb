## Was sind die grundlegenden Datentypen?

#### 1. Numerische Datentypen: Integer, Float, Complex

##### Integer: Wird für ganze Zahlen verwendet

```python
a=int()
print(a)
print(type(a))
b=25
print(b)
print(type(b))
```

##### Float: Wird für Dezimalzahlen verwendet

```python
x=float()
print(x)
print(type(x))
y=12.5
print(y)
print(type(y))
```

##### Complex: Wird für komplexe Zahlen verwendet

```python
k=complex()
print(k)
print(type(k))
l=1+2j
print(l)
print(type(l))
```

#### 2. Zeichenketten-Datentyp: String

##### Kann mit doppelten Anführungszeichen erstellt werden

```python
str1="max"
print(str1)
print(type(str1))
```

##### Kann mit einfachen Anführungszeichen erstellt werden

```python
str2='erste'
print(str2)
print(type(str2))
```

##### Kann mit dreifachen Anführungszeichen erstellt werden

```python
str3="""
mark
erste
"""
print(str3)
print(type(str3))
```

##### Kann mit dem Schlüsselwort str() erstellt werden

```python
str4=str()
print(str4)
print(type(str4))
```

##### Auf einzelne Zeichen kann direkt mit [] zugegriffen werden

```python
str5="karlerste"
print(str5)
print(str5[0])
print(str5[:])
```

##### ACHTUNG!
Verschachtelte doppelte oder einfache Anführungszeichen können nicht direkt verwendet werden!

```python
text='Max Mustermann nimmt gerade am Kurs "Python-Programmierung" teil'
print(text)
```

#### 3. Logischer Datentyp: True / False

```python
a=True
b=False
c=bool()
print(type(a))
print(a)
print(type(b))
print(b)
print(type(c))
print(c)
```

```python
c=1==3
print(c)
print(type(c))
```
