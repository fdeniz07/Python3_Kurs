## Wie erfolgt die Umwandlung zwischen Datentypen?


### Umwandlung in einen Integer

```python
a=12.5
b=int(a)
print(a)
print(b)
print(type(a))
print(type(b))
```

```python
c="123"
d=int(c)
print(c)
print(d)
print(type(c))
print(type(d))
```

```python
a=input("Bitte geben Sie eine Zahl ein: ")
print(a)
print(type(a))
a=int(a)
# Zahl um 1 erhöhen
print(a)
print(type(a))
```

### Umwandlung in eine Fließkommazahl

```python
a=12
b=float(a)
print(a)
print(b)
print(type(a))
print(type(b))
```

```python
c="123"
d=float(c)
print(c)
print(d)
print(type(c))
print(type(d))
```

```python
a=float(input("Bitte geben Sie eine Zahl ein: "))
a=a+5
print(type(a))
print(a)
```

### Umwandlung in eine Zeichenkette

```python
a=123
b=str(a)
print(a)
print(b)
print(type(a))
print(type(b))
```

### Umwandlung in ein Zeichen

```python
a=85
b=chr(a)
print(type(a))
print(type(b))
print(a)
print(b)
```

### Umwandlung eines Zeichens in den ASCII-/Unicode-Wert

```python
a='U'
b=ord(a)
print(a)
print(b)
```

### Umwandlung in einen booleschen Datentyp

##### Alle Werte außer 0 sowie nicht-leere Zeichenketten werden bei der Umwandlung in bool als True interpretiert

```python
a=0
b=1
c=bool(a)
d=bool(b)
print(type(a))
print(type(b))
print(type(c))
print(type(d))
print(c)
print(d)
```

```python
a=""
b="Max"
print(bool(a))
print(bool(b))
```
