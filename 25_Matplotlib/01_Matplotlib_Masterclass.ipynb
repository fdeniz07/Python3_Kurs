# Theorie

Visualisierung ist Interpretation. Gute Diagramme reduzieren Missverstaendnisse und beschleunigen Entscheidungen.

# Erklaerung

Wichtige Diagrammtypen:
- Line Plot fuer Zeitreihen
- Bar Plot fuer Vergleiche
- Scatter fuer Zusammenhaenge

# Syntax

```python
import matplotlib.pyplot as plt
plt.plot(x, y)
plt.show()
```

# Merke

Ein Diagramm ohne Kontext (Titel, Achsen, Einheit) ist oft wertlos.

# Parameter

- `figsize`
- `color`, `linestyle`, `marker`
- `alpha`, `label`

# Rueckgabewert

Matplotlib liefert Figure/Axes-Objekte, die weiter angepasst werden koennen.

````python
import matplotlib.pyplot as plt

# Beispiel 1
x = [1,2,3,4,5]
y = [2,3,5,4,6]
plt.figure(figsize=(6,3))
plt.plot(x, y, marker='o', label='Umsatztrend')
plt.title('Linienchart')
plt.xlabel('Monat')
plt.ylabel('Wert')
plt.legend()
plt.grid(True, alpha=0.3)
plt.show()
````

````python
# Beispiel 2
kategorien = ['A', 'B', 'C']
werte = [30, 45, 25]
plt.bar(kategorien, werte, color=['#2E86AB','#A23B72','#F18F01'])
plt.title('Balkendiagramm')
plt.show()
````

````python
# Beispiel 3
x = [1,2,3,4,5,6]
y = [1.5,2.2,2.8,3.7,4.2,5.1]
plt.scatter(x, y, alpha=0.8)
plt.title('Scatterplot')
plt.xlabel('Lernstunden')
plt.ylabel('Score')
plt.show()
````
