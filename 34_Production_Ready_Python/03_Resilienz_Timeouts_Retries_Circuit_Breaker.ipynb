# Kapitel 34.3 - Resilienz: Timeouts, Retries, Circuit Breaker

Robuste Systeme erwarten Fehler und reagieren kontrolliert. Dieses Notebook behandelt Resilienz-Bausteine fuer produktive Python-Services.

## Lernziele

- Timeouts konsequent einsetzen
- Retries mit Backoff planen
- Circuit-Breaker-Prinzip verstehen

```python
import time

def call_external(attempt):
    if attempt < 3:
        raise RuntimeError('temporary failure')
    return 'ok'

for attempt in range(1, 5):
    try:
        result = call_external(attempt)
        print('success', result)
        break
    except RuntimeError as exc:
        wait = attempt * 0.2
        print(f'retry {attempt} after {wait}s because {exc}')
        time.sleep(wait)
```

## Best Practices

- Retry nur fuer transiente Fehler einsetzen
- idempotente Operationen bevorzugen
- Circuit Breaker mit Monitoring kombinieren
