## In der print-Funktion verwendete Sonderzeichen

### \n-Zeichen
Ermöglicht den Wechsel in die nächste Zeile

```python
print("Max\nErste\nZweite\nDritte")
```

### \t-Zeichen
Fügt einen Tabulatorabstand ein

```python
print("Max\tErste")
print("MaxErste")
```

### \b-Zeichen
Löscht ein Zeichen und entspricht der Funktion der Rücktaste (Backspace)

```python
print("Max Erste")
print("Max \bErste") 
```

### \r-Zeichen
Setzt den Cursor an den Zeilenanfang zurück und überschreibt den vorhandenen Text

```python
print("Max\rErste")
```

### Verwendung von doppeltem \
Wenn wir \\\\ verwenden, können wir direkt einen Backslash (\\) ausgeben

```python
print("Max\\Erste")
```
