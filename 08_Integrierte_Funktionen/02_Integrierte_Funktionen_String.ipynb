## Integrierte Funktionen - String-Funktionen (Built-in String Functions)

#### Bestimmte Zeichen in einem String ersetzen
1. `replace()`-Funktion: Wird verwendet, um ein bestimmtes Zeichen oder eine Sequenz von Zeichen innerhalb eines Strings durch andere zu ersetzen.

```python
text1 = text2 = text3 = "fatihdeniz"

text1 = text1.replace("i", "*")
text2 = text2.replace("fatih", "*")
text3 = text3.replace("i", "*", 1)

print(text1)
print(text2)
print(text3)
```

#### Strings in Groß- oder Kleinschreibung umwandeln
1. `upper()`-Funktion: Wird verwendet, um alle Zeichen in Großbuchstaben umzuwandeln.
2. `lower()`-Funktion: Wird verwendet, um alle Zeichen in Kleinbuchstaben umzuwandeln.

```python
text = "fatih"
text = text.upper()
print(text)
text = text.lower()
print(text)
```

#### Ersten Buchstaben vergrößern - Alle Wörter im Titelformat formatieren - Groß- und Kleinschreibung vertauschen
1. `capitalize()`-Funktion: Wird verwendet, um nur den allerersten Buchstaben des gesamten Strings in einen Großbuchstaben umzuwandeln.
2. `title()`-Funktion: Wird verwendet, um den jeweils ersten Buchstaben eines jeden Wortes im String in einen Großbuchstaben umzuwandeln.
3. `swapcase()`-Funktion: Wird verwendet, um alle Großbuchstaben im String in Kleinbuchstaben und alle Kleinbuchstaben in Großbuchstaben umzuwandeln.

```python
text = "fatih DeNiZ bEreItEt gERaDe eIne PyThoN-ProGRamMierStUnde Vor."

capitalizeText = text.capitalize()
titleText = text.title()
swapcaseText = text.swapcase()

print(capitalizeText)
print(titleText)
print(swapcaseText)
```

#### Unnötige Zeichen aus einem String entfernen (Trimming)
1. `strip()`-Funktion: Wird verwendet, um Whitespaces oder Sonderzeichen wie "\n", " ", "\t", "\r", "\v" an beiden Enden (links und rechts) des Strings zu entfernen.
2. `lstrip()`-Funktion: Wird verwendet, um die genannten Zeichen nur am linken Ende (Anfang) des Strings zu entfernen.
3. `rstrip()`-Funktion: Wird verwendet, um die genannten Zeichen nur am rechten Ende (Schluss) des Strings zu entfernen.

```python
text = "     Fatih Deniz bereitet gerade eine Python-Programmierstunde vor.    \t\t"
textStrip = text.strip()
textLStrip = text.lstrip()
textRStrip = text.rstrip()

print(len(text))
print(len(textStrip))
print(len(textLStrip))
print(len(textRStrip))

print(text)
print(textStrip)
print(textLStrip)
print(textRStrip)
```

#### Anfang und Ende eines Strings überprüfen
1. `startswith()`-Funktion: Wird verwendet, um zu prüfen, ob ein String mit dem oder den angegebenen Zeichen beginnt (gibt `True` oder `False` zurück).
2. `endswith()`-Funktion: Wird verwendet, um zu prüfen, ob ein String mit dem oder den angegebenen Zeichen endet (gibt `True` oder `False` zurück).

```python
text = "fatih deniz bereitet gerade eine Python-Programmierstunde vor"

print(text.startswith("fatih"))
print(text.startswith("a"))
print(text.endswith("r"))
print(text.endswith("vorr"))
```
