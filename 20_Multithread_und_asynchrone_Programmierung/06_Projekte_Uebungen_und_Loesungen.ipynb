# Uebung

Implementiere einen Async-Aggregator mit 5 simulierten Diensten, davon 1 Timeout und 1 Fehler.

````python
# Beispiel-Loesungsskizze
import asyncio

async def dienst(name, delay, fail=False):
    await asyncio.sleep(delay)
    if fail:
        raise RuntimeError(f'Dienst {name} fehlgeschlagen')
    return {name: 'ok'}

async def aggregate():
    jobs = [
        asyncio.wait_for(dienst('a', 0.2), timeout=1),
        asyncio.wait_for(dienst('b', 0.5), timeout=1),
        asyncio.wait_for(dienst('c', 2.0), timeout=1),
        asyncio.wait_for(dienst('d', 0.3, fail=True), timeout=1),
        asyncio.wait_for(dienst('e', 0.1), timeout=1),
    ]
    ergebnisse = await asyncio.gather(*jobs, return_exceptions=True)
    print(ergebnisse)

asyncio.run(aggregate())
````

````python
# asyncio Timeout-Guard
import asyncio
async def slow_job():
    await asyncio.sleep(2)
    return "done"
async def main():
    try:
        print(await asyncio.wait_for(slow_job(), timeout=1.0))
    except asyncio.TimeoutError:
        print("timeout")
asyncio.run(main())
````
