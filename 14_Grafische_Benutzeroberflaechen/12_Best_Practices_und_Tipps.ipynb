In diesem Kapitel lernst du, wie du Tkinter-Programme übersichtlicher,

# Warum Best Practices?

Mit zunehmender Größe eines Projekts wird guter Code immer wichtiger.

Vorteile:

- bessere Lesbarkeit
- einfachere Fehlersuche
- leichtere Erweiterung
- Wiederverwendbarkeit


# Aussagekräftige Variablennamen


Verwende verständliche Namen.

Gut:

- name_entry
- speichern_button

Schlecht:

- a
- x1
- test2


# Code in Funktionen aufteilen


Anstatt den gesamten Code im Hauptprogramm zu schreiben,
sollten Aktionen in Funktionen ausgelagert werden.


# Widgets logisch gruppieren


Verwende Frame oder LabelFrame,
um zusammengehörige Widgets zu gruppieren.


# Kommentare sinnvoll einsetzen


Kommentare erklären den Zweck des Codes,
nicht jede einzelne Zeile.


# Keine doppelten Codeblöcke


Wiederhole denselben Code möglichst nicht.
Nutze stattdessen Funktionen.


# Fehler behandeln


Verwende try/except,
wenn Benutzereingaben fehlerhaft sein können.


# Konstanten verwenden


Speichere häufig verwendete Werte
in Konstanten.



# Beispiel: Schlechter Code


```python

import tkinter as tk

root=tk.Tk()

entry=tk.Entry(root)
entry.pack()

button=tk.Button(root,text="OK")
button.pack()

root.mainloop()
```


Probleme:

- keine Funktionen
- keine Kommentare
- keine Struktur



# Beispiel: Verbesserter Code


```python

import tkinter as tk

def begruessen():
    print("Hallo")

root=tk.Tk()

eingabe=tk.Entry(root)
eingabe.pack()

button=tk.Button(
    root,
    text="Begrüßen",
    command=begruessen
)

button.pack()

root.mainloop()
```


# Typische Anfängerfehler

- mainloop() vergessen
- Widgets nicht packen
- pack() und grid() im gleichen Container mischen
- get() vergessen
- destroy() falsch verwenden
- Klammern bei command setzen
