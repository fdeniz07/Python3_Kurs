## Theorie
Gute objektorientierte Programme sind nicht nur korrekt, sondern auch klar, wartbar und erweiterbar.
Die SOLID-Prinzipien helfen dabei, solche Programme zu gestalten.

## Erklärung
- S: `Single Responsibility Principle (Prinzip der einzigen Verantwortung)` – eine Klasse hat eine Aufgabe
- O: `Open/Closed Principle (Offen/Geschlossen-Prinzip)` – offen für Erweiterung, geschlossen für Änderung
- L: `Liskov Substitution Principle (Liskovsches Substitutionsprinzip)` – Unterklassen dürfen Oberklassen ersetzen
- I: `Interface Segregation Principl (Prinzip der Schnittstellentrennung)` – kleine, klare Schnittstellen
- D: `Dependency Inversion Principle (Prinzip der Abhängigkeitsumkehr)` – abhängig von Abstraktionen statt Konkreta

```python
class Rechnung:
    def __init__(self, betrag):
        self.betrag = betrag

    def speichern(self):
        print(f"Rechnung {self.betrag} gespeichert")

    def senden(self):
        print(f"Rechnung {self.betrag} gesendet")

r = Rechnung(100)
r.speichern()
r.senden()
```

## Beispiel 1
Eine Klasse sollte nicht gleichzeitig speichern, senden und berechnen, wenn das zu viel Verantwortung ist.

```python
class EmailService:
    def senden(self, nachricht):
        print(f"Sende: {nachricht}")

class Benachrichtigung:
    def __init__(self, service):
        self.service = service

    def senden(self, text):
        self.service.senden(text)

b = Benachrichtigung(EmailService())
b.senden("Hallo Welt")
```

## Beispiel 2
Hier wird ein Service in die Klasse injiziert. Das ist ein einfaches Dependency-Injection-Muster.

```python
from typing import Protocol

class Druckbar(Protocol):
    def drucken(self) -> None:
        ...

class Bericht:
    def __init__(self, ziel: Druckbar):
        self.ziel = ziel

    def ausgeben(self) -> None:
        self.ziel.drucken()

class ConsoleDrucker:
    def drucken(self) -> None:
        print("Drucke auf Konsole")

bericht = Bericht(ConsoleDrucker())
bericht.ausgeben()
```

## Praxisbeispiel
Ein kleines Beispiel zu Duck Typing und Protokollen.

## Häufige Fehler
- Zu viele Verantwortlichkeiten in einer Klasse
- Abhängigkeiten direkt im Code verankern
- Zu starke Vererbung statt klarer Schnittstellen

## Best Practice
Halte Klassen klein, benenne sie klar und trenne Verantwortlichkeiten.

## Übung
Erstelle eine kleine Klasse `Logger`, die per Dependency Injection an eine `MailService`-Klasse übergeben wird.

```python
class MailService:
    def senden(self, text):
        print(f"Mail gesendet: {text}")

class Logger:
    def __init__(self, mail_service):
        self.mail_service = mail_service

    def loggen(self, text):
        self.mail_service.senden(text)

logger = Logger(MailService())
logger.loggen("Fehler aufgetreten")
```
