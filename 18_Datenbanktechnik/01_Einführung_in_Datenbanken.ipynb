# Kapitel 18.1 — Einführung in Datenbanken

## Lernziele
- Grundkonzepte von Datenbanken verstehen.
- Den Unterschied zwischen relationalen und NoSQL-Datenbanken erklären.
- ACID-Eigenschaften verstehen.

## Theorie

Eine Datenbank ist ein strukturiertes System zur Speicherung und Verwaltung großer Mengen von Daten. Datenbanken ermöglichen es, Daten:
- zu speichern
- zu suchen
- zu bearbeiten
- zu löschen

Es gibt verschiedene Typen von Datenbanken.

## Erklärung

**Relationale Datenbanken:**
- Daten sind in Tabellen organisiert
- Tabellen haben Spalten (Attribute) und Zeilen (Datensätze)
- Es gibt Beziehungen zwischen Tabellen
- Beispiele: SQLite, MySQL, PostgreSQL

**NoSQL Datenbanken:**
- Daten sind nicht in Tabellen organisiert
- Flexiblere Strukturen (Dokumente, Key-Value-Paare)
- Besser für unstrukturierte Daten geeignet
- Beispiele: MongoDB, Redis, Cassandra

## ACID-Eigenschaften

- **A (Atomarität)**: Eine Transaktion wird ganz oder gar nicht ausgeführt
- **C (Konsistenz)**: Die Datenbank bleibt in einem konsistenten Zustand
- **I (Isolation)**: Transaktionen beeinflussen sich nicht gegenseitig
- **D (Dauerhaftigkeit)**: Gespeicherte Daten bleiben erhalten

## Beispiel 1
Wie sind Daten typischerweise in einer relationalen Datenbank organisiert?

```
Tabelle: Schüler
| ID | Name       | Klasse |
|----|------------|--------|
| 1  | Anna       | 10A    |
| 2  | Beni       | 10B    |
| 3  | Carla      | 10A    |
```

## Beispiel 2
Vergleich: MongoDB (NoSQL) speichert Daten als Dokumente:

```json
{
  "_id": 1,
  "name": "Anna",
  "klasse": "10A",
  "noten": [5, 4, 5]
}
```

## Praxisbeispiel
Eine einfache Abfrage in SQL.

```sql
SELECT * FROM schüler WHERE klasse = '10A';
```

## Häufige Fehler
- Falsche Datenbank-Typ für den Anwendungsfall wählen
- Normalisierung ignorieren
- Keine Backups erstellen

## Best Practice
Wähle die richtige Datenbank für dein Problem: Relational für strukturierte Daten, NoSQL für flexible Strukturen.

## Zusammenfassung
Datenbanken sind zentral für moderne Anwendungen. Jede Datenbank hat ihre Stärken und Schwächen.
