## Wörterbücher (Dictionaries)

#### Warum brauchen wir Wörterbücher?

Sie werden verwendet, um Elemente in einer Schlüssel-Wert-Struktur (Key-Value) zu speichern. Auf Wörterbüchern können Operationen wie Hinzufügen, Löschen und Aktualisieren durchgeführt werden.

#### Wie werden Wörterbücher definiert?
1. Mit dem Schlüsselwort `dict()`
2. Mit geschweiften Klammern `{}`

##### Ein leeres Wörterbuch definieren

```python
woerterbuch = dict()
woerterbuch2 = {}
print(type(woerterbuch))
print(type(woerterbuch2))
```

##### Ein Wörterbuch mit Elementen definieren und auf Werte zugreifen
##### Syntax-Struktur
`{"schluessel1": "wert1", "schluessel2": "wert2"}`

```python
zahlenWoerterbuch = {"1": "eins", "2": "zwei", "3": "drei"}
print(zahlenWoerterbuch["1"])
print(zahlenWoerterbuch["2"])
print(zahlenWoerterbuch["3"])
```

##### Zugriff mithilfe einer Schleife

```python
for schluessel in zahlenWoerterbuch:
    print("{}:{}".format(schluessel, zahlenWoerterbuch[schluessel]))
```

### ACHTUNG
Wenn ein Schlüssel verwendet wird, der nicht im Wörterbuch existiert, wird ein `KeyError` ausgelöst.

1. `in`-Operator: Gibt True zurück, wenn der Schlüssel im Wörterbuch existiert, andernfalls False.

2. `not in`-Operator: Gibt False zurück, wenn der Schlüssel im Wörterbuch existiert, andernfalls True.

```python
if "3" in zahlenWoerterbuch:
    print(zahlenWoerterbuch["3"])
else:
    print("Dieser Schluessel existiert nicht!")
```

##### GET-METHODE
Ermöglicht das Abrufen des Wertes, der dem übergebenen Schlüssel entspricht. Wenn der Schlüssel nicht existiert, wird kein Fehler ausgelöst, sondern `None` zurückgegeben.

```python
print(zahlenWoerterbuch.get("1"))
print(zahlenWoerterbuch.get("2"))
print(zahlenWoerterbuch.get("3"))
print(zahlenWoerterbuch.get("4"))
```

##### Elemente hinzufügen, Werte ändern und Elemente löschen
Beim Hinzufügen einer neuen Schlüssel-Wert-Beziehung wird die folgende Syntax-Struktur verwendet:

`woerterbuchName["schluessel"] = "wert"`

```python
zahlenWoerterbuch["4"] = "vier"
zahlenWoerterbuch["0"] = "null"
print(zahlenWoerterbuch)
zahlenWoerterbuch["4"] = "Vier"
print(zahlenWoerterbuch)
del zahlenWoerterbuch["4"]
print(zahlenWoerterbuch)
```

##### Die Konzepte `items()`, `keys()` und `values()`

1. `items()`: Ermöglicht den gleichzeitigen Zugriff auf Schlüssel und Werte.

2. `keys()`: Ermöglicht den einzelnen Zugriff auf die Schlüssel im Wörterbuch.

3. `values()`: Ermöglicht den einzelnen Zugriff auf die Werte im Wörterbuch.

```python
for schluessel, wert in zahlenWoerterbuch.items():
    print("{}:{}".format(schluessel, wert))
```

```python
for schluessel in zahlenWoerterbuch.keys():
    print(schluessel)
```

```python
for wert in zahlenWoerterbuch.values():
    print(wert)
```

##### Anzahl der Elemente in einem Wörterbuch ermitteln

`len()`: Gibt die Länge (Anzahl der Elemente) der übergebenen Datenstruktur zurück.

```python
print("Das Zahlenwoerterbuch enthaelt {} Elemente".format(len(zahlenWoerterbuch)))
```

##### Wörterbuch aktualisieren, kopieren, Elemente leeren und Wörterbuch löschen

1. `update()`: Wird verwendet, um das Wörterbuch mit einem neuen Wörterbuch zu aktualisieren. Werte bestehender Schlüssel werden überschrieben, neue Schlüssel-Wert-Paare werden hinzugefügt.

2. `copy()`: Wird verwendet, um eine flache Kopie der Elemente des Wörterbuchs zu erstellen.

3. `clear()`: Löscht alle Elemente aus dem Wörterbuch, das Wörterbuch selbst bleibt jedoch im Speicher vorhanden.

4. `del()`: Wird verwendet, um das Wörterbuch vollständig aus dem Speicher zu löschen.

```python
zahlenAktuell = {"1": "EINS", "2": "ZWEI", "3": "DREI", "4": "VIER", "5": "FUENF"}
print(zahlenWoerterbuch)
zahlenWoerterbuch.update(zahlenAktuell)
print(zahlenWoerterbuch)
```

```python
zahlenWoerterbuch2 = zahlenWoerterbuch
zahlenWoerterbuch2["6"] = "SECHS"
zahlenWoerterbuch["7"] = "SIEBEN"
print(zahlenWoerterbuch)
print(zahlenWoerterbuch2)
print(id(zahlenWoerterbuch))
print(id(zahlenWoerterbuch2))

zahlenWoerterbuch3 = zahlenWoerterbuch.copy()
zahlenWoerterbuch3["8"] = "ACHT"
print(zahlenWoerterbuch)
print(zahlenWoerterbuch3)
print(id(zahlenWoerterbuch))
print(id(zahlenWoerterbuch3))
```

```python
zahlenWoerterbuch.clear()
print(zahlenWoerterbuch)
print(zahlenWoerterbuch2)
print(zahlenWoerterbuch3)
```

```python
del zahlenWoerterbuch
del zahlenWoerterbuch2
del zahlenWoerterbuch3
print(zahlenWoerterbuch)
print(zahlenWoerterbuch2)
print(zahlenWoerterbuch3)
```
