# Kapitel 18.4 — SQL Erweitert

## Lernziele
- JOIN-Operationen verstehen.
- Aggregate Funktionen verwenden.
- GROUP BY und ORDER BY anwenden.

## Theorie

Erweiterte SQL-Funktionen helfen, komplexere Abfragen durchzuführen und Daten aus mehreren Tabellen zu kombinieren.

## Erklärung

**JOIN-Typen:**
- **INNER JOIN**: Nur übereinstimmende Zeilen
- **LEFT JOIN**: Alle Zeilen aus der linken Tabelle
- **RIGHT JOIN**: Alle Zeilen aus der rechten Tabelle
- **FULL JOIN**: Alle Zeilen aus beiden Tabellen

## Beispiel 1
INNER JOIN zum Kombinieren von Daten.

```sql
SELECT schüler.name, kurse.kursname
FROM schüler
INNER JOIN kurse ON schüler.kurs_id = kurse.kurs_id;
```

## Beispiel 2
Aggregat-Funktionen.

```sql
SELECT COUNT(*) FROM schüler;              -- Anzahl
SELECT SUM(alter) FROM schüler;            -- Summe
SELECT AVG(alter) FROM schüler;            -- Durchschnitt
SELECT MAX(alter) FROM schüler;            -- Maximum
SELECT MIN(alter) FROM schüler;            -- Minimum
```

## Beispiel 3
GROUP BY für Gruppierung.

```sql
SELECT klasse, COUNT(*) as anzahl
FROM schüler
GROUP BY klasse;
```

## Praxisbeispiel
Komplexe Abfrage mit JOIN, GROUP BY und ORDER BY.

```sql
SELECT kurse.kursname, COUNT(schüler.id) as schüleranzahl
FROM kurse
LEFT JOIN schüler ON kurse.kurs_id = schüler.kurs_id
GROUP BY kurse.kursname
ORDER BY schüleranzahl DESC;
```

## Häufige Fehler
- Falsche JOIN-Syntax
- GROUP BY ohne alle nicht-aggregierten Spalten
- Komplexe Abfragen ohne Testdaten zu schreiben

## Best Practice
Teste Abfragen mit kleinen Datenmengen zuerst.

## Zusammenfassung
Erweiterte SQL-Funktionen ermöglichen powerful Datenanalysen.
