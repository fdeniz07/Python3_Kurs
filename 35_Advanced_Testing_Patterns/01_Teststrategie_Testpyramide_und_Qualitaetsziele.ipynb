## Haeufige Fehler und Debugging-Checkliste

- Sind Annahmen ueber Umgebungswerte oder Testdaten explizit validiert?
- Gibt es reproduzierbare Schritte fuer den Fehlerfall?
- Wurden Logs, Metriken und relevante Konfiguration gemeinsam betrachtet?
- Ist die Fehlerbehebung durch einen neuen Test abgesichert?
- Existiert eine kurze Dokumentation fuer Lessons Learned?

## Pruefungsfragen und Kurzloesungen

1. Warum reicht hohe Test-Coverage allein nicht fuer Produktionsreife?
Kurzloesung: Coverage misst Umfang, nicht automatisch Risikoabdeckung, Datenqualitaet oder Betriebsfaehigkeit.
2. Welche Rolle spielt Reproduzierbarkeit in Debugging und Testing?
Kurzloesung: Nur reproduzierbare Fehler lassen sich verifizieren, dauerhaft beheben und regressionssicher testen.
3. Was ist ein guter erster Schritt bei sporadischen Fehlern?
Kurzloesung: Beobachtbarkeit verbessern (Logs/Metriken/Tracing) und den Fehler systematisch eingrenzen.
