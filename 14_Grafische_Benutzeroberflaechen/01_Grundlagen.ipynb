# Was ist Tkinter?

Tkinter ist die Standardbibliothek von Python zur Erstellung grafischer Benutzeroberflächen (GUI).

## Vorteile

- Bereits in Python enthalten
- Einfach zu lernen
- Plattformunabhängig
- Viele Widgets verfügbar

## Nachteile

- Weniger modern als PyQt oder Kivy
- Für sehr große Projekte eingeschränkt


# GUI vs CLI

| CLI | GUI |
|------|-----|
| Tastatureingabe | Fenster und Buttons |
| Schnell | Benutzerfreundlich |
| Terminal | Grafische Oberfläche |


````python
import tkinter as tk

fenster = tk.Tk()
fenster.mainloop()
````

## Erklärung

`Tk()` erzeugt das Hauptfenster.

`mainloop()` startet die Ereignisschleife. Ohne `mainloop()` erscheint kein Fenster.


## title()

Ändert den Fenstertitel.

````python
import tkinter as tk

fenster=tk.Tk()
fenster.title('Meine Anwendung')
fenster.mainloop()
````

## geometry()

Legt die Fenstergröße fest.

````python
import tkinter as tk

fenster=tk.Tk()
fenster.geometry('500x300')
fenster.mainloop()
````

## resizable()

Verhindert das Vergrößern.

````python
import tkinter as tk

fenster=tk.Tk()
fenster.resizable(False, False)
fenster.mainloop()
````

## minsize()

Minimale Fenstergröße.

````python
import tkinter as tk

fenster=tk.Tk()
fenster.minsize(300,200)
fenster.mainloop()
````

## maxsize()

Maximale Fenstergröße.

````python
import tkinter as tk

fenster=tk.Tk()
fenster.maxsize(900,600)
fenster.mainloop()
````

## configure()

Hintergrundfarbe ändern.

````python
import tkinter as tk

fenster=tk.Tk()
fenster.configure(bg='lightblue')
fenster.mainloop()
````

# Farben

Tkinter unterstützt Farbnamen und Hex-Farben.

Beispiele:

- red
- blue
- green
- yellow
- white
- black
- #00FF00
- #FFAA00


````python
import tkinter as tk

fenster=tk.Tk()
fenster.configure(bg="lightgreen")
fenster.mainloop()
````

# Schriftarten

Eine Schrift wird als Tupel angegeben.

```python
("Arial",16)
("Arial",16,"bold")
("Arial",16,"italic")
```


````python
import tkinter as tk

fenster=tk.Tk()

label=tk.Label(
    fenster,
    text="Hallo Tkinter",
    font=("Arial",18,"bold")
)

label.pack()

fenster.mainloop()
````

# Erstes Widget: Label

Ein Label dient zur Anzeige von Texten oder Bildern.

Wichtige Parameter

- text
- font
- fg
- bg
- width
- height


````python
import tkinter as tk

fenster=tk.Tk()

label=tk.Label(
    fenster,
    text="Willkommen bei Tkinter!",
    font=("Arial",16),
    fg="blue"
)

label.pack()

fenster.mainloop()
````

# Erstes Widget: Button

Ein Button führt beim Anklicken eine Funktion aus.


````python
import tkinter as tk

def hallo():
    print("Hallo Welt")

fenster=tk.Tk()

button=tk.Button(
    fenster,
    text="Klick mich",
    command=hallo
)

button.pack()

fenster.mainloop()
````

````python
import tkinter as tk

fenster=tk.Tk()
fenster.title("Meine erste GUI")
fenster.configure(bg="blue")
fenster.geometry("400x200")

label=tk.Label(
    master=fenster,
    text="Willkommen bei Tkinter!",
    justify="center",
    font=("Arial",20,"bold"),
    fg="red"
)

button=tk.Button(
    fenster,
    text="Close",
    command=fenster.destroy
)

label.pack()
button.pack()


fenster.mainloop()
````
