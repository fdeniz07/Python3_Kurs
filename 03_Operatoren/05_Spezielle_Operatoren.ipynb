## Was sind Operatoren?

##### Arithmetische Operatoren: Addition, Subtraktion, Multiplikation, Ganzzahldivision, Gleitkommadivision, Modulo, Potenzierung
##### Vergleichsoperatoren: ==, !=, >, <, >=, <=
##### Logische Operatoren: AND, OR, NOT
##### Bitweise Operatoren: &, |, ^, ~, <<, >>
##### Extra: in, not in, is, is not

### Spezielle Operatoren

##### in-Operator
Wenn der erste Ausdruck im zweiten Ausdruck enthalten ist, wird True zurückgegeben, andernfalls False.

```python
a = "merhaba"
b = "ba"
c = "c"

print(b in a) # Der erste Ausdruck ist im zweiten enthalten
print(c in a) # Der erste Ausdruck ist nicht im zweiten enthalten
```

```python
a = [1, 2, 3, 4, 5]
print(5 in a) # True, da die Zahl 5 in der Liste ist
print('a' in a) # False, da das Zeichen 'a' nicht in der Liste ist
```

##### not in-Operator
Wenn der erste Ausdruck im zweiten Ausdruck enthalten ist, wird False zurückgegeben, andernfalls True.

```python
a = "merhaba"
b = "ba"
c = "c"

print(b not in a) # Der erste Ausdruck ist im zweiten enthalten
print(c not in a) # Der erste Ausdruck ist nicht im zweiten enthalten
```

```python
a = [1, 2, 3, 4, 5]
print('a' not in a) # True, da das Zeichen 'a' nicht in der Liste ist
print(1 not in a) # False, da die Zahl 1 in der Liste ist
```

##### is-Operator
Wenn sich beide Variablen im selben Speicherbereich befinden, wird True zurückgegeben, andernfalls False.

```python
x = 5
y = 10
print(id(x))
print(id(y))
print(x is y)
```

```python
x = 5
y = 5
print(id(x))
print(id(y))
print(x is y)
```

##### is not-Operator
Wenn sich beide Variablen im selben Speicherbereich befinden, wird False zurückgegeben, andernfalls True.

```python
x = 5
y = 10
print(id(x))
print(id(y))
print(x is not y)
```

```python
x = 5
y = 5
print(id(x))
print(id(y))
print(x is not y)
```
