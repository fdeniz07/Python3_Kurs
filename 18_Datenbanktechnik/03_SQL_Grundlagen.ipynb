# Kapitel 18.3 — SQL Grundlagen

## Lernziele
- SQL verstehen und anwenden.
- SELECT, INSERT, UPDATE, DELETE beherrschen.
- WHERE-Klauseln verwenden.

## Theorie

SQL (Structured Query Language) ist eine Sprache zur Abfrage und Manipulation von Datenbanken. Die vier Grundoperationen sind:
- **CREATE**: Tabellen erstellen
- **READ (SELECT)**: Daten abrufen
- **UPDATE**: Daten ändern
- **DELETE**: Daten löschen

## Erklärung

**CREATE TABLE:**
```sql
CREATE TABLE schüler (
  id INTEGER PRIMARY KEY,
  name TEXT NOT NULL,
  klasse TEXT
);
```

**INSERT:**
```sql
INSERT INTO schüler (id, name, klasse) VALUES (1, 'Anna', '10A');
```

## Beispiel 1
Daten abrufen mit SELECT.

```sql
SELECT * FROM schüler;
SELECT name, klasse FROM schüler;
SELECT * FROM schüler WHERE klasse = '10A';
```

## Beispiel 2
Daten ändern und löschen.

```sql
UPDATE schüler SET klasse = '10B' WHERE name = 'Anna';
DELETE FROM schüler WHERE id = 1;
```

## WHERE-Klauseln

```sql
WHERE klasse = '10A'           -- Gleichheit
WHERE id > 5                   -- Größer als
WHERE name LIKE 'A%'           -- Muster-Matching
WHERE id IN (1, 2, 3)          -- Wert in Liste
WHERE klasse = '10A' AND id > 1 -- UND-Bedingung
```

## Praxisbeispiel
Komplexere SELECT-Abfrage mit mehreren Bedingungen.

```sql
SELECT name FROM schüler
WHERE klasse = '10A' AND name LIKE 'A%'
ORDER BY name ASC;
```

## Häufige Fehler
- Vergessen von WHERE führt zu unbeabsichtigten Änderungen/Löschungen aller Zeilen
- Falsche SQL-Syntax
- Nicht in korrekter Reihenfolge arbeiten (CREATE vor INSERT)

## Best Practice
Schreibe SQL korrekt und verwende immer WHERE-Klauseln bei UPDATE und DELETE.

## Zusammenfassung
SQL ist die Standardsprache für relationale Datenbanken.
