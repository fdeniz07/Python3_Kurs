# Merke- Ein Test prueft eine fachliche Aussage.- Fehlermeldungen sollen Ursache schnell sichtbar machen.

# ParameterTypische Testparameter: Eingabe, erwartetes Ergebnis, Fehlertyp, Randfaelle.

# RueckgabewertTests liefern meist keinen Fachwert, sondern Bestehen/Fehlschlag mit Diagnose.

````python
# Beispiel 1: Abhaengigkeit simulierendef hole_preis_von_service(produkt_id):    # Normalerweise externer Aufruf    return 9.99def berechne_gesamtpreis(produkt_id, menge, preis_funktion=hole_preis_von_service):    preis = preis_funktion(produkt_id)    return round(preis * menge, 2)print(berechne_gesamtpreis(1, 3))
````

````python
# Beispiel 2: Fake-Funktion als Test Doubledef fake_preis(_produkt_id):    return 5.0assert berechne_gesamtpreis(42, 4, preis_funktion=fake_preis) == 20.0print('Mocking-Idee ueber Dependency Injection OK')
````

````python
# Beispiel 3: Fehler simulierendef failing_preis(_produkt_id):    raise RuntimeError('Service nicht erreichbar')try:    berechne_gesamtpreis(1, 2, preis_funktion=failing_preis)except RuntimeError as fehler:    print('Fehler korrekt simuliert:', fehler)
````

# PraxisbeispielWende die Teststrategie auf ein eigenes Modul an und dokumentiere Testfaelle fuer Happy Path und Edge Cases.

# Haeufige Fehler1. Zu grosse Tests mit mehreren Verantwortlichkeiten.2. Abhaengigkeit von Zufall/Uhrzeit/Netzwerk ohne Kontrolle.3. Nur Happy Path testen.

# Best Practice- Testdaten klar benennen.- Pro Test eine klare Erwartung.- Fehler reproduzierbar halten.

# TippSchreibe zuerst den Testnamen als fachlichen Satz, danach den Testcode.

# UebungErstelle mindestens 5 Tests: 3 Erfolgsfaelle, 2 Fehlerfaelle, inklusive Randwerten.

````python
# Loesung: Robuste Variantedef berechne_gesamtpreis_robust(produkt_id, menge, preis_funktion=hole_preis_von_service):    try:        preis = preis_funktion(produkt_id)        return {'ok': True, 'gesamt': round(preis * menge, 2)}    except Exception as fehler:        return {'ok': False, 'fehler': str(fehler)}print(berechne_gesamtpreis_robust(1, 2, fake_preis))print(berechne_gesamtpreis_robust(1, 2, failing_preis))
````
