# Dateioperationen mit 'with' Statement (Context Manager)

### Warum benutzen wir 'with'?
In Python ist das `with`-Statement der sicherste Weg, um Dateien zu öffnen. Man nennt es einen *Context Manager*. Wenn wir Dateien manuell mit `open()` öffnen und `close()` schließen, besteht die Gefahr, dass die Datei bei einem Fehler im Programm nicht geschlossen wird, was zu Datenverlust führen kann.

Der `with`-Block sorgt dafür, dass die Datei **automatisch geschlossen wird**, sobald der Code-Block verlassen wird – selbst wenn während der Verarbeitung ein Fehler (Exception) auftritt.

### Der Vorteil des 'with'-Statements
Vergleich der beiden Methoden:

**1. Ohne 'with' (Risikoreich):**
```python
f = open("datei.txt", "r")
inhalt = f.read()
f.close()  # Wenn hier ein Fehler passiert, wird f.close() nie erreicht!
```

**2. Mit 'with' (Sicher):**
```python
with open("datei.txt", "r") as f:
    inhalt = f.read()
# Die Datei wurde hier automatisch und sicher geschlossen.
```

### Beispiel: Sicheres Lesen und Schreiben
Hier zeigen wir, wie wir mit `with` Daten effizient in eine Datei schreiben und wieder auslesen.

````python
# Schreiben in eine Datei mit 'with'
with open("logbuch_neu.txt", "w", encoding="utf-8") as f:
    f.write("Eintrag 1: Systemstart\n")
    f.write("Eintrag 2: Datensicherung erfolgreich\n")

# Lesen aus einer Datei mit 'with'
with open("logbuch_neu.txt", "r", encoding="utf-8") as f:
    zeilen = f.readlines()
    for zeile in zeilen:
        print(zeile.strip())
````

### Was ist ein Kontextmanager?
Das `with`-Statement funktioniert nicht nur mit Dateien, sondern mit allen Objekten, die zwei besondere Methoden besitzen: `__enter__()` und `__exit__()`.

* **`__enter__()`**: Diese Methode wird aufgerufen, wenn der `with`-Block startet. Bei Dateien öffnet sie die Datei.
* **`__exit__()`**: Diese Methode wird aufgerufen, sobald der `with`-Block verlassen wird. Sie stellt sicher, dass die Datei ordnungsgemäß geschlossen wird, auch bei Fehlern.

Objekte, die diese Methoden implementieren, nennt man **Kontextmanager**.

### Beispiel: Sicherer Dateizugriff
Hier ist ein einfaches Beispiel für die Verwendung des `with`-Statements:

````python
# Sicherer Schreib- und Lesevorgang
with open("beispiel.txt", "w", encoding="utf-8") as f:
    f.write("Diese Datei wird automatisch geschlossen.")

with open("beispiel.txt", "r", encoding="utf-8") as f:
    inhalt = f.read()
    print(inhalt)
````
