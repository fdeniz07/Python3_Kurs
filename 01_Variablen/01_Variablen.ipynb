## Warum benötigen wir Variablen?

Für jede Information, die wir in einem Programm verwenden, wird ein Speicherbereich reserviert. Alle Operationen werden über diese Speicherbereiche ausgeführt. Die Arbeit mit Speicheradressen ist jedoch kompliziert und unübersichtlich. Deshalb verwenden wir Variablen, um auf Daten im Speicher zuzugreifen. Mithilfe von Variablennamen können wir diese Speicherbereiche einfach ansprechen und verwalten.

```python
x=10
y=10
print(x)
print(id(x))
print(y)
print(id(y))
```

## Wie definiert man Variablen und weist ihnen Werte zu?

### Werte einzeln zuweisen

```python
x=123
print(x)
```

```python
y='a'
print(y)
```

```python
z="Max"
print(z)
```

```python
w=True
print(w)
```

### Mehrere Werte gleichzeitig zuweisen

```python
a,b,c=1,2,3
print(a,b,c)
```

### Allen Variablen denselben Wert zuweisen

```python
a=b=c=0
print(a,b,c)
```

### Werte zwischen Variablen übertragen

```python
x,y=1,2
print(x,y)
x,y=y,x
print(x,y)
```
