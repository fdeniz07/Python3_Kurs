## 1. Theoretische Grundlage: Daten aus dem Internet
Um auf Dateien im World Wide Web zuzugreifen, nutzen wir `urllib.request`. Ein **URL** identifiziert den Speicherort. 
- **`urlopen(url)`**: Öffnet einen Stream zu einer Internet-Ressource (liefert ein `HTTPResponse`-Objekt).
- **`stream.read()`**: Liest den Inhalt als *Bytestring* aus.
- **`decode()`**: Wandelt den Bytestring in einen lesbaren String um.

### Die Rolle des Kontextmanagers
Auch hier nutzen wir das `with`-Statement. Es garantiert, dass der Internet-Stream nach dem Lesen der Daten (`__exit__`) korrekt geschlossen wird, was Ressourcen spart.

```python
from urllib.request import urlopen

def rezept_aus_internet_laden(url):
    """Lädt Rezeptdaten sicher via Kontextmanager."""
    with urlopen(url) as stream:
        daten = stream.read()
        return daten.decode('utf-8')

# Beispielaufruf (Platzhalter-URL)
# rezept = rezept_aus_internet_laden("http://example.com/rezept.txt")
# print(rezept)
url = 'http://www.textfiles.com/food/richbred.txt'
rezept_inhalt = rezept_aus_internet_laden(url)
print(rezept_inhalt)


```
