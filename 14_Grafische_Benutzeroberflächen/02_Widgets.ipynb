# Widget: Label

## Beschreibung

Zeigt Texte oder Bilder an

## Wichtige Eigenschaften

- `text`
- `font`
- `fg`
- `bg`
- `width`
- `height`

## Tipp

Probiere das Beispiel aus und ändere anschließend einige Eigenschaften selbst.


```python
import tkinter as tk

root=tk.Tk()

label=tk.Label(root,text="Hallo Welt",font=("Arial",16))
label.pack()

root.mainloop()
```

### Kleine Übung

1. Erstelle selbst ein Label.
2. Ändere mindestens zwei Eigenschaften.
3. Starte das Programm erneut und beobachte die Änderungen.


# Widget: Button

## Beschreibung

Führt eine Funktion aus

## Wichtige Eigenschaften

- `text`
- `command`
- `state`
- `width`

## Tipp

Probiere das Beispiel aus und ändere anschließend einige Eigenschaften selbst.


```python
import tkinter as tk

def hallo():
    print("Hallo")

root=tk.Tk()
root.title("Hallo Welt")
root.geometry("300x200")


button=tk.Button(root,text="Klick",command=hallo)
button.pack()

button=tk.Button(root,text="Schließen",command=root.destroy)
button.pack()

root.mainloop()
```

### Kleine Übung

1. Erstelle selbst ein Button.
2. Ändere mindestens zwei Eigenschaften.
3. Starte das Programm erneut und beobachte die Änderungen.


# Widget: Entry

## Beschreibung

## Wichtige Eigenschaften

- `get()`
- `insert()`
- `delete()`

## Tipp

Probiere das Beispiel aus und ändere anschließend einige Eigenschaften selbst.

# Widget: Checkbutton

## Beschreibung

## Wichtige Eigenschaften

- `variable`
- `onvalue`

Probiere das Beispiel aus und ändere anschließend einige Eigenschaften selbst.

# Widget: Radiobutton

## Beschreibung

## Wichtige Eigenschaften

- `variable`

## Tipp

Probiere das Beispiel aus und ändere anschließend einige Eigenschaften selbst.
